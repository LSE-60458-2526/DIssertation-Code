```python
# %%
# ============================================================
# WSJ AV News Data Cleaning, Scope Screening,
# Abstract-Hash Deduplication, and Abstract Extraction
#
# Moderately relaxed screening version:
#
# 1. Load and inspect the original LexisNexis dataset
# 2. Standardise publication dates
# 3. Extract ABSTRACT to End of Document
# 4. Remove unusable records
# 5. Apply automated AV scope screening
# 6. Retain ambiguous ADAS/company-context cases for review
# 7. Apply completed manual scope-review decisions
# 8. Deduplicate using normalised abstract SHA-256 hashes
# 9. Produce audit files and descriptive statistics
# 10. Save the final analytical dataset
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import hashlib
import unicodedata


# %%
# ============================================================
# 1. File paths
# ============================================================

INPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/excel/"
    "lexis_structured_WSJ_1-389.xlsx"
)

OUTPUT_DIR = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_WSJ"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

FINAL_DATA_PATH = (
    OUTPUT_DIR /
    "WSJ_AV_final_analytical_dataset.xlsx"
)

MANUAL_REVIEW_COMPLETED_PATH = (
    OUTPUT_DIR /
    "WSJ_manual_scope_review_completed.xlsx"
)


# %%
# ============================================================
# 2. Load data
# ============================================================

df = pd.read_excel(INPUT_PATH)
n_original = len(df)

print("=" * 75)
print("WSJ AV DATA CLEANING")
print("=" * 75)

print("\nData loaded successfully.")
print("Original shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# %%
# ============================================================
# 3. Define key columns
# ============================================================

DATE_COL = "date"
TITLE_COL = "title"
CONTENT_COL = "content"
SOURCE_COL = "source"

required_columns = [DATE_COL, TITLE_COL, CONTENT_COL]

missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_required_columns:
    raise KeyError(
        "The following required columns are missing: "
        f"{missing_required_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

if SOURCE_COL not in df.columns:
    print(
        f"\nWARNING: '{SOURCE_COL}' column was not found. "
        "A default source value will be created."
    )
    df[SOURCE_COL] = "Wall Street Journal Abstracts"

df["_original_order"] = range(len(df))

df[TITLE_COL] = (
    df[TITLE_COL]
    .fillna("")
    .astype(str)
)

df[CONTENT_COL] = (
    df[CONTENT_COL]
    .fillna("")
    .astype(str)
)


# %%
# ============================================================
# 4. Missing-value and column summary
# ============================================================

column_summary = pd.DataFrame({
    "column": df.columns,
    "non_null_count": df.notna().sum().values,
    "missing_count": df.isna().sum().values,
    "missing_percent": (
        df.isna().mean().values * 100
    ).round(2),
    "dtype": df.dtypes.astype(str).values
})

display(column_summary)

column_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_column_missing_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 5. Standardise publication dates
#
# Example:
# December 14, 2023 Thursday
# -> 2023-12-14
# ============================================================

MONTH_PATTERN = (
    r"(?:January|February|March|April|May|June|"
    r"July|August|September|October|November|December)"
)

DATE_PATTERN = (
    rf"{MONTH_PATTERN}\s+"
    r"\d{1,2},\s+\d{4}"
)


def extract_english_date(value):
    """
    Extract an English-language date from a messy
    LexisNexis-style date string.
    """

    if pd.isna(value):
        return pd.NaT

    text = str(value).strip()

    match = re.search(
        DATE_PATTERN,
        text,
        flags=re.IGNORECASE
    )

    if match is None:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        format="%B %d, %Y",
        errors="coerce"
    )


df["date_original"] = df[DATE_COL]

df["date_parsed_direct"] = pd.to_datetime(
    df[DATE_COL],
    errors="coerce"
)

df["date_fixed"] = (
    df[DATE_COL]
    .apply(extract_english_date)
)

date_summary = pd.DataFrame([
    {
        "metric": "Total original articles",
        "value": len(df)
    },
    {
        "metric": "Dates parsed directly by pandas",
        "value": int(
            df["date_parsed_direct"].notna().sum()
        )
    },
    {
        "metric": "Dates recovered using date extraction",
        "value": int(
            df["date_fixed"].notna().sum()
        )
    },
    {
        "metric": "Dates still missing after extraction",
        "value": int(
            df["date_fixed"].isna().sum()
        )
    }
])

print("\nDate parsing summary:")
display(date_summary)

date_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_date_parsing_summary.xlsx",
    index=False
)

failed_date_rows = df[
    df["date_fixed"].isna()
].copy()

failed_date_rows.to_excel(
    OUTPUT_DIR /
    "WSJ_invalid_date_rows_check.xlsx",
    index=False
)


# %%
# ============================================================
# 6. Normalise line breaks
# ============================================================

def normalize_line_breaks(text):
    """
    Standardise line breaks while preserving text structure.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    if "\\n" in text and "\n" not in text:
        text = text.replace("\\n", "\n")

    text = re.sub(r"[ \t]+\n", "\n", text)
    text = re.sub(r"\n[ \t]+", "\n", text)
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


# %%
# ============================================================
# 7. Extract WSJ abstract
#
# Extract text between:
# ABSTRACT
# ...
# End of Document
# ============================================================

def extract_abstract(text):
    """
    Extract the WSJ abstract from a LexisNexis record.

    Preferred format:
    ABSTRACT
    abstract text
    End of Document

    A fallback pattern handles unusual line formatting.
    """

    text = normalize_line_breaks(text)

    if text == "":
        return ""

    main_pattern = re.compile(
        r"(?ims)"
        r"^[ \t]*ABSTRACT[ \t]*:?[ \t]*\n"
        r"(.*?)"
        r"(?=^[ \t]*End\s+of\s+Document[ \t]*$)"
    )

    match = main_pattern.search(text)

    if match is None:
        fallback_pattern = re.compile(
            r"(?is)"
            r"(?:^|\n|\s)ABSTRACT\s*:?\s*"
            r"(.*?)"
            r"\s*End\s+of\s+Document"
        )
        match = fallback_pattern.search(text)

    if match is None:
        return ""

    abstract = match.group(1).strip()

    abstract = re.sub(r"[\r\n\t]+", " ", abstract)
    abstract = re.sub(r"\s+", " ", abstract)
    abstract = abstract.strip(" \n\r\t:-")

    return abstract.strip()


df["abstract"] = (
    df[CONTENT_COL]
    .apply(extract_abstract)
)

df["abstract_extracted"] = (
    df["abstract"]
    .fillna("")
    .str
    .strip()
    .ne("")
)

print("\nAbstract extraction status:")
print(
    df["abstract_extracted"]
    .value_counts(dropna=False)
)

failed_abstract_rows = df[
    ~df["abstract_extracted"]
].copy()

failed_abstract_rows.to_excel(
    OUTPUT_DIR /
    "WSJ_failed_abstract_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 8. Basic text measures
# ============================================================

def count_english_words(text):
    """
    Approximate English word count.

    Expressions such as self-driving, company's, COVID-19
    and U.S. are generally treated as single word units.
    """

    if pd.isna(text):
        return 0

    text = str(text).strip()

    if text == "":
        return 0

    words = re.findall(
        r"\b[A-Za-z0-9]+"
        r"(?:['’.-][A-Za-z0-9]+)*\b",
        text
    )

    return len(words)


df["abstract_char_length"] = (
    df["abstract"]
    .fillna("")
    .str
    .len()
)

df["abstract_word_count"] = (
    df["abstract"]
    .apply(count_english_words)
)


# %%
# ============================================================
# 9. Initial usability exclusions
#
# Exclude:
# - missing or unrecoverable publication dates
# - failed or empty abstract extraction
# ============================================================

df["exclude_invalid_date"] = (
    df["date_fixed"].isna()
)

df["exclude_missing_abstract"] = (
    ~df["abstract_extracted"]
)

df["exclude_initial_unusable_record"] = (
    df["exclude_invalid_date"]
    | df["exclude_missing_abstract"]
)


def create_initial_exclusion_reason(row):

    reasons = []

    if row["exclude_invalid_date"]:
        reasons.append(
            "Missing or invalid publication date"
        )

    if row["exclude_missing_abstract"]:
        reasons.append(
            "Abstract could not be extracted"
        )

    return "; ".join(reasons)


df["initial_exclusion_reason"] = (
    df.apply(
        create_initial_exclusion_reason,
        axis=1
    )
)

initial_exclusion_summary = pd.DataFrame([
    {
        "rule": "Missing or invalid publication date",
        "article_count": int(
            df["exclude_invalid_date"].sum()
        )
    },
    {
        "rule": "Abstract could not be extracted",
        "article_count": int(
            df["exclude_missing_abstract"].sum()
        )
    },
    {
        "rule": "Excluded by one or more initial rules",
        "article_count": int(
            df[
                "exclude_initial_unusable_record"
            ].sum()
        )
    }
])

initial_exclusion_summary["percent_of_original"] = (
    initial_exclusion_summary["article_count"]
    / len(df)
    * 100
).round(2)

display(initial_exclusion_summary)

initial_exclusion_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_initial_exclusion_summary.xlsx",
    index=False
)

initially_excluded = df[
    df["exclude_initial_unusable_record"]
].copy()

initially_excluded.to_excel(
    OUTPUT_DIR /
    "WSJ_initially_excluded_records_check.xlsx",
    index=False
)

usable_data = df[
    ~df["exclude_initial_unusable_record"]
].copy()

n_after_initial_cleaning = len(usable_data)

n_removed_initial_cleaning = (
    len(df) - len(usable_data)
)

very_long_abstracts = usable_data[
    usable_data["abstract_word_count"] > 250
].copy()

very_long_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_over_250_words_check.xlsx",
    index=False
)

print(
    "\nAbstracts longer than 250 words, retained for audit:",
    len(very_long_abstracts)
)


# %%
# ============================================================
# 10. Create automated AV-screening text
#
# For WSJ, use title plus the complete extracted abstract.
# ============================================================

df_scope = usable_data.copy()

df_scope["automated_screen_text"] = (
    df_scope[TITLE_COL]
    + " "
    + df_scope["abstract"].fillna("")
)


def contains_any_pattern(text, patterns):
    """
    Return True if at least one regex pattern appears in text.
    """

    if pd.isna(text):
        return False

    text = str(text)

    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        is not None
        for pattern in patterns
    )


# %%
# ============================================================
# 11. Define English AV-scope patterns
# ============================================================

general_road_av_patterns = [
    r"\bautonomous\s+vehicles?\b",
    r"\bautonomous\s+cars?\b",
    r"\bautomated\s+vehicles?\b",
    r"\bautomated\s+cars?\b",
    r"\bautonomous\s+driving\b",
    r"\bautomated\s+driving\b",
    r"\bautomated\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+technology\b",
    r"\bself[-\s]?driving\s+technology\b",
    r"\bself[-\s]?driving\b",
    r"\bself[-\s]?driving\s+vehicles?\b",
    r"\bself[-\s]?driving\s+cars?\b",
    r"\bdriverless\b",
    r"\bdriverless\s+vehicles?\b",
    r"\bdriverless\s+cars?\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bself[-\s]?driving\s+taxis?\b",
    r"\bvehicle\s+automation\b",
    r"\bautomotive\s+automation\b",
    r"\bautomated\s+mobility\b",
    r"\bautonomous\s+mobility\b",
    r"\bfull\s+self[-\s]?driving\b",
]

higher_automation_patterns = [
    r"\bSAE\s+(?:Level\s*)?[345]\b",
    r"\bLevel[-\s]?[345]\b",
    r"\bL[345]\b",
    r"\bfully\s+autonomous\b",
    r"\bfull\s+autonomy\b",
    r"\bhighly\s+automated\b",
    r"\bhigh[-\s]?automation\b",
    r"\bdriverless\b",
    r"\bwithout\s+(?:a\s+)?safety\s+driver\b",
    r"\bno\s+safety\s+driver\b",
    r"\bremove(?:d|s|ing)?\s+(?:the\s+)?safety\s+driver\b",
    r"\bcommercial\s+autonomous\s+service\b",
    r"\bcommercial\s+driverless\s+service\b",
    r"\bcommercial\s+robotaxi\s+service\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+ride[-\s]?hailing\b",
]

passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bpersonal\s+vehicles?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\bminivans?\b",
    r"\btaxis?\b",
    r"\bcabs?\b",
    r"\bride[-\s]?hailing\b",
    r"\bride[-\s]?sharing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]

strong_passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\btaxis?\b",
    r"\bride[-\s]?hailing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]

bus_public_transport_patterns = [
    r"\bautonomous\s+buses?\b",
    r"\bdriverless\s+buses?\b",
    r"\bself[-\s]?driving\s+buses?\b",
    r"\bautomated\s+buses?\b",
    r"\bautonomous\s+shuttles?\b",
    r"\bdriverless\s+shuttles?\b",
    r"\bself[-\s]?driving\s+shuttles?\b",
    r"\bautomated\s+shuttles?\b",
    r"\bautonomous\s+minibuses?\b",
    r"\bdriverless\s+minibuses?\b",
    r"\bpublic\s+transit\b",
    r"\bpublic\s+transportation\b",
    r"\btransit\s+service\b",
    r"\bpeople\s+movers?\b",
]

adas_patterns = [
    r"\bADAS\b",
    r"\badvanced\s+driver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assist(?:ance)?\b",
    r"\bautomated\s+parking\b",
    r"\bautomatic\s+parking\b",
    r"\bself[-\s]?parking\b",
    r"\badaptive\s+cruise\s+control\b",
    r"\blane[-\s]?keeping\b",
    r"\blane[-\s]?centering\b",
    r"\blane[-\s]?departure\s+warning\b",
    r"\bautomatic\s+emergency\s+braking\b",
    r"\bblind[-\s]?spot\s+monitoring\b",
    r"\bcollision\s+warning\b",
    r"\bparking\s+assist\b",
    r"\bSuper\s+Cruise\b",
    r"\bBlueCruise\b",
    r"\bAutopilot\b",
]

road_vehicle_context_patterns = [
    r"\bvehicles?\b",
    r"\bcars?\b",
    r"\bautomobiles?\b",
    r"\bpassenger\s+vehicles?\b",
    r"\btaxis?\b",
    r"\broads?\b",
    r"\bhighways?\b",
    r"\bstreets?\b",
    r"\bautomakers?\b",
    r"\bcar\s+companies\b",
    r"\broad\s+testing\b",
    r"\bpublic\s+roads?\b",
    r"\btest\s+vehicles?\b",
]

strong_non_road_patterns = [
    r"\bautonomous\s+drones?\b",
    r"\bdriverless\s+drones?\b",
    r"\bunmanned\s+aerial\s+vehicles?\b",
    r"\bUAVs?\b",
    r"\bautonomous\s+aircraft\b",
    r"\bpilotless\s+aircraft\b",
    r"\bautonomous\s+ships?\b",
    r"\bunmanned\s+ships?\b",
    r"\bdriverless\s+ships?\b",
    r"\bautonomous\s+boats?\b",
    r"\bunmanned\s+boats?\b",
    r"\bautonomous\s+vessels?\b",
    r"\bdriverless\s+trains?\b",
    r"\bautonomous\s+trains?\b",
    r"\bautomated\s+trains?\b",
    r"\bdriverless\s+metros?\b",
    r"\bautomated\s+metros?\b",
    r"\bautonomous\s+mining\s+trucks?\b",
    r"\bdriverless\s+mining\s+trucks?\b",
    r"\bautonomous\s+mine\s+vehicles?\b",
    r"\bdriverless\s+mine\s+vehicles?\b",
    r"\bautonomous\s+port\s+vehicles?\b",
    r"\bdriverless\s+port\s+vehicles?\b",
    r"\bautomated\s+container\s+vehicles?\b",
    r"\bautonomous\s+tractors?\b",
    r"\bdriverless\s+tractors?\b",
    r"\bautonomous\s+farm\s+equipment\b",
    r"\bunmanned\s+farm\s+equipment\b",
]

general_non_road_context_patterns = [
    r"\bdrones?\b",
    r"\baircraft\b",
    r"\baviation\b",
    r"\bships?\b",
    r"\bboats?\b",
    r"\bvessels?\b",
    r"\brailways?\b",
    r"\btrains?\b",
    r"\bmetros?\b",
    r"\bsubways?\b",
    r"\bmines?\b",
    r"\bmining\b",
    r"\bports?\b",
    r"\bharbors?\b",
    r"\btractors?\b",
    r"\bagricultural\s+machinery\b",
]

av_company_patterns = [
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
    r"\bMotional\b",
    r"\bArgo\s+AI\b",
    r"\bAurora\b",
    r"\bNuro\b",
    r"\bMay\s+Mobility\b",
    r"\bMobileye\b",
    r"\bTesla\s+Autopilot\b",
    r"\bFull\s+Self[-\s]?Driving\b",
    r"\bTesla\s+FSD\b",
    r"\bGoogle\s+self[-\s]?driving\b",
    r"\bUber\s+autonomous\b",
    r"\bUber\s+self[-\s]?driving\b",
    r"\bGM\s+Cruise\b",
]


# %%
# ============================================================
# 12. Generate automated screening indicators
# ============================================================

pattern_groups = {
    "has_general_road_av_term":
        general_road_av_patterns,
    "has_higher_automation_term":
        higher_automation_patterns,
    "has_passenger_robotaxi_term":
        passenger_robotaxi_patterns,
    "has_strong_passenger_robotaxi_term":
        strong_passenger_robotaxi_patterns,
    "has_bus_public_transport_term":
        bus_public_transport_patterns,
    "has_adas_term":
        adas_patterns,
    "has_road_vehicle_context":
        road_vehicle_context_patterns,
    "has_strong_non_road_term":
        strong_non_road_patterns,
    "has_general_non_road_context":
        general_non_road_context_patterns,
    "has_av_company_term":
        av_company_patterns
}

for output_column, patterns in pattern_groups.items():
    df_scope[output_column] = (
        df_scope["automated_screen_text"]
        .apply(
            lambda text: contains_any_pattern(
                text,
                patterns
            )
        )
    )

df_scope["has_non_road_term"] = (
    df_scope["has_strong_non_road_term"]
    | df_scope["has_general_non_road_context"]
)


# %%
# ============================================================
# 13. Create broader AV-relevance indicators
# ============================================================

df_scope["has_explicit_av_signal"] = (
    df_scope["has_general_road_av_term"]
    | df_scope["has_higher_automation_term"]
    | df_scope["has_passenger_robotaxi_term"]
)

df_scope["has_company_supported_av_signal"] = (
    df_scope["has_av_company_term"]
    & (
        df_scope["has_road_vehicle_context"]
        | df_scope["has_adas_term"]
        | df_scope["has_higher_automation_term"]
    )
)

df_scope["has_broad_road_av_relevance"] = (
    df_scope["has_explicit_av_signal"]
    | df_scope["has_company_supported_av_signal"]
)


# %%
# ============================================================
# 14. Define moderately relaxed exclusion rules
# ============================================================

df_scope["remove_no_relevant_road_av_term"] = (
    ~df_scope["has_broad_road_av_relevance"]
)

df_scope["remove_pure_non_road"] = (
    df_scope["has_strong_non_road_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)

df_scope["remove_bus_public_transport_only"] = (
    df_scope["has_bus_public_transport_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)

df_scope["review_adas_only"] = (
    df_scope["has_adas_term"]
    & ~df_scope["has_higher_automation_term"]
)

df_scope["remove_adas_only"] = False

df_scope["exclude_by_automated_scope_screening"] = (
    df_scope["remove_no_relevant_road_av_term"]
    | df_scope["remove_pure_non_road"]
    | df_scope["remove_bus_public_transport_only"]
)

df_scope["requires_manual_scope_review"] = (
    df_scope["review_adas_only"]
    | (
        ~df_scope["has_explicit_av_signal"]
        & df_scope["has_company_supported_av_signal"]
    )
)


# %%
# ============================================================
# 15. Create exclusion and manual-review reasons
# ============================================================

def create_scope_exclusion_reason(row):

    reasons = []

    if row["remove_no_relevant_road_av_term"]:
        reasons.append(
            "No broad road autonomous-vehicle relevance signal"
        )

    if row["remove_pure_non_road"]:
        reasons.append(
            "Pure non-road autonomous-system article"
        )

    if row["remove_bus_public_transport_only"]:
        reasons.append(
            "Bus, shuttle or public-transport-only article"
        )

    return "; ".join(reasons)


df_scope["automated_scope_exclusion_reason"] = (
    df_scope.apply(
        create_scope_exclusion_reason,
        axis=1
    )
)


def create_manual_review_reason(row):

    reasons = []

    if row["review_adas_only"]:
        reasons.append(
            "ADAS terminology without explicit "
            "higher-level automation indicator"
        )

    if (
        not row["has_explicit_av_signal"]
        and row["has_company_supported_av_signal"]
    ):
        reasons.append(
            "AV relevance inferred from company/programme "
            "and road-vehicle context"
        )

    return "; ".join(reasons)


df_scope["manual_scope_review_reason"] = (
    df_scope.apply(
        create_manual_review_reason,
        axis=1
    )
)


# %%
# ============================================================
# 16. Automated scope-screening summary
# ============================================================

scope_summary = pd.DataFrame([
    {
        "screening_rule":
            "No broad road autonomous-vehicle relevance signal",
        "article_count":
            int(
                df_scope[
                    "remove_no_relevant_road_av_term"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Pure non-road autonomous-system article",
        "article_count":
            int(
                df_scope[
                    "remove_pure_non_road"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Bus, shuttle or public-transport-only article",
        "article_count":
            int(
                df_scope[
                    "remove_bus_public_transport_only"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "ADAS case without explicit higher-level automation",
        "article_count":
            int(
                df_scope[
                    "review_adas_only"
                ].sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Company/context-supported borderline AV case",
        "article_count":
            int(
                (
                    ~df_scope["has_explicit_av_signal"]
                    & df_scope[
                        "has_company_supported_av_signal"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Excluded by one or more automatic scope rules",
        "article_count":
            int(
                df_scope[
                    "exclude_by_automated_scope_screening"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Retained articles requiring manual scope review",
        "article_count":
            int(
                (
                    df_scope[
                        "requires_manual_scope_review"
                    ]
                    & ~df_scope[
                        "exclude_by_automated_scope_screening"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    }
])

scope_summary["percent_of_scope_input"] = (
    scope_summary["article_count"]
    / len(df_scope)
    * 100
).round(2)

print("\nAutomated AV-scope screening summary:")
display(scope_summary)

scope_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_automated_AV_scope_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 17. Export scope-screening audit files
# ============================================================

scope_audit_columns = [
    "date_fixed",
    DATE_COL,
    TITLE_COL,
    "has_general_road_av_term",
    "has_higher_automation_term",
    "has_passenger_robotaxi_term",
    "has_strong_passenger_robotaxi_term",
    "has_bus_public_transport_term",
    "has_adas_term",
    "has_road_vehicle_context",
    "has_strong_non_road_term",
    "has_general_non_road_context",
    "has_non_road_term",
    "has_av_company_term",
    "has_explicit_av_signal",
    "has_company_supported_av_signal",
    "has_broad_road_av_relevance",
    "remove_no_relevant_road_av_term",
    "remove_pure_non_road",
    "remove_bus_public_transport_only",
    "remove_adas_only",
    "review_adas_only",
    "requires_manual_scope_review",
    "automated_scope_exclusion_reason",
    "manual_scope_review_reason",
    "automated_screen_text",
    "abstract"
]

scope_audit_columns = [
    column
    for column in scope_audit_columns
    if column in df_scope.columns
]

automatically_excluded_scope = df_scope[
    df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

automatically_retained_scope = df_scope[
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

automatically_excluded_scope[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_articles_excluded_by_AV_scope.xlsx",
    index=False
)

manual_scope_review_articles[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_articles_requiring_manual_scope_review.xlsx",
    index=False
)

df_scope[
    df_scope["remove_no_relevant_road_av_term"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_no_broad_AV_signal_excluded_check.xlsx",
    index=False
)

df_scope[
    df_scope["remove_pure_non_road"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_non_road_articles_excluded_check.xlsx",
    index=False
)

df_scope[
    df_scope["remove_bus_public_transport_only"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_bus_only_articles_excluded_check.xlsx",
    index=False
)

adas_review_cases = df_scope[
    df_scope["review_adas_only"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

adas_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_ADAS_cases_retained_for_review.xlsx",
    index=False
)

company_context_review_cases = df_scope[
    ~df_scope["has_explicit_av_signal"]
    & df_scope[
        "has_company_supported_av_signal"
    ]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

company_context_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_company_context_AV_cases_retained_for_review.xlsx",
    index=False
)

mixed_bus_retained = df_scope[
    df_scope["has_bus_public_transport_term"]
    & df_scope["has_strong_passenger_robotaxi_term"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_bus_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_mixed_bus_passenger_articles_retained_check.xlsx",
    index=False
)

mixed_adas_retained = df_scope[
    df_scope["has_adas_term"]
    & df_scope["has_higher_automation_term"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_adas_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_mixed_ADAS_higher_automation_retained_check.xlsx",
    index=False
)


# %%
# ============================================================
# 17B. Apply completed manual scope-review decisions
#
# Run the script once to create:
# WSJ_articles_requiring_manual_scope_review.xlsx
#
# Complete the review and save a copy as:
# WSJ_manual_scope_review_completed.xlsx
#
# Required decision values:
# INCLUDE or EXCLUDE
# ============================================================

if not MANUAL_REVIEW_COMPLETED_PATH.exists():

    raise FileNotFoundError(
        "\nThe completed manual-review file could not be found:\n"
        f"{MANUAL_REVIEW_COMPLETED_PATH}\n\n"
        "First review WSJ_articles_requiring_manual_scope_review.xlsx. "
        "Then save the completed file as "
        "WSJ_manual_scope_review_completed.xlsx inside OUTPUT_DIR."
    )

manual_review_completed = pd.read_excel(
    MANUAL_REVIEW_COMPLETED_PATH,
    sheet_name="Sheet1"
)

print("\nCompleted manual-review file loaded:")
print(MANUAL_REVIEW_COMPLETED_PATH)

print("\nManual-review file shape:")
print(manual_review_completed.shape)

print("\nManual-review file columns:")
print(manual_review_completed.columns.tolist())

required_manual_review_columns = [
    TITLE_COL,
    "abstract",
    "manual_review_decision",
    "manual_review_in_scope",
    "manual_review_reason_final",
    "manual_review_category"
]

missing_manual_review_columns = [
    column
    for column in required_manual_review_columns
    if column not in manual_review_completed.columns
]

if missing_manual_review_columns:
    raise KeyError(
        "The completed manual-review file is missing these columns: "
        f"{missing_manual_review_columns}\n"
        f"Available columns: "
        f"{manual_review_completed.columns.tolist()}"
    )


def normalize_manual_review_match_text(text):
    """
    Normalise title/abstract text for matching the manually
    reviewed file back to the main dataset.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = (
        text
        .replace("\u00a0", " ")
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()


def create_manual_review_match_key(title, abstract):
    """
    Create a SHA-256 matching key from the normalised title and
    abstract.
    """

    normalized_title = (
        normalize_manual_review_match_text(title)
    )

    normalized_abstract = (
        normalize_manual_review_match_text(abstract)
    )

    combined_text = (
        normalized_title
        + "\n"
        + normalized_abstract
    )

    return hashlib.sha256(
        combined_text.encode("utf-8")
    ).hexdigest()


df_scope["_manual_review_match_key"] = (
    df_scope.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["abstract"]
        ),
        axis=1
    )
)

manual_review_completed[
    "_manual_review_match_key"
] = (
    manual_review_completed.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["abstract"]
        ),
        axis=1
    )
)

manual_review_completed[
    "manual_review_decision"
] = (
    manual_review_completed[
        "manual_review_decision"
    ]
    .fillna("")
    .astype(str)
    .str
    .strip()
    .str
    .upper()
)

valid_manual_decisions = {
    "INCLUDE",
    "EXCLUDE"
}

invalid_manual_decision_rows = (
    manual_review_completed[
        ~manual_review_completed[
            "manual_review_decision"
        ].isin(valid_manual_decisions)
    ]
    .copy()
)

if len(invalid_manual_decision_rows) > 0:

    display_columns = [
        column
        for column in [
            "date_fixed",
            TITLE_COL,
            "manual_review_decision"
        ]
        if column
        in invalid_manual_decision_rows.columns
    ]

    print(
        "\nInvalid or missing manual-review decisions:"
    )

    display(
        invalid_manual_decision_rows[
            display_columns
        ]
    )

    raise ValueError(
        "Every reviewed article must have either INCLUDE "
        "or EXCLUDE in manual_review_decision."
    )

decision_conflicts = (
    manual_review_completed
    .groupby(
        "_manual_review_match_key"
    )["manual_review_decision"]
    .nunique()
)

conflicting_keys = set(
    decision_conflicts[
        decision_conflicts > 1
    ].index
)

if conflicting_keys:

    conflicting_rows = (
        manual_review_completed[
            manual_review_completed[
                "_manual_review_match_key"
            ].isin(conflicting_keys)
        ]
        .copy()
    )

    display(
        conflicting_rows[
            [
                TITLE_COL,
                "manual_review_decision",
                "manual_review_reason_final"
            ]
        ]
    )

    raise ValueError(
        "Conflicting manual decisions were found for "
        "the same title/abstract record."
    )

manual_decision_lookup = (
    manual_review_completed[
        [
            "_manual_review_match_key",
            "manual_review_decision",
            "manual_review_in_scope",
            "manual_review_reason_final",
            "manual_review_category"
        ]
    ]
    .drop_duplicates(
        subset=["_manual_review_match_key"],
        keep="first"
    )
    .copy()
)

manual_decision_lookup = (
    manual_decision_lookup
    .rename(
        columns={
            "manual_review_in_scope":
                "manual_review_in_scope_final"
        }
    )
)

df_scope = df_scope.merge(
    manual_decision_lookup,
    on="_manual_review_match_key",
    how="left",
    validate="many_to_one"
)

unmatched_manual_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & df_scope[
        "manual_review_decision"
    ].isna()
].copy()

if len(unmatched_manual_review_articles) > 0:

    unmatched_manual_review_articles.to_excel(
        OUTPUT_DIR /
        "WSJ_unmatched_manual_review_articles_check.xlsx",
        index=False
    )

    print(
        "\nArticles requiring manual review but not matched "
        "to the completed review file:"
    )

    display(
        unmatched_manual_review_articles[
            [
                "date_fixed",
                TITLE_COL,
                "manual_scope_review_reason"
            ]
        ]
    )

    raise ValueError(
        f"{len(unmatched_manual_review_articles)} article(s) "
        "requiring manual review could not be matched. "
        "See WSJ_unmatched_manual_review_articles_check.xlsx."
    )

df_scope["exclude_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "EXCLUDE"
    )
)

df_scope["include_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "INCLUDE"
    )
)

df_scope["final_scope_inclusion"] = (
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & ~df_scope[
        "exclude_by_manual_scope_review"
    ]
)

manually_included_articles = df_scope[
    df_scope["include_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manually_excluded_articles = df_scope[
    df_scope["exclude_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manually_included_articles.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_included_after_manual_review.xlsx",
    index=False
)

manually_excluded_articles.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_excluded_after_manual_review.xlsx",
    index=False
)

automatically_retained_scope = df_scope[
    df_scope["final_scope_inclusion"]
].copy()

manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

n_before_scope_screening = len(df_scope)

n_removed_automatic_scope = int(
    df_scope[
        "exclude_by_automated_scope_screening"
    ].sum()
)

n_before_manual_review = int(
    (
        df_scope[
            "requires_manual_scope_review"
        ]
        & ~df_scope[
            "exclude_by_automated_scope_screening"
        ]
    ).sum()
)

n_manual_review_included = int(
    df_scope[
        "include_by_manual_scope_review"
    ].sum()
)

n_manual_review_excluded = int(
    df_scope[
        "exclude_by_manual_scope_review"
    ].sum()
)

n_after_scope_screening = len(
    automatically_retained_scope
)

n_removed_scope_screening = (
    n_before_scope_screening
    - n_after_scope_screening
)

print("\n" + "=" * 75)
print("MANUAL SCOPE REVIEW APPLIED")
print("=" * 75)

print("\nInput to automated scope screening:")
print(n_before_scope_screening)

print("\nExcluded by automatic scope screening:")
print(n_removed_automatic_scope)

print("\nArticles sent for manual review:")
print(n_before_manual_review)

print("\nIncluded after manual review:")
print(n_manual_review_included)

print("\nExcluded after manual review:")
print(n_manual_review_excluded)

print("\nTotal removed by automatic and manual scope screening:")
print(n_removed_scope_screening)

print("\nRetained after automatic and manual scope screening:")
print(n_after_scope_screening)


# %%
# ============================================================
# 18. Normalise abstract for SHA-256 hashing
# ============================================================

df_hash = automatically_retained_scope.copy()


def normalize_content_for_hash(text):
    """
    Normalise abstract before exact-content hashing.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"[\u200b\u200c\u200d\ufeff]",
        "",
        text
    )

    text = (
        text
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(r"\s+", "", text)

    return text.strip()


def generate_sha256_hash(normalized_text):
    """
    Generate SHA-256 hash for non-empty normalised text.
    """

    if pd.isna(normalized_text):
        return pd.NA

    normalized_text = str(normalized_text)

    if normalized_text == "":
        return pd.NA

    return hashlib.sha256(
        normalized_text.encode("utf-8")
    ).hexdigest()


df_hash["content_normalized_for_hash"] = (
    df_hash["abstract"]
    .apply(normalize_content_for_hash)
)

df_hash["content_sha256"] = (
    df_hash["content_normalized_for_hash"]
    .apply(generate_sha256_hash)
)

df_hash["normalized_content_length"] = (
    df_hash["content_normalized_for_hash"]
    .str
    .len()
)

df_hash["_content_dedup_key"] = (
    df_hash["content_sha256"]
    .astype("string")
)

empty_hash_mask = (
    df_hash["content_sha256"].isna()
)

df_hash.loc[
    empty_hash_mask,
    "_content_dedup_key"
] = (
    "EMPTY_CONTENT_ROW_"
    + df_hash.loc[
        empty_hash_mask,
        "_original_order"
    ].astype(str)
)

df_hash["is_content_duplicate_group"] = (
    df_hash[
        "_content_dedup_key"
    ].duplicated(
        keep=False
    )
)

duplicate_content_rows = df_hash[
    df_hash["is_content_duplicate_group"]
].copy()

print(
    "\nRows in duplicated-abstract groups:",
    len(duplicate_content_rows)
)

print(
    "Number of duplicated-abstract groups:",
    duplicate_content_rows[
        "_content_dedup_key"
    ].nunique()
)

duplicate_content_rows.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
).to_excel(
    OUTPUT_DIR /
    "WSJ_duplicated_abstract_articles_check.xlsx",
    index=False
)


# %%
# ============================================================
# 19. Remove exact-abstract duplicates
#
# Retention rule:
# 1. Keep most recent publication date.
# 2. If dates are equal, keep earliest original row.
# ============================================================

df_hash_sorted = df_hash.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
)

df_no_content_duplicates = (
    df_hash_sorted
    .drop_duplicates(
        subset=["_content_dedup_key"],
        keep="first"
    )
    .copy()
)

kept_original_orders = set(
    df_no_content_duplicates[
        "_original_order"
    ]
)

removed_content_duplicates = df_hash[
    ~df_hash["_original_order"].isin(
        kept_original_orders
    )
].copy()

removed_content_duplicates.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_removed_by_abstract_hash.xlsx",
    index=False
)

n_before_hash_deduplication = len(df_hash)
n_after_hash_deduplication = len(
    df_no_content_duplicates
)

n_removed_content_duplicates = (
    n_before_hash_deduplication
    - n_after_hash_deduplication
)

print("\nBefore abstract-hash deduplication:")
print(n_before_hash_deduplication)

print("\nAfter abstract-hash deduplication:")
print(n_after_hash_deduplication)

print("\nRemoved by abstract-hash deduplication:")
print(n_removed_content_duplicates)


# %%
# ============================================================
# 20. Create final analytical corpus
# ============================================================

data = df_no_content_duplicates.copy()

data["date_fixed"] = pd.to_datetime(
    data["date_fixed"],
    errors="coerce"
)

data["year"] = (
    data["date_fixed"]
    .dt
    .year
)

data["month"] = (
    data["date_fixed"]
    .dt
    .month
)

data["month_name"] = (
    data["date_fixed"]
    .dt
    .month_name()
)

data["year_month"] = (
    data["date_fixed"]
    .dt
    .to_period("M")
    .astype(str)
)

data["abstract_char_length"] = (
    data["abstract"]
    .fillna("")
    .str
    .len()
)

data["abstract_word_count"] = (
    data["abstract"]
    .apply(count_english_words)
)

data = (
    data
    .sort_values(
        "date_fixed",
        na_position="last"
    )
    .reset_index(drop=True)
)


# %%
# ============================================================
# 21. Abstract extraction audit variables
# ============================================================

data["abstract_extracted"] = (
    data["abstract"]
    .fillna("")
    .str
    .strip()
    .ne("")
)

data["abstract_equals_full_content"] = (
    data["abstract"]
    .fillna("")
    .str
    .strip()
    ==
    data[CONTENT_COL]
    .fillna("")
    .str
    .strip()
)

data["abstract_starts_with_metadata"] = (
    data["abstract"]
    .fillna("")
    .str
    .contains(
        r"^\s*(?:ABSTRACT|Wall Street Journal Abstracts|"
        r"Copyright|Section:|Length:|Byline:)",
        regex=True,
        case=False,
        na=False
    )
)

failed_abstract_final = data[
    ~data["abstract_extracted"]
].copy()

short_abstracts = data[
    data["abstract_extracted"]
    & (
        data["abstract_word_count"] < 10
    )
].copy()

long_abstracts = data[
    data["abstract_word_count"] > 200
].copy()

remaining_metadata_problems = data[
    data["abstract_starts_with_metadata"]
].copy()

abstract_equals_full = data[
    data["abstract_extracted"]
    & data["abstract_equals_full_content"]
].copy()

failed_abstract_final.to_excel(
    OUTPUT_DIR /
    "WSJ_failed_abstract_extraction_final.xlsx",
    index=False
)

short_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_under_10_words_check.xlsx",
    index=False
)

long_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_over_200_words_check.xlsx",
    index=False
)

remaining_metadata_problems.to_excel(
    OUTPUT_DIR /
    "WSJ_remaining_abstract_metadata_check.xlsx",
    index=False
)

abstract_equals_full.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_equals_full_content_check.xlsx",
    index=False
)


# %%
# ============================================================
# 22. Abstract descriptive statistics
# ============================================================

valid_abstracts = data[
    data["abstract_extracted"]
].copy()

if len(valid_abstracts) == 0:
    raise ValueError(
        "No abstracts were successfully extracted."
    )

abstract_statistics = pd.DataFrame([
    {
        "metric": "Total final analytical articles",
        "value": len(data)
    },
    {
        "metric": "Successfully extracted abstracts",
        "value": len(valid_abstracts)
    },
    {
        "metric": "Failed abstract extractions",
        "value": len(failed_abstract_final)
    },
    {
        "metric": "Extraction success rate (%)",
        "value": round(
            len(valid_abstracts)
            / len(data)
            * 100,
            2
        )
    },
    {
        "metric": "Remaining possible metadata problems",
        "value": len(remaining_metadata_problems)
    },
    {
        "metric": "Abstract equals complete raw content",
        "value": len(abstract_equals_full)
    },
    {
        "metric": "Abstracts under 10 words",
        "value": len(short_abstracts)
    },
    {
        "metric": "Abstracts over 200 words",
        "value": len(long_abstracts)
    },
    {
        "metric": "Mean abstract word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].mean()
    },
    {
        "metric": "Median abstract word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].median()
    },
    {
        "metric": "Standard deviation",
        "value": valid_abstracts[
            "abstract_word_count"
        ].std()
    },
    {
        "metric": "Variance",
        "value": valid_abstracts[
            "abstract_word_count"
        ].var()
    },
    {
        "metric": "Minimum word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].min()
    },
    {
        "metric": "10th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.10)
    },
    {
        "metric": "25th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.25)
    },
    {
        "metric": "50th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.50)
    },
    {
        "metric": "75th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.75)
    },
    {
        "metric": "90th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.90)
    },
    {
        "metric": "95th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.95)
    },
    {
        "metric": "Maximum word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].max()
    },
    {
        "metric": "Skewness",
        "value": valid_abstracts[
            "abstract_word_count"
        ].skew()
    }
])

abstract_statistics["value"] = (
    abstract_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)

display(abstract_statistics)

abstract_statistics.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 23. Cleaning-process summary
# ============================================================

cleaning_summary = pd.DataFrame([
    {
        "step": "Original WSJ dataset",
        "article_count": n_original,
        "removed_at_step": 0,
        "retained_for_manual_review": 0,
        "screening_type": "Input data"
    },
    {
        "step":
            "After removing invalid dates and "
            "failed abstract extractions",
        "article_count":
            n_after_initial_cleaning,
        "removed_at_step":
            n_removed_initial_cleaning,
        "retained_for_manual_review":
            0,
        "screening_type":
            "Source-format and usability cleaning"
    },
    {
        "step":
            "After automated AV scope screening",
        "article_count":
            n_before_scope_screening
            - n_removed_automatic_scope,
        "removed_at_step":
            n_removed_automatic_scope,
        "retained_for_manual_review":
            n_before_manual_review,
        "screening_type":
            (
                "Automatic exclusion of clearly non-relevant, "
                "non-road and public-transport-only articles"
            )
    },
    {
        "step":
            "After manual scope review",
        "article_count":
            n_after_scope_screening,
        "removed_at_step":
            n_manual_review_excluded,
        "retained_for_manual_review":
            n_manual_review_included,
        "screening_type":
            (
                "Manual adjudication of borderline ADAS and "
                "company-context cases"
            )
    },
    {
        "step":
            "After abstract SHA-256 hash deduplication",
        "article_count":
            n_after_hash_deduplication,
        "removed_at_step":
            n_removed_content_duplicates,
        "retained_for_manual_review":
            int(
                df_no_content_duplicates[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Automated exact-abstract deduplication"
    },
    {
        "step":
            "Final analytical dataset",
        "article_count":
            len(data),
        "removed_at_step":
            0,
        "retained_for_manual_review":
            int(
                data[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Final retained sample"
    }
])

display(cleaning_summary)

cleaning_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_cleaning_process_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 24. Overall descriptive statistics
# ============================================================

overall_statistics = pd.DataFrame([
    {
        "metric": "Total final articles",
        "value": len(data)
    },
    {
        "metric":
            "Articles included after manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & (
                        data[
                            "manual_review_decision"
                        ]
                        == "INCLUDE"
                    )
                ).sum()
            )
    },
    {
        "metric":
            "Articles still awaiting manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & data[
                        "manual_review_decision"
                    ].isna()
                ).sum()
            )
    },
    {
        "metric": "Unique titles",
        "value": data[TITLE_COL].nunique()
    },
    {
        "metric": "Repeated titles remaining",
        "value": int(
            data[TITLE_COL]
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Unique abstract hashes",
        "value": data[
            "content_sha256"
        ].nunique(
            dropna=True
        )
    },
    {
        "metric": "Duplicate abstract hashes remaining",
        "value": int(
            data[
                "content_sha256"
            ]
            .dropna()
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Missing dates",
        "value": int(
            data[
                "date_fixed"
            ].isna().sum()
        )
    },
    {
        "metric": "Mean abstract word count",
        "value": data[
            "abstract_word_count"
        ].mean()
    },
    {
        "metric": "Median abstract word count",
        "value": data[
            "abstract_word_count"
        ].median()
    },
    {
        "metric": "Minimum abstract word count",
        "value": data[
            "abstract_word_count"
        ].min()
    },
    {
        "metric": "Maximum abstract word count",
        "value": data[
            "abstract_word_count"
        ].max()
    }
])

overall_statistics["value"] = (
    overall_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)

display(overall_statistics)

overall_statistics.to_excel(
    OUTPUT_DIR /
    "WSJ_overall_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 25. Article counts by year
# ============================================================

year_counts = (
    data.dropna(
        subset=["year"]
    )
    .groupby("year")
    .size()
    .reset_index(name="article_count")
    .sort_values("year")
)

display(year_counts)

year_counts.to_excel(
    OUTPUT_DIR /
    "WSJ_final_article_counts_by_year.xlsx",
    index=False
)

if len(year_counts) > 0:

    plt.figure(figsize=(10, 6))

    plt.bar(
        year_counts["year"]
        .astype(int)
        .astype(str),
        year_counts["article_count"]
    )

    plt.xlabel("Year")
    plt.ylabel("Number of Articles")

    plt.title(
        "Number of WSJ Autonomous-Vehicle Articles by Year"
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_article_counts_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 26. Complete monthly article timeline
# ============================================================

valid_date_data = data[
    data["date_fixed"].notna()
].copy()

if len(valid_date_data) > 0:

    year_month_counts = (
        valid_date_data
        .groupby("year_month")
        .size()
        .reset_index(name="article_count")
        .sort_values("year_month")
    )

    full_month_range = pd.period_range(
        start=valid_date_data[
            "date_fixed"
        ].min().to_period("M"),
        end=valid_date_data[
            "date_fixed"
        ].max().to_period("M"),
        freq="M"
    )

    complete_month_table = pd.DataFrame({
        "year_month":
            full_month_range.astype(str)
    })

    complete_month_table = (
        complete_month_table
        .merge(
            year_month_counts,
            on="year_month",
            how="left"
        )
    )

    complete_month_table["article_count"] = (
        complete_month_table[
            "article_count"
        ]
        .fillna(0)
        .astype(int)
    )

    complete_month_table.to_excel(
        OUTPUT_DIR /
        "WSJ_final_complete_monthly_counts.xlsx",
        index=False
    )

    plt.figure(figsize=(16, 6))

    plt.plot(
        complete_month_table["year_month"],
        complete_month_table["article_count"],
        marker="o",
        markersize=3
    )

    label_interval = max(
        1,
        len(complete_month_table) // 24
    )

    tick_positions = list(
        range(
            0,
            len(complete_month_table),
            label_interval
        )
    )

    plt.xticks(
        ticks=tick_positions,
        labels=complete_month_table.iloc[
            tick_positions
        ]["year_month"],
        rotation=90
    )

    plt.xlabel("Year-Month")
    plt.ylabel("Number of Articles")

    plt.title(
        "Monthly Number of WSJ "
        "Autonomous-Vehicle Articles"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_monthly_article_trend.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

else:
    print(
        "\nWARNING: No valid dates were available for "
        "the monthly timeline."
    )


# %%
# ============================================================
# 27. Abstract word-count distribution
# ============================================================

if len(valid_abstracts) > 0:

    plt.figure(figsize=(10, 6))

    plt.hist(
        valid_abstracts[
            "abstract_word_count"
        ],
        bins=30,
        edgecolor="black"
    )

    abstract_mean = valid_abstracts[
        "abstract_word_count"
    ].mean()

    abstract_median = valid_abstracts[
        "abstract_word_count"
    ].median()

    plt.axvline(
        abstract_mean,
        linestyle="--",
        linewidth=2,
        label=f"Mean = {abstract_mean:.1f}"
    )

    plt.axvline(
        abstract_median,
        linestyle=":",
        linewidth=2,
        label=f"Median = {abstract_median:.1f}"
    )

    plt.xlabel("Abstract Word Count")
    plt.ylabel("Number of Articles")

    plt.title(
        "Distribution of WSJ Abstract Lengths"
    )

    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_abstract_word_distribution.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 28. Abstract word-count statistics by year
# ============================================================

yearly_word_count = (
    data
    .dropna(
        subset=[
            "year",
            "abstract_word_count"
        ]
    )
    .groupby("year")[
        "abstract_word_count"
    ]
    .agg(
        article_count="count",
        mean_word_count="mean",
        median_word_count="median",
        minimum_word_count="min",
        maximum_word_count="max"
    )
    .reset_index()
)

yearly_word_count[
    [
        "mean_word_count",
        "median_word_count"
    ]
] = yearly_word_count[
    [
        "mean_word_count",
        "median_word_count"
    ]
].round(2)

display(yearly_word_count)

yearly_word_count.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_word_count_by_year.xlsx",
    index=False
)

if len(yearly_word_count) > 0:

    plt.figure(figsize=(10, 6))

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["mean_word_count"],
        marker="o",
        label="Mean"
    )

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["median_word_count"],
        marker="o",
        label="Median"
    )

    plt.xlabel("Year")
    plt.ylabel("Abstract Word Count")

    plt.title(
        "Mean and Median WSJ Abstract Word Count by Year"
    )

    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_abstract_word_count_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 29. Create combined abstract-extraction check file
# ============================================================

extraction_check = pd.concat(
    [
        data.head(100),
        remaining_metadata_problems,
        short_abstracts,
        long_abstracts,
        abstract_equals_full,
        failed_abstract_final
    ],
    ignore_index=True
)

if "content_sha256" in extraction_check.columns:

    non_empty_hash = (
        extraction_check[
            "content_sha256"
        ]
        .fillna("")
        .ne("")
    )

    extraction_check_with_hash = (
        extraction_check[
            non_empty_hash
        ]
        .drop_duplicates(
            subset=["content_sha256"],
            keep="first"
        )
    )

    extraction_check_without_hash = (
        extraction_check[
            ~non_empty_hash
        ]
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )

    extraction_check = pd.concat(
        [
            extraction_check_with_hash,
            extraction_check_without_hash
        ],
        ignore_index=True
    )

else:
    extraction_check = (
        extraction_check
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )

check_columns = [
    "date_fixed",
    TITLE_COL,
    "abstract",
    "abstract_word_count",
    "abstract_char_length",
    "abstract_extracted",
    "abstract_equals_full_content",
    "abstract_starts_with_metadata",
    CONTENT_COL
]

check_columns = [
    column
    for column in check_columns
    if column in extraction_check.columns
]

extraction_check[
    check_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 30. Remove internal helper columns from final output
# ============================================================

temporary_columns_to_remove = [
    "_original_order",
    "_content_dedup_key",
    "_manual_review_match_key"
]

data = data.drop(
    columns=[
        column
        for column in temporary_columns_to_remove
        if column in data.columns
    ]
)


# %%
# ============================================================
# 31. Save final analytical dataset
# ============================================================

data.to_excel(
    FINAL_DATA_PATH,
    index=False
)


# %%
# ============================================================
# 32. Final verification
# ============================================================

print("\n" + "=" * 75)
print("WSJ CLEANING COMPLETED")
print("=" * 75)

print("\nOriginal articles:")
print(n_original)

print(
    "\nRemoved because of invalid dates "
    "or failed abstract extraction:"
)
print(n_removed_initial_cleaning)

print("\nRemoved by automatic AV scope screening:")
print(n_removed_automatic_scope)

print("\nSent for manual scope review:")
print(n_before_manual_review)

print("\nIncluded after manual scope review:")
print(n_manual_review_included)

print("\nExcluded after manual scope review:")
print(n_manual_review_excluded)

print(
    "\nRetained after automatic and manual "
    "scope screening, before deduplication:"
)
print(n_after_scope_screening)

print("\nRemoved by abstract SHA-256 deduplication:")
print(n_removed_content_duplicates)

print("\nFinal analytical articles:")
print(len(data))

final_manually_included_count = int(
    (
        data["requires_manual_scope_review"]
        & (
            data["manual_review_decision"]
            == "INCLUDE"
        )
    ).sum()
)

final_pending_manual_review_count = int(
    (
        data["requires_manual_scope_review"]
        & data[
            "manual_review_decision"
        ].isna()
    ).sum()
)

print("\nFinal articles included after manual scope review:")
print(final_manually_included_count)

print("\nFinal articles still awaiting manual review:")
print(final_pending_manual_review_count)

print("\nMissing dates in final dataset:")
print(
    data["date_fixed"].isna().sum()
)

print("\nDuplicate abstract hashes remaining:")
print(
    data[
        "content_sha256"
    ]
    .dropna()
    .duplicated()
    .sum()
)

print("\nSuccessfully extracted abstracts:")
print(
    int(
        data[
            "abstract_extracted"
        ].sum()
    )
)

print("\nFailed abstract extractions:")
print(len(failed_abstract_final))

print("\nAbstracts under 10 words:")
print(len(short_abstracts))

print("\nAbstracts over 200 words:")
print(len(long_abstracts))

print("\nRemaining possible abstract metadata problems:")
print(len(remaining_metadata_problems))

print("\nMean abstract word count:")
print(
    round(
        valid_abstracts[
            "abstract_word_count"
        ].mean(),
        2
    )
)

print("\nMedian abstract word count:")
print(
    round(
        valid_abstracts[
            "abstract_word_count"
        ].median(),
        2
    )
)

print("\nFinal analytical dataset saved to:")
print(FINAL_DATA_PATH)

print("\nManual scope-review input file:")
print(
    OUTPUT_DIR /
    "WSJ_articles_requiring_manual_scope_review.xlsx"
)

print("\nAutomatically excluded scope file:")
print(
    OUTPUT_DIR /
    "WSJ_articles_excluded_by_AV_scope.xlsx"
)

print("\nAll audit and descriptive files saved in:")
print(OUTPUT_DIR)
```

```python
# %%
# ============================================================
# WSJ AV News Data Cleaning, Scope Screening,
# Abstract-Hash Deduplication, and Abstract Extraction
#
# Moderately relaxed screening version:
#
# 1. Load and inspect the original LexisNexis dataset
# 2. Standardise publication dates
# 3. Extract ABSTRACT to End of Document
# 4. Remove unusable records
# 5. Apply automated AV scope screening
# 6. Retain ambiguous ADAS/company-context cases for review
# 7. Apply completed manual scope-review decisions
# 8. Deduplicate using normalised abstract SHA-256 hashes
# 9. Produce audit files and descriptive statistics
# 10. Save the final analytical dataset
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import hashlib
import unicodedata


# %%
# ============================================================
# 1. File paths
# ============================================================

INPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/excel/"
    "lexis_structured_WSJ_1-389.xlsx"
)

OUTPUT_DIR = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_WSJ"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

FINAL_DATA_PATH = (
    OUTPUT_DIR /
    "WSJ_AV_final_analytical_dataset.xlsx"
)

MANUAL_REVIEW_COMPLETED_PATH = (
    OUTPUT_DIR /
    "WSJ_manual_scope_review_completed.xlsx"
)


# %%
# ============================================================
# 2. Load data
# ============================================================

df = pd.read_excel(INPUT_PATH)
n_original = len(df)

print("=" * 75)
print("WSJ AV DATA CLEANING")
print("=" * 75)

print("\nData loaded successfully.")
print("Original shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# %%
# ============================================================
# 3. Define key columns
# ============================================================

DATE_COL = "date"
TITLE_COL = "title"
CONTENT_COL = "content"
SOURCE_COL = "source"

required_columns = [DATE_COL, TITLE_COL, CONTENT_COL]

missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_required_columns:
    raise KeyError(
        "The following required columns are missing: "
        f"{missing_required_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

if SOURCE_COL not in df.columns:
    print(
        f"\nWARNING: '{SOURCE_COL}' column was not found. "
        "A default source value will be created."
    )
    df[SOURCE_COL] = "Wall Street Journal Abstracts"

df["_original_order"] = range(len(df))

df[TITLE_COL] = (
    df[TITLE_COL]
    .fillna("")
    .astype(str)
)

df[CONTENT_COL] = (
    df[CONTENT_COL]
    .fillna("")
    .astype(str)
)


# %%
# ============================================================
# 4. Missing-value and column summary
# ============================================================

column_summary = pd.DataFrame({
    "column": df.columns,
    "non_null_count": df.notna().sum().values,
    "missing_count": df.isna().sum().values,
    "missing_percent": (
        df.isna().mean().values * 100
    ).round(2),
    "dtype": df.dtypes.astype(str).values
})

display(column_summary)

column_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_column_missing_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 5. Standardise publication dates
#
# Example:
# December 14, 2023 Thursday
# -> 2023-12-14
# ============================================================

MONTH_PATTERN = (
    r"(?:January|February|March|April|May|June|"
    r"July|August|September|October|November|December)"
)

DATE_PATTERN = (
    rf"{MONTH_PATTERN}\s+"
    r"\d{1,2},\s+\d{4}"
)


def extract_english_date(value):
    """
    Extract an English-language date from a messy
    LexisNexis-style date string.
    """

    if pd.isna(value):
        return pd.NaT

    text = str(value).strip()

    match = re.search(
        DATE_PATTERN,
        text,
        flags=re.IGNORECASE
    )

    if match is None:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        format="%B %d, %Y",
        errors="coerce"
    )


df["date_original"] = df[DATE_COL]

df["date_parsed_direct"] = pd.to_datetime(
    df[DATE_COL],
    errors="coerce"
)

df["date_fixed"] = (
    df[DATE_COL]
    .apply(extract_english_date)
)

date_summary = pd.DataFrame([
    {
        "metric": "Total original articles",
        "value": len(df)
    },
    {
        "metric": "Dates parsed directly by pandas",
        "value": int(
            df["date_parsed_direct"].notna().sum()
        )
    },
    {
        "metric": "Dates recovered using date extraction",
        "value": int(
            df["date_fixed"].notna().sum()
        )
    },
    {
        "metric": "Dates still missing after extraction",
        "value": int(
            df["date_fixed"].isna().sum()
        )
    }
])

print("\nDate parsing summary:")
display(date_summary)

date_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_date_parsing_summary.xlsx",
    index=False
)

failed_date_rows = df[
    df["date_fixed"].isna()
].copy()

failed_date_rows.to_excel(
    OUTPUT_DIR /
    "WSJ_invalid_date_rows_check.xlsx",
    index=False
)


# %%
# ============================================================
# 6. Normalise line breaks
# ============================================================

def normalize_line_breaks(text):
    """
    Standardise line breaks while preserving text structure.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    if "\\n" in text and "\n" not in text:
        text = text.replace("\\n", "\n")

    text = re.sub(r"[ \t]+\n", "\n", text)
    text = re.sub(r"\n[ \t]+", "\n", text)
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


# %%
# ============================================================
# 7. Extract WSJ abstract
#
# Extract text between:
# ABSTRACT
# ...
# End of Document
# ============================================================

def extract_abstract(text):
    """
    Extract the WSJ abstract from a LexisNexis record.

    Preferred format:
    ABSTRACT
    abstract text
    End of Document

    A fallback pattern handles unusual line formatting.
    """

    text = normalize_line_breaks(text)

    if text == "":
        return ""

    main_pattern = re.compile(
        r"(?ims)"
        r"^[ \t]*ABSTRACT[ \t]*:?[ \t]*\n"
        r"(.*?)"
        r"(?=^[ \t]*End\s+of\s+Document[ \t]*$)"
    )

    match = main_pattern.search(text)

    if match is None:
        fallback_pattern = re.compile(
            r"(?is)"
            r"(?:^|\n|\s)ABSTRACT\s*:?\s*"
            r"(.*?)"
            r"\s*End\s+of\s+Document"
        )
        match = fallback_pattern.search(text)

    if match is None:
        return ""

    abstract = match.group(1).strip()

    abstract = re.sub(r"[\r\n\t]+", " ", abstract)
    abstract = re.sub(r"\s+", " ", abstract)
    abstract = abstract.strip(" \n\r\t:-")

    return abstract.strip()


df["abstract"] = (
    df[CONTENT_COL]
    .apply(extract_abstract)
)

df["abstract_extracted"] = (
    df["abstract"]
    .fillna("")
    .str
    .strip()
    .ne("")
)

print("\nAbstract extraction status:")
print(
    df["abstract_extracted"]
    .value_counts(dropna=False)
)

failed_abstract_rows = df[
    ~df["abstract_extracted"]
].copy()

failed_abstract_rows.to_excel(
    OUTPUT_DIR /
    "WSJ_failed_abstract_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 8. Basic text measures
# ============================================================

def count_english_words(text):
    """
    Approximate English word count.

    Expressions such as self-driving, company's, COVID-19
    and U.S. are generally treated as single word units.
    """

    if pd.isna(text):
        return 0

    text = str(text).strip()

    if text == "":
        return 0

    words = re.findall(
        r"\b[A-Za-z0-9]+"
        r"(?:['’.-][A-Za-z0-9]+)*\b",
        text
    )

    return len(words)


df["abstract_char_length"] = (
    df["abstract"]
    .fillna("")
    .str
    .len()
)

df["abstract_word_count"] = (
    df["abstract"]
    .apply(count_english_words)
)


# %%
# ============================================================
# 9. Initial usability exclusions
#
# Exclude:
# - missing or unrecoverable publication dates
# - failed or empty abstract extraction
# ============================================================

df["exclude_invalid_date"] = (
    df["date_fixed"].isna()
)

df["exclude_missing_abstract"] = (
    ~df["abstract_extracted"]
)

df["exclude_initial_unusable_record"] = (
    df["exclude_invalid_date"]
    | df["exclude_missing_abstract"]
)


def create_initial_exclusion_reason(row):

    reasons = []

    if row["exclude_invalid_date"]:
        reasons.append(
            "Missing or invalid publication date"
        )

    if row["exclude_missing_abstract"]:
        reasons.append(
            "Abstract could not be extracted"
        )

    return "; ".join(reasons)


df["initial_exclusion_reason"] = (
    df.apply(
        create_initial_exclusion_reason,
        axis=1
    )
)

initial_exclusion_summary = pd.DataFrame([
    {
        "rule": "Missing or invalid publication date",
        "article_count": int(
            df["exclude_invalid_date"].sum()
        )
    },
    {
        "rule": "Abstract could not be extracted",
        "article_count": int(
            df["exclude_missing_abstract"].sum()
        )
    },
    {
        "rule": "Excluded by one or more initial rules",
        "article_count": int(
            df[
                "exclude_initial_unusable_record"
            ].sum()
        )
    }
])

initial_exclusion_summary["percent_of_original"] = (
    initial_exclusion_summary["article_count"]
    / len(df)
    * 100
).round(2)

display(initial_exclusion_summary)

initial_exclusion_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_initial_exclusion_summary.xlsx",
    index=False
)

initially_excluded = df[
    df["exclude_initial_unusable_record"]
].copy()

initially_excluded.to_excel(
    OUTPUT_DIR /
    "WSJ_initially_excluded_records_check.xlsx",
    index=False
)

usable_data = df[
    ~df["exclude_initial_unusable_record"]
].copy()

n_after_initial_cleaning = len(usable_data)

n_removed_initial_cleaning = (
    len(df) - len(usable_data)
)

very_long_abstracts = usable_data[
    usable_data["abstract_word_count"] > 250
].copy()

very_long_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_over_250_words_check.xlsx",
    index=False
)

print(
    "\nAbstracts longer than 250 words, retained for audit:",
    len(very_long_abstracts)
)


# %%
# ============================================================
# 10. Create automated AV-screening text
#
# For WSJ, use title plus the complete extracted abstract.
# ============================================================

df_scope = usable_data.copy()

df_scope["automated_screen_text"] = (
    df_scope[TITLE_COL]
    + " "
    + df_scope["abstract"].fillna("")
)


def contains_any_pattern(text, patterns):
    """
    Return True if at least one regex pattern appears in text.
    """

    if pd.isna(text):
        return False

    text = str(text)

    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        is not None
        for pattern in patterns
    )


# %%
# ============================================================
# 11. Define English AV-scope patterns
# ============================================================

general_road_av_patterns = [
    r"\bautonomous\s+vehicles?\b",
    r"\bautonomous\s+cars?\b",
    r"\bautomated\s+vehicles?\b",
    r"\bautomated\s+cars?\b",
    r"\bautonomous\s+driving\b",
    r"\bautomated\s+driving\b",
    r"\bautomated\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+technology\b",
    r"\bself[-\s]?driving\s+technology\b",
    r"\bself[-\s]?driving\b",
    r"\bself[-\s]?driving\s+vehicles?\b",
    r"\bself[-\s]?driving\s+cars?\b",
    r"\bdriverless\b",
    r"\bdriverless\s+vehicles?\b",
    r"\bdriverless\s+cars?\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bself[-\s]?driving\s+taxis?\b",
    r"\bvehicle\s+automation\b",
    r"\bautomotive\s+automation\b",
    r"\bautomated\s+mobility\b",
    r"\bautonomous\s+mobility\b",
    r"\bfull\s+self[-\s]?driving\b",
]

higher_automation_patterns = [
    r"\bSAE\s+(?:Level\s*)?[345]\b",
    r"\bLevel[-\s]?[345]\b",
    r"\bL[345]\b",
    r"\bfully\s+autonomous\b",
    r"\bfull\s+autonomy\b",
    r"\bhighly\s+automated\b",
    r"\bhigh[-\s]?automation\b",
    r"\bdriverless\b",
    r"\bwithout\s+(?:a\s+)?safety\s+driver\b",
    r"\bno\s+safety\s+driver\b",
    r"\bremove(?:d|s|ing)?\s+(?:the\s+)?safety\s+driver\b",
    r"\bcommercial\s+autonomous\s+service\b",
    r"\bcommercial\s+driverless\s+service\b",
    r"\bcommercial\s+robotaxi\s+service\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+ride[-\s]?hailing\b",
]

passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bpersonal\s+vehicles?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\bminivans?\b",
    r"\btaxis?\b",
    r"\bcabs?\b",
    r"\bride[-\s]?hailing\b",
    r"\bride[-\s]?sharing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]

strong_passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\btaxis?\b",
    r"\bride[-\s]?hailing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]

bus_public_transport_patterns = [
    r"\bautonomous\s+buses?\b",
    r"\bdriverless\s+buses?\b",
    r"\bself[-\s]?driving\s+buses?\b",
    r"\bautomated\s+buses?\b",
    r"\bautonomous\s+shuttles?\b",
    r"\bdriverless\s+shuttles?\b",
    r"\bself[-\s]?driving\s+shuttles?\b",
    r"\bautomated\s+shuttles?\b",
    r"\bautonomous\s+minibuses?\b",
    r"\bdriverless\s+minibuses?\b",
    r"\bpublic\s+transit\b",
    r"\bpublic\s+transportation\b",
    r"\btransit\s+service\b",
    r"\bpeople\s+movers?\b",
]

adas_patterns = [
    r"\bADAS\b",
    r"\badvanced\s+driver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assist(?:ance)?\b",
    r"\bautomated\s+parking\b",
    r"\bautomatic\s+parking\b",
    r"\bself[-\s]?parking\b",
    r"\badaptive\s+cruise\s+control\b",
    r"\blane[-\s]?keeping\b",
    r"\blane[-\s]?centering\b",
    r"\blane[-\s]?departure\s+warning\b",
    r"\bautomatic\s+emergency\s+braking\b",
    r"\bblind[-\s]?spot\s+monitoring\b",
    r"\bcollision\s+warning\b",
    r"\bparking\s+assist\b",
    r"\bSuper\s+Cruise\b",
    r"\bBlueCruise\b",
    r"\bAutopilot\b",
]

road_vehicle_context_patterns = [
    r"\bvehicles?\b",
    r"\bcars?\b",
    r"\bautomobiles?\b",
    r"\bpassenger\s+vehicles?\b",
    r"\btaxis?\b",
    r"\broads?\b",
    r"\bhighways?\b",
    r"\bstreets?\b",
    r"\bautomakers?\b",
    r"\bcar\s+companies\b",
    r"\broad\s+testing\b",
    r"\bpublic\s+roads?\b",
    r"\btest\s+vehicles?\b",
]

strong_non_road_patterns = [
    r"\bautonomous\s+drones?\b",
    r"\bdriverless\s+drones?\b",
    r"\bunmanned\s+aerial\s+vehicles?\b",
    r"\bUAVs?\b",
    r"\bautonomous\s+aircraft\b",
    r"\bpilotless\s+aircraft\b",
    r"\bautonomous\s+ships?\b",
    r"\bunmanned\s+ships?\b",
    r"\bdriverless\s+ships?\b",
    r"\bautonomous\s+boats?\b",
    r"\bunmanned\s+boats?\b",
    r"\bautonomous\s+vessels?\b",
    r"\bdriverless\s+trains?\b",
    r"\bautonomous\s+trains?\b",
    r"\bautomated\s+trains?\b",
    r"\bdriverless\s+metros?\b",
    r"\bautomated\s+metros?\b",
    r"\bautonomous\s+mining\s+trucks?\b",
    r"\bdriverless\s+mining\s+trucks?\b",
    r"\bautonomous\s+mine\s+vehicles?\b",
    r"\bdriverless\s+mine\s+vehicles?\b",
    r"\bautonomous\s+port\s+vehicles?\b",
    r"\bdriverless\s+port\s+vehicles?\b",
    r"\bautomated\s+container\s+vehicles?\b",
    r"\bautonomous\s+tractors?\b",
    r"\bdriverless\s+tractors?\b",
    r"\bautonomous\s+farm\s+equipment\b",
    r"\bunmanned\s+farm\s+equipment\b",
]

general_non_road_context_patterns = [
    r"\bdrones?\b",
    r"\baircraft\b",
    r"\baviation\b",
    r"\bships?\b",
    r"\bboats?\b",
    r"\bvessels?\b",
    r"\brailways?\b",
    r"\btrains?\b",
    r"\bmetros?\b",
    r"\bsubways?\b",
    r"\bmines?\b",
    r"\bmining\b",
    r"\bports?\b",
    r"\bharbors?\b",
    r"\btractors?\b",
    r"\bagricultural\s+machinery\b",
]

av_company_patterns = [
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
    r"\bMotional\b",
    r"\bArgo\s+AI\b",
    r"\bAurora\b",
    r"\bNuro\b",
    r"\bMay\s+Mobility\b",
    r"\bMobileye\b",
    r"\bTesla\s+Autopilot\b",
    r"\bFull\s+Self[-\s]?Driving\b",
    r"\bTesla\s+FSD\b",
    r"\bGoogle\s+self[-\s]?driving\b",
    r"\bUber\s+autonomous\b",
    r"\bUber\s+self[-\s]?driving\b",
    r"\bGM\s+Cruise\b",
]


# %%
# ============================================================
# 12. Generate automated screening indicators
# ============================================================

pattern_groups = {
    "has_general_road_av_term":
        general_road_av_patterns,
    "has_higher_automation_term":
        higher_automation_patterns,
    "has_passenger_robotaxi_term":
        passenger_robotaxi_patterns,
    "has_strong_passenger_robotaxi_term":
        strong_passenger_robotaxi_patterns,
    "has_bus_public_transport_term":
        bus_public_transport_patterns,
    "has_adas_term":
        adas_patterns,
    "has_road_vehicle_context":
        road_vehicle_context_patterns,
    "has_strong_non_road_term":
        strong_non_road_patterns,
    "has_general_non_road_context":
        general_non_road_context_patterns,
    "has_av_company_term":
        av_company_patterns
}

for output_column, patterns in pattern_groups.items():
    df_scope[output_column] = (
        df_scope["automated_screen_text"]
        .apply(
            lambda text: contains_any_pattern(
                text,
                patterns
            )
        )
    )

df_scope["has_non_road_term"] = (
    df_scope["has_strong_non_road_term"]
    | df_scope["has_general_non_road_context"]
)


# %%
# ============================================================
# 13. Create broader AV-relevance indicators
# ============================================================

df_scope["has_explicit_av_signal"] = (
    df_scope["has_general_road_av_term"]
    | df_scope["has_higher_automation_term"]
    | df_scope["has_passenger_robotaxi_term"]
)

df_scope["has_company_supported_av_signal"] = (
    df_scope["has_av_company_term"]
    & (
        df_scope["has_road_vehicle_context"]
        | df_scope["has_adas_term"]
        | df_scope["has_higher_automation_term"]
    )
)

df_scope["has_broad_road_av_relevance"] = (
    df_scope["has_explicit_av_signal"]
    | df_scope["has_company_supported_av_signal"]
)


# %%
# ============================================================
# 14. Define moderately relaxed exclusion rules
# ============================================================

df_scope["remove_no_relevant_road_av_term"] = (
    ~df_scope["has_broad_road_av_relevance"]
)

df_scope["remove_pure_non_road"] = (
    df_scope["has_strong_non_road_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)

df_scope["remove_bus_public_transport_only"] = (
    df_scope["has_bus_public_transport_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)

df_scope["review_adas_only"] = (
    df_scope["has_adas_term"]
    & ~df_scope["has_higher_automation_term"]
)

df_scope["remove_adas_only"] = False

df_scope["exclude_by_automated_scope_screening"] = (
    df_scope["remove_no_relevant_road_av_term"]
    | df_scope["remove_pure_non_road"]
    | df_scope["remove_bus_public_transport_only"]
)

df_scope["requires_manual_scope_review"] = (
    df_scope["review_adas_only"]
    | (
        ~df_scope["has_explicit_av_signal"]
        & df_scope["has_company_supported_av_signal"]
    )
)


# %%
# ============================================================
# 15. Create exclusion and manual-review reasons
# ============================================================

def create_scope_exclusion_reason(row):

    reasons = []

    if row["remove_no_relevant_road_av_term"]:
        reasons.append(
            "No broad road autonomous-vehicle relevance signal"
        )

    if row["remove_pure_non_road"]:
        reasons.append(
            "Pure non-road autonomous-system article"
        )

    if row["remove_bus_public_transport_only"]:
        reasons.append(
            "Bus, shuttle or public-transport-only article"
        )

    return "; ".join(reasons)


df_scope["automated_scope_exclusion_reason"] = (
    df_scope.apply(
        create_scope_exclusion_reason,
        axis=1
    )
)


def create_manual_review_reason(row):

    reasons = []

    if row["review_adas_only"]:
        reasons.append(
            "ADAS terminology without explicit "
            "higher-level automation indicator"
        )

    if (
        not row["has_explicit_av_signal"]
        and row["has_company_supported_av_signal"]
    ):
        reasons.append(
            "AV relevance inferred from company/programme "
            "and road-vehicle context"
        )

    return "; ".join(reasons)


df_scope["manual_scope_review_reason"] = (
    df_scope.apply(
        create_manual_review_reason,
        axis=1
    )
)


# %%
# ============================================================
# 16. Automated scope-screening summary
# ============================================================

scope_summary = pd.DataFrame([
    {
        "screening_rule":
            "No broad road autonomous-vehicle relevance signal",
        "article_count":
            int(
                df_scope[
                    "remove_no_relevant_road_av_term"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Pure non-road autonomous-system article",
        "article_count":
            int(
                df_scope[
                    "remove_pure_non_road"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Bus, shuttle or public-transport-only article",
        "article_count":
            int(
                df_scope[
                    "remove_bus_public_transport_only"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "ADAS case without explicit higher-level automation",
        "article_count":
            int(
                df_scope[
                    "review_adas_only"
                ].sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Company/context-supported borderline AV case",
        "article_count":
            int(
                (
                    ~df_scope["has_explicit_av_signal"]
                    & df_scope[
                        "has_company_supported_av_signal"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Excluded by one or more automatic scope rules",
        "article_count":
            int(
                df_scope[
                    "exclude_by_automated_scope_screening"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Retained articles requiring manual scope review",
        "article_count":
            int(
                (
                    df_scope[
                        "requires_manual_scope_review"
                    ]
                    & ~df_scope[
                        "exclude_by_automated_scope_screening"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    }
])

scope_summary["percent_of_scope_input"] = (
    scope_summary["article_count"]
    / len(df_scope)
    * 100
).round(2)

print("\nAutomated AV-scope screening summary:")
display(scope_summary)

scope_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_automated_AV_scope_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 17. Export scope-screening audit files
# ============================================================

scope_audit_columns = [
    "date_fixed",
    DATE_COL,
    TITLE_COL,
    "has_general_road_av_term",
    "has_higher_automation_term",
    "has_passenger_robotaxi_term",
    "has_strong_passenger_robotaxi_term",
    "has_bus_public_transport_term",
    "has_adas_term",
    "has_road_vehicle_context",
    "has_strong_non_road_term",
    "has_general_non_road_context",
    "has_non_road_term",
    "has_av_company_term",
    "has_explicit_av_signal",
    "has_company_supported_av_signal",
    "has_broad_road_av_relevance",
    "remove_no_relevant_road_av_term",
    "remove_pure_non_road",
    "remove_bus_public_transport_only",
    "remove_adas_only",
    "review_adas_only",
    "requires_manual_scope_review",
    "automated_scope_exclusion_reason",
    "manual_scope_review_reason",
    "automated_screen_text",
    "abstract"
]

scope_audit_columns = [
    column
    for column in scope_audit_columns
    if column in df_scope.columns
]

automatically_excluded_scope = df_scope[
    df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

automatically_retained_scope = df_scope[
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

automatically_excluded_scope[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_articles_excluded_by_AV_scope.xlsx",
    index=False
)

manual_scope_review_articles[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_articles_requiring_manual_scope_review.xlsx",
    index=False
)

df_scope[
    df_scope["remove_no_relevant_road_av_term"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_no_broad_AV_signal_excluded_check.xlsx",
    index=False
)

df_scope[
    df_scope["remove_pure_non_road"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_non_road_articles_excluded_check.xlsx",
    index=False
)

df_scope[
    df_scope["remove_bus_public_transport_only"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "WSJ_bus_only_articles_excluded_check.xlsx",
    index=False
)

adas_review_cases = df_scope[
    df_scope["review_adas_only"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

adas_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_ADAS_cases_retained_for_review.xlsx",
    index=False
)

company_context_review_cases = df_scope[
    ~df_scope["has_explicit_av_signal"]
    & df_scope[
        "has_company_supported_av_signal"
    ]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

company_context_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_company_context_AV_cases_retained_for_review.xlsx",
    index=False
)

mixed_bus_retained = df_scope[
    df_scope["has_bus_public_transport_term"]
    & df_scope["has_strong_passenger_robotaxi_term"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_bus_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_mixed_bus_passenger_articles_retained_check.xlsx",
    index=False
)

mixed_adas_retained = df_scope[
    df_scope["has_adas_term"]
    & df_scope["has_higher_automation_term"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_adas_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_mixed_ADAS_higher_automation_retained_check.xlsx",
    index=False
)


# %%
# ============================================================
# 17B. Apply completed manual scope-review decisions
#
# Run the script once to create:
# WSJ_articles_requiring_manual_scope_review.xlsx
#
# Complete the review and save a copy as:
# WSJ_manual_scope_review_completed.xlsx
#
# Required decision values:
# INCLUDE or EXCLUDE
# ============================================================

if not MANUAL_REVIEW_COMPLETED_PATH.exists():

    raise FileNotFoundError(
        "\nThe completed manual-review file could not be found:\n"
        f"{MANUAL_REVIEW_COMPLETED_PATH}\n\n"
        "First review WSJ_articles_requiring_manual_scope_review.xlsx. "
        "Then save the completed file as "
        "WSJ_manual_scope_review_completed.xlsx inside OUTPUT_DIR."
    )

manual_review_completed = pd.read_excel(
    MANUAL_REVIEW_COMPLETED_PATH,
    sheet_name="Sheet1"
)

print("\nCompleted manual-review file loaded:")
print(MANUAL_REVIEW_COMPLETED_PATH)

print("\nManual-review file shape:")
print(manual_review_completed.shape)

print("\nManual-review file columns:")
print(manual_review_completed.columns.tolist())

required_manual_review_columns = [
    TITLE_COL,
    "abstract",
    "manual_review_decision",
    "manual_review_in_scope",
    "manual_review_reason_final",
    "manual_review_category"
]

missing_manual_review_columns = [
    column
    for column in required_manual_review_columns
    if column not in manual_review_completed.columns
]

if missing_manual_review_columns:
    raise KeyError(
        "The completed manual-review file is missing these columns: "
        f"{missing_manual_review_columns}\n"
        f"Available columns: "
        f"{manual_review_completed.columns.tolist()}"
    )


def normalize_manual_review_match_text(text):
    """
    Normalise title/abstract text for matching the manually
    reviewed file back to the main dataset.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = (
        text
        .replace("\u00a0", " ")
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()


def create_manual_review_match_key(title, abstract):
    """
    Create a SHA-256 matching key from the normalised title and
    abstract.
    """

    normalized_title = (
        normalize_manual_review_match_text(title)
    )

    normalized_abstract = (
        normalize_manual_review_match_text(abstract)
    )

    combined_text = (
        normalized_title
        + "\n"
        + normalized_abstract
    )

    return hashlib.sha256(
        combined_text.encode("utf-8")
    ).hexdigest()


df_scope["_manual_review_match_key"] = (
    df_scope.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["abstract"]
        ),
        axis=1
    )
)

manual_review_completed[
    "_manual_review_match_key"
] = (
    manual_review_completed.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["abstract"]
        ),
        axis=1
    )
)

manual_review_completed[
    "manual_review_decision"
] = (
    manual_review_completed[
        "manual_review_decision"
    ]
    .fillna("")
    .astype(str)
    .str
    .strip()
    .str
    .upper()
)

valid_manual_decisions = {
    "INCLUDE",
    "EXCLUDE"
}

invalid_manual_decision_rows = (
    manual_review_completed[
        ~manual_review_completed[
            "manual_review_decision"
        ].isin(valid_manual_decisions)
    ]
    .copy()
)

if len(invalid_manual_decision_rows) > 0:

    display_columns = [
        column
        for column in [
            "date_fixed",
            TITLE_COL,
            "manual_review_decision"
        ]
        if column
        in invalid_manual_decision_rows.columns
    ]

    print(
        "\nInvalid or missing manual-review decisions:"
    )

    display(
        invalid_manual_decision_rows[
            display_columns
        ]
    )

    raise ValueError(
        "Every reviewed article must have either INCLUDE "
        "or EXCLUDE in manual_review_decision."
    )

decision_conflicts = (
    manual_review_completed
    .groupby(
        "_manual_review_match_key"
    )["manual_review_decision"]
    .nunique()
)

conflicting_keys = set(
    decision_conflicts[
        decision_conflicts > 1
    ].index
)

if conflicting_keys:

    conflicting_rows = (
        manual_review_completed[
            manual_review_completed[
                "_manual_review_match_key"
            ].isin(conflicting_keys)
        ]
        .copy()
    )

    display(
        conflicting_rows[
            [
                TITLE_COL,
                "manual_review_decision",
                "manual_review_reason_final"
            ]
        ]
    )

    raise ValueError(
        "Conflicting manual decisions were found for "
        "the same title/abstract record."
    )

manual_decision_lookup = (
    manual_review_completed[
        [
            "_manual_review_match_key",
            "manual_review_decision",
            "manual_review_in_scope",
            "manual_review_reason_final",
            "manual_review_category"
        ]
    ]
    .drop_duplicates(
        subset=["_manual_review_match_key"],
        keep="first"
    )
    .copy()
)

manual_decision_lookup = (
    manual_decision_lookup
    .rename(
        columns={
            "manual_review_in_scope":
                "manual_review_in_scope_final"
        }
    )
)

df_scope = df_scope.merge(
    manual_decision_lookup,
    on="_manual_review_match_key",
    how="left",
    validate="many_to_one"
)

unmatched_manual_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & df_scope[
        "manual_review_decision"
    ].isna()
].copy()

if len(unmatched_manual_review_articles) > 0:

    unmatched_manual_review_articles.to_excel(
        OUTPUT_DIR /
        "WSJ_unmatched_manual_review_articles_check.xlsx",
        index=False
    )

    print(
        "\nArticles requiring manual review but not matched "
        "to the completed review file:"
    )

    display(
        unmatched_manual_review_articles[
            [
                "date_fixed",
                TITLE_COL,
                "manual_scope_review_reason"
            ]
        ]
    )

    raise ValueError(
        f"{len(unmatched_manual_review_articles)} article(s) "
        "requiring manual review could not be matched. "
        "See WSJ_unmatched_manual_review_articles_check.xlsx."
    )

df_scope["exclude_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "EXCLUDE"
    )
)

df_scope["include_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "INCLUDE"
    )
)

df_scope["final_scope_inclusion"] = (
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & ~df_scope[
        "exclude_by_manual_scope_review"
    ]
)

manually_included_articles = df_scope[
    df_scope["include_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manually_excluded_articles = df_scope[
    df_scope["exclude_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

manually_included_articles.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_included_after_manual_review.xlsx",
    index=False
)

manually_excluded_articles.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_excluded_after_manual_review.xlsx",
    index=False
)

automatically_retained_scope = df_scope[
    df_scope["final_scope_inclusion"]
].copy()

manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

n_before_scope_screening = len(df_scope)

n_removed_automatic_scope = int(
    df_scope[
        "exclude_by_automated_scope_screening"
    ].sum()
)

n_before_manual_review = int(
    (
        df_scope[
            "requires_manual_scope_review"
        ]
        & ~df_scope[
            "exclude_by_automated_scope_screening"
        ]
    ).sum()
)

n_manual_review_included = int(
    df_scope[
        "include_by_manual_scope_review"
    ].sum()
)

n_manual_review_excluded = int(
    df_scope[
        "exclude_by_manual_scope_review"
    ].sum()
)

n_after_scope_screening = len(
    automatically_retained_scope
)

n_removed_scope_screening = (
    n_before_scope_screening
    - n_after_scope_screening
)

print("\n" + "=" * 75)
print("MANUAL SCOPE REVIEW APPLIED")
print("=" * 75)

print("\nInput to automated scope screening:")
print(n_before_scope_screening)

print("\nExcluded by automatic scope screening:")
print(n_removed_automatic_scope)

print("\nArticles sent for manual review:")
print(n_before_manual_review)

print("\nIncluded after manual review:")
print(n_manual_review_included)

print("\nExcluded after manual review:")
print(n_manual_review_excluded)

print("\nTotal removed by automatic and manual scope screening:")
print(n_removed_scope_screening)

print("\nRetained after automatic and manual scope screening:")
print(n_after_scope_screening)


# %%
# ============================================================
# 18. Normalise abstract for SHA-256 hashing
# ============================================================

df_hash = automatically_retained_scope.copy()


def normalize_content_for_hash(text):
    """
    Normalise abstract before exact-content hashing.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"[\u200b\u200c\u200d\ufeff]",
        "",
        text
    )

    text = (
        text
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(r"\s+", "", text)

    return text.strip()


def generate_sha256_hash(normalized_text):
    """
    Generate SHA-256 hash for non-empty normalised text.
    """

    if pd.isna(normalized_text):
        return pd.NA

    normalized_text = str(normalized_text)

    if normalized_text == "":
        return pd.NA

    return hashlib.sha256(
        normalized_text.encode("utf-8")
    ).hexdigest()


df_hash["content_normalized_for_hash"] = (
    df_hash["abstract"]
    .apply(normalize_content_for_hash)
)

df_hash["content_sha256"] = (
    df_hash["content_normalized_for_hash"]
    .apply(generate_sha256_hash)
)

df_hash["normalized_content_length"] = (
    df_hash["content_normalized_for_hash"]
    .str
    .len()
)

df_hash["_content_dedup_key"] = (
    df_hash["content_sha256"]
    .astype("string")
)

empty_hash_mask = (
    df_hash["content_sha256"].isna()
)

df_hash.loc[
    empty_hash_mask,
    "_content_dedup_key"
] = (
    "EMPTY_CONTENT_ROW_"
    + df_hash.loc[
        empty_hash_mask,
        "_original_order"
    ].astype(str)
)

df_hash["is_content_duplicate_group"] = (
    df_hash[
        "_content_dedup_key"
    ].duplicated(
        keep=False
    )
)

duplicate_content_rows = df_hash[
    df_hash["is_content_duplicate_group"]
].copy()

print(
    "\nRows in duplicated-abstract groups:",
    len(duplicate_content_rows)
)

print(
    "Number of duplicated-abstract groups:",
    duplicate_content_rows[
        "_content_dedup_key"
    ].nunique()
)

duplicate_content_rows.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
).to_excel(
    OUTPUT_DIR /
    "WSJ_duplicated_abstract_articles_check.xlsx",
    index=False
)


# %%
# ============================================================
# 19. Remove exact-abstract duplicates
#
# Retention rule:
# 1. Keep most recent publication date.
# 2. If dates are equal, keep earliest original row.
# ============================================================

df_hash_sorted = df_hash.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
)

df_no_content_duplicates = (
    df_hash_sorted
    .drop_duplicates(
        subset=["_content_dedup_key"],
        keep="first"
    )
    .copy()
)

kept_original_orders = set(
    df_no_content_duplicates[
        "_original_order"
    ]
)

removed_content_duplicates = df_hash[
    ~df_hash["_original_order"].isin(
        kept_original_orders
    )
].copy()

removed_content_duplicates.to_excel(
    OUTPUT_DIR /
    "WSJ_articles_removed_by_abstract_hash.xlsx",
    index=False
)

n_before_hash_deduplication = len(df_hash)
n_after_hash_deduplication = len(
    df_no_content_duplicates
)

n_removed_content_duplicates = (
    n_before_hash_deduplication
    - n_after_hash_deduplication
)

print("\nBefore abstract-hash deduplication:")
print(n_before_hash_deduplication)

print("\nAfter abstract-hash deduplication:")
print(n_after_hash_deduplication)

print("\nRemoved by abstract-hash deduplication:")
print(n_removed_content_duplicates)


# %%
# ============================================================
# 20. Create final analytical corpus
# ============================================================

data = df_no_content_duplicates.copy()

data["date_fixed"] = pd.to_datetime(
    data["date_fixed"],
    errors="coerce"
)

data["year"] = (
    data["date_fixed"]
    .dt
    .year
)

data["month"] = (
    data["date_fixed"]
    .dt
    .month
)

data["month_name"] = (
    data["date_fixed"]
    .dt
    .month_name()
)

data["year_month"] = (
    data["date_fixed"]
    .dt
    .to_period("M")
    .astype(str)
)

data["abstract_char_length"] = (
    data["abstract"]
    .fillna("")
    .str
    .len()
)

data["abstract_word_count"] = (
    data["abstract"]
    .apply(count_english_words)
)

data = (
    data
    .sort_values(
        "date_fixed",
        na_position="last"
    )
    .reset_index(drop=True)
)


# %%
# ============================================================
# 21. Abstract extraction audit variables
# ============================================================

data["abstract_extracted"] = (
    data["abstract"]
    .fillna("")
    .str
    .strip()
    .ne("")
)

data["abstract_equals_full_content"] = (
    data["abstract"]
    .fillna("")
    .str
    .strip()
    ==
    data[CONTENT_COL]
    .fillna("")
    .str
    .strip()
)

data["abstract_starts_with_metadata"] = (
    data["abstract"]
    .fillna("")
    .str
    .contains(
        r"^\s*(?:ABSTRACT|Wall Street Journal Abstracts|"
        r"Copyright|Section:|Length:|Byline:)",
        regex=True,
        case=False,
        na=False
    )
)

failed_abstract_final = data[
    ~data["abstract_extracted"]
].copy()

short_abstracts = data[
    data["abstract_extracted"]
    & (
        data["abstract_word_count"] < 10
    )
].copy()

long_abstracts = data[
    data["abstract_word_count"] > 200
].copy()

remaining_metadata_problems = data[
    data["abstract_starts_with_metadata"]
].copy()

abstract_equals_full = data[
    data["abstract_extracted"]
    & data["abstract_equals_full_content"]
].copy()

failed_abstract_final.to_excel(
    OUTPUT_DIR /
    "WSJ_failed_abstract_extraction_final.xlsx",
    index=False
)

short_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_under_10_words_check.xlsx",
    index=False
)

long_abstracts.to_excel(
    OUTPUT_DIR /
    "WSJ_abstracts_over_200_words_check.xlsx",
    index=False
)

remaining_metadata_problems.to_excel(
    OUTPUT_DIR /
    "WSJ_remaining_abstract_metadata_check.xlsx",
    index=False
)

abstract_equals_full.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_equals_full_content_check.xlsx",
    index=False
)


# %%
# ============================================================
# 22. Abstract descriptive statistics
# ============================================================

valid_abstracts = data[
    data["abstract_extracted"]
].copy()

if len(valid_abstracts) == 0:
    raise ValueError(
        "No abstracts were successfully extracted."
    )

abstract_statistics = pd.DataFrame([
    {
        "metric": "Total final analytical articles",
        "value": len(data)
    },
    {
        "metric": "Successfully extracted abstracts",
        "value": len(valid_abstracts)
    },
    {
        "metric": "Failed abstract extractions",
        "value": len(failed_abstract_final)
    },
    {
        "metric": "Extraction success rate (%)",
        "value": round(
            len(valid_abstracts)
            / len(data)
            * 100,
            2
        )
    },
    {
        "metric": "Remaining possible metadata problems",
        "value": len(remaining_metadata_problems)
    },
    {
        "metric": "Abstract equals complete raw content",
        "value": len(abstract_equals_full)
    },
    {
        "metric": "Abstracts under 10 words",
        "value": len(short_abstracts)
    },
    {
        "metric": "Abstracts over 200 words",
        "value": len(long_abstracts)
    },
    {
        "metric": "Mean abstract word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].mean()
    },
    {
        "metric": "Median abstract word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].median()
    },
    {
        "metric": "Standard deviation",
        "value": valid_abstracts[
            "abstract_word_count"
        ].std()
    },
    {
        "metric": "Variance",
        "value": valid_abstracts[
            "abstract_word_count"
        ].var()
    },
    {
        "metric": "Minimum word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].min()
    },
    {
        "metric": "10th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.10)
    },
    {
        "metric": "25th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.25)
    },
    {
        "metric": "50th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.50)
    },
    {
        "metric": "75th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.75)
    },
    {
        "metric": "90th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.90)
    },
    {
        "metric": "95th percentile",
        "value": valid_abstracts[
            "abstract_word_count"
        ].quantile(0.95)
    },
    {
        "metric": "Maximum word count",
        "value": valid_abstracts[
            "abstract_word_count"
        ].max()
    },
    {
        "metric": "Skewness",
        "value": valid_abstracts[
            "abstract_word_count"
        ].skew()
    }
])

abstract_statistics["value"] = (
    abstract_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)

display(abstract_statistics)

abstract_statistics.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 23. Cleaning-process summary
# ============================================================

cleaning_summary = pd.DataFrame([
    {
        "step": "Original WSJ dataset",
        "article_count": n_original,
        "removed_at_step": 0,
        "retained_for_manual_review": 0,
        "screening_type": "Input data"
    },
    {
        "step":
            "After removing invalid dates and "
            "failed abstract extractions",
        "article_count":
            n_after_initial_cleaning,
        "removed_at_step":
            n_removed_initial_cleaning,
        "retained_for_manual_review":
            0,
        "screening_type":
            "Source-format and usability cleaning"
    },
    {
        "step":
            "After automated AV scope screening",
        "article_count":
            n_before_scope_screening
            - n_removed_automatic_scope,
        "removed_at_step":
            n_removed_automatic_scope,
        "retained_for_manual_review":
            n_before_manual_review,
        "screening_type":
            (
                "Automatic exclusion of clearly non-relevant, "
                "non-road and public-transport-only articles"
            )
    },
    {
        "step":
            "After manual scope review",
        "article_count":
            n_after_scope_screening,
        "removed_at_step":
            n_manual_review_excluded,
        "retained_for_manual_review":
            n_manual_review_included,
        "screening_type":
            (
                "Manual adjudication of borderline ADAS and "
                "company-context cases"
            )
    },
    {
        "step":
            "After abstract SHA-256 hash deduplication",
        "article_count":
            n_after_hash_deduplication,
        "removed_at_step":
            n_removed_content_duplicates,
        "retained_for_manual_review":
            int(
                df_no_content_duplicates[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Automated exact-abstract deduplication"
    },
    {
        "step":
            "Final analytical dataset",
        "article_count":
            len(data),
        "removed_at_step":
            0,
        "retained_for_manual_review":
            int(
                data[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Final retained sample"
    }
])

display(cleaning_summary)

cleaning_summary.to_excel(
    OUTPUT_DIR /
    "WSJ_cleaning_process_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 24. Overall descriptive statistics
# ============================================================

overall_statistics = pd.DataFrame([
    {
        "metric": "Total final articles",
        "value": len(data)
    },
    {
        "metric":
            "Articles included after manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & (
                        data[
                            "manual_review_decision"
                        ]
                        == "INCLUDE"
                    )
                ).sum()
            )
    },
    {
        "metric":
            "Articles still awaiting manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & data[
                        "manual_review_decision"
                    ].isna()
                ).sum()
            )
    },
    {
        "metric": "Unique titles",
        "value": data[TITLE_COL].nunique()
    },
    {
        "metric": "Repeated titles remaining",
        "value": int(
            data[TITLE_COL]
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Unique abstract hashes",
        "value": data[
            "content_sha256"
        ].nunique(
            dropna=True
        )
    },
    {
        "metric": "Duplicate abstract hashes remaining",
        "value": int(
            data[
                "content_sha256"
            ]
            .dropna()
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Missing dates",
        "value": int(
            data[
                "date_fixed"
            ].isna().sum()
        )
    },
    {
        "metric": "Mean abstract word count",
        "value": data[
            "abstract_word_count"
        ].mean()
    },
    {
        "metric": "Median abstract word count",
        "value": data[
            "abstract_word_count"
        ].median()
    },
    {
        "metric": "Minimum abstract word count",
        "value": data[
            "abstract_word_count"
        ].min()
    },
    {
        "metric": "Maximum abstract word count",
        "value": data[
            "abstract_word_count"
        ].max()
    }
])

overall_statistics["value"] = (
    overall_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)

display(overall_statistics)

overall_statistics.to_excel(
    OUTPUT_DIR /
    "WSJ_overall_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 25. Article counts by year
# ============================================================

year_counts = (
    data.dropna(
        subset=["year"]
    )
    .groupby("year")
    .size()
    .reset_index(name="article_count")
    .sort_values("year")
)

display(year_counts)

year_counts.to_excel(
    OUTPUT_DIR /
    "WSJ_final_article_counts_by_year.xlsx",
    index=False
)

if len(year_counts) > 0:

    plt.figure(figsize=(10, 6))

    plt.bar(
        year_counts["year"]
        .astype(int)
        .astype(str),
        year_counts["article_count"]
    )

    plt.xlabel("Year")
    plt.ylabel("Number of Articles")

    plt.title(
        "Number of WSJ Autonomous-Vehicle Articles by Year"
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_article_counts_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 26. Complete monthly article timeline
# ============================================================

valid_date_data = data[
    data["date_fixed"].notna()
].copy()

if len(valid_date_data) > 0:

    year_month_counts = (
        valid_date_data
        .groupby("year_month")
        .size()
        .reset_index(name="article_count")
        .sort_values("year_month")
    )

    full_month_range = pd.period_range(
        start=valid_date_data[
            "date_fixed"
        ].min().to_period("M"),
        end=valid_date_data[
            "date_fixed"
        ].max().to_period("M"),
        freq="M"
    )

    complete_month_table = pd.DataFrame({
        "year_month":
            full_month_range.astype(str)
    })

    complete_month_table = (
        complete_month_table
        .merge(
            year_month_counts,
            on="year_month",
            how="left"
        )
    )

    complete_month_table["article_count"] = (
        complete_month_table[
            "article_count"
        ]
        .fillna(0)
        .astype(int)
    )

    complete_month_table.to_excel(
        OUTPUT_DIR /
        "WSJ_final_complete_monthly_counts.xlsx",
        index=False
    )

    plt.figure(figsize=(16, 6))

    plt.plot(
        complete_month_table["year_month"],
        complete_month_table["article_count"],
        marker="o",
        markersize=3
    )

    label_interval = max(
        1,
        len(complete_month_table) // 24
    )

    tick_positions = list(
        range(
            0,
            len(complete_month_table),
            label_interval
        )
    )

    plt.xticks(
        ticks=tick_positions,
        labels=complete_month_table.iloc[
            tick_positions
        ]["year_month"],
        rotation=90
    )

    plt.xlabel("Year-Month")
    plt.ylabel("Number of Articles")

    plt.title(
        "Monthly Number of WSJ "
        "Autonomous-Vehicle Articles"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_monthly_article_trend.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

else:
    print(
        "\nWARNING: No valid dates were available for "
        "the monthly timeline."
    )


# %%
# ============================================================
# 27. Abstract word-count distribution
# ============================================================

if len(valid_abstracts) > 0:

    plt.figure(figsize=(10, 6))

    plt.hist(
        valid_abstracts[
            "abstract_word_count"
        ],
        bins=30,
        edgecolor="black"
    )

    abstract_mean = valid_abstracts[
        "abstract_word_count"
    ].mean()

    abstract_median = valid_abstracts[
        "abstract_word_count"
    ].median()

    plt.axvline(
        abstract_mean,
        linestyle="--",
        linewidth=2,
        label=f"Mean = {abstract_mean:.1f}"
    )

    plt.axvline(
        abstract_median,
        linestyle=":",
        linewidth=2,
        label=f"Median = {abstract_median:.1f}"
    )

    plt.xlabel("Abstract Word Count")
    plt.ylabel("Number of Articles")

    plt.title(
        "Distribution of WSJ Abstract Lengths"
    )

    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_final_abstract_word_distribution.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 28. Abstract word-count statistics by year
# ============================================================

yearly_word_count = (
    data
    .dropna(
        subset=[
            "year",
            "abstract_word_count"
        ]
    )
    .groupby("year")[
        "abstract_word_count"
    ]
    .agg(
        article_count="count",
        mean_word_count="mean",
        median_word_count="median",
        minimum_word_count="min",
        maximum_word_count="max"
    )
    .reset_index()
)

yearly_word_count[
    [
        "mean_word_count",
        "median_word_count"
    ]
] = yearly_word_count[
    [
        "mean_word_count",
        "median_word_count"
    ]
].round(2)

display(yearly_word_count)

yearly_word_count.to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_word_count_by_year.xlsx",
    index=False
)

if len(yearly_word_count) > 0:

    plt.figure(figsize=(10, 6))

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["mean_word_count"],
        marker="o",
        label="Mean"
    )

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["median_word_count"],
        marker="o",
        label="Median"
    )

    plt.xlabel("Year")
    plt.ylabel("Abstract Word Count")

    plt.title(
        "Mean and Median WSJ Abstract Word Count by Year"
    )

    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_abstract_word_count_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 29. Create combined abstract-extraction check file
# ============================================================

extraction_check = pd.concat(
    [
        data.head(100),
        remaining_metadata_problems,
        short_abstracts,
        long_abstracts,
        abstract_equals_full,
        failed_abstract_final
    ],
    ignore_index=True
)

if "content_sha256" in extraction_check.columns:

    non_empty_hash = (
        extraction_check[
            "content_sha256"
        ]
        .fillna("")
        .ne("")
    )

    extraction_check_with_hash = (
        extraction_check[
            non_empty_hash
        ]
        .drop_duplicates(
            subset=["content_sha256"],
            keep="first"
        )
    )

    extraction_check_without_hash = (
        extraction_check[
            ~non_empty_hash
        ]
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )

    extraction_check = pd.concat(
        [
            extraction_check_with_hash,
            extraction_check_without_hash
        ],
        ignore_index=True
    )

else:
    extraction_check = (
        extraction_check
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )

check_columns = [
    "date_fixed",
    TITLE_COL,
    "abstract",
    "abstract_word_count",
    "abstract_char_length",
    "abstract_extracted",
    "abstract_equals_full_content",
    "abstract_starts_with_metadata",
    CONTENT_COL
]

check_columns = [
    column
    for column in check_columns
    if column in extraction_check.columns
]

extraction_check[
    check_columns
].to_excel(
    OUTPUT_DIR /
    "WSJ_abstract_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 30. Remove internal helper columns from final output
# ============================================================

temporary_columns_to_remove = [
    "_original_order",
    "_content_dedup_key",
    "_manual_review_match_key"
]

data = data.drop(
    columns=[
        column
        for column in temporary_columns_to_remove
        if column in data.columns
    ]
)


# %%
# ============================================================
# 31. Save final analytical dataset
# ============================================================

data.to_excel(
    FINAL_DATA_PATH,
    index=False
)


# %%
# ============================================================
# 32. Final verification
# ============================================================

print("\n" + "=" * 75)
print("WSJ CLEANING COMPLETED")
print("=" * 75)

print("\nOriginal articles:")
print(n_original)

print(
    "\nRemoved because of invalid dates "
    "or failed abstract extraction:"
)
print(n_removed_initial_cleaning)

print("\nRemoved by automatic AV scope screening:")
print(n_removed_automatic_scope)

print("\nSent for manual scope review:")
print(n_before_manual_review)

print("\nIncluded after manual scope review:")
print(n_manual_review_included)

print("\nExcluded after manual scope review:")
print(n_manual_review_excluded)

print(
    "\nRetained after automatic and manual "
    "scope screening, before deduplication:"
)
print(n_after_scope_screening)

print("\nRemoved by abstract SHA-256 deduplication:")
print(n_removed_content_duplicates)

print("\nFinal analytical articles:")
print(len(data))

final_manually_included_count = int(
    (
        data["requires_manual_scope_review"]
        & (
            data["manual_review_decision"]
            == "INCLUDE"
        )
    ).sum()
)

final_pending_manual_review_count = int(
    (
        data["requires_manual_scope_review"]
        & data[
            "manual_review_decision"
        ].isna()
    ).sum()
)

print("\nFinal articles included after manual scope review:")
print(final_manually_included_count)

print("\nFinal articles still awaiting manual review:")
print(final_pending_manual_review_count)

print("\nMissing dates in final dataset:")
print(
    data["date_fixed"].isna().sum()
)

print("\nDuplicate abstract hashes remaining:")
print(
    data[
        "content_sha256"
    ]
    .dropna()
    .duplicated()
    .sum()
)

print("\nSuccessfully extracted abstracts:")
print(
    int(
        data[
            "abstract_extracted"
        ].sum()
    )
)

print("\nFailed abstract extractions:")
print(len(failed_abstract_final))

print("\nAbstracts under 10 words:")
print(len(short_abstracts))

print("\nAbstracts over 200 words:")
print(len(long_abstracts))

print("\nRemaining possible abstract metadata problems:")
print(len(remaining_metadata_problems))

print("\nMean abstract word count:")
print(
    round(
        valid_abstracts[
            "abstract_word_count"
        ].mean(),
        2
    )
)

print("\nMedian abstract word count:")
print(
    round(
        valid_abstracts[
            "abstract_word_count"
        ].median(),
        2
    )
)

print("\nFinal analytical dataset saved to:")
print(FINAL_DATA_PATH)

print("\nManual scope-review input file:")
print(
    OUTPUT_DIR /
    "WSJ_articles_requiring_manual_scope_review.xlsx"
)

print("\nAutomatically excluded scope file:")
print(
    OUTPUT_DIR /
    "WSJ_articles_excluded_by_AV_scope.xlsx"
)

print("\nAll audit and descriptive files saved in:")
print(OUTPUT_DIR)
```

```python
# %%
# ============================================================
# Extract WSJ abstracts and calculate word-count statistics
#
# Extract text between:
# "ABSTRACT" and "End of Document"
#
# Input:
# WSJ_dataset_after_invalid_date_removal.xlsx
#
# Outputs:
# 1. Dataset with abstract and abstract_word_count
# 2. Abstract extraction summary
# 3. Failed abstract extraction records
# 4. Word-count descriptive statistics
# 5. Word-count distribution figure
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import re
from pathlib import Path


# ------------------------------------------------------------
# 1. File paths
# ------------------------------------------------------------

INPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_WSJ/"
    "WSJ_dataset_after_invalid_date_removal.xlsx"
)

OUTPUT_DIR = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_WSJ"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 2. Load data
# ------------------------------------------------------------

df = pd.read_excel(INPUT_PATH)

print("Data loaded successfully.")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# ------------------------------------------------------------
# 3. Check content column
# ------------------------------------------------------------

CONTENT_COL = "content"

if CONTENT_COL not in df.columns:
    raise KeyError(
        f"Column '{CONTENT_COL}' was not found. "
        f"Available columns are: {df.columns.tolist()}"
    )

print("\nMissing content values:")
print(df[CONTENT_COL].isna().sum())


# ------------------------------------------------------------
# 4. Define abstract extraction function
# ------------------------------------------------------------

def extract_abstract(text):
    """
    Extract text located between 'ABSTRACT' and 'End of Document'.

    The function:
    - ignores uppercase/lowercase differences;
    - allows spaces around ABSTRACT;
    - handles line breaks;
    - returns pd.NA when no valid section is found.
    """

    # Return missing value when content is empty
    if pd.isna(text):
        return pd.NA

    text = str(text)

    # Extract everything between ABSTRACT and End of Document
    match = re.search(
        r"\bABSTRACT\b\s*:?\s*(.*?)\s*\bEnd\s+of\s+Document\b",
        text,
        flags=re.IGNORECASE | re.DOTALL
    )

    if match is None:
        return pd.NA

    abstract = match.group(1)

    # Replace line breaks and tabs with spaces
    abstract = re.sub(
        r"[\r\n\t]+",
        " ",
        abstract
    )

    # Remove repeated spaces
    abstract = re.sub(
        r"\s+",
        " ",
        abstract
    )

    # Remove unnecessary punctuation or spaces at beginning/end
    abstract = abstract.strip(
        " \n\r\t:-"
    )

    # Treat empty extraction as missing
    if abstract == "":
        return pd.NA

    return abstract


# ------------------------------------------------------------
# 5. Extract abstracts
# ------------------------------------------------------------

df["abstract"] = df[CONTENT_COL].apply(
    extract_abstract
)

print("\nAbstract extraction completed.")


# ------------------------------------------------------------
# 6. Check extraction results
# ------------------------------------------------------------

n_total = len(df)

n_abstract_success = (
    df["abstract"].notna().sum()
)

n_abstract_failed = (
    df["abstract"].isna().sum()
)

abstract_success_rate = (
    n_abstract_success / n_total * 100
    if n_total > 0
    else 0
)

print("\nAbstract extraction results:")
print("Total articles:", n_total)
print("Successfully extracted abstracts:", n_abstract_success)
print("Failed abstract extractions:", n_abstract_failed)
print(
    f"Extraction success rate: "
    f"{abstract_success_rate:.2f}%"
)


# ------------------------------------------------------------
# 7. Display extracted abstract examples
# ------------------------------------------------------------

example_columns = [
    col for col in
    [
        "file",
        "title",
        "date",
        "abstract"
    ]
    if col in df.columns
]

print("\nExtracted abstract examples:")

display(
    df.loc[
        df["abstract"].notna(),
        example_columns
    ].head(10)
)


# ------------------------------------------------------------
# 8. Create word-count function
# ------------------------------------------------------------

def count_words(text):
    """
    Count English words and word-like tokens.

    Examples counted as one token:
    - self-driving
    - company's
    - U.S.
    - 2023
    """

    if pd.isna(text):
        return pd.NA

    text = str(text).strip()

    if text == "":
        return pd.NA

    # Count words, numbers, apostrophe words and hyphenated words
    words = re.findall(
        r"\b[\w]+(?:[-'][\w]+)*\b",
        text,
        flags=re.UNICODE
    )

    return len(words)


# ------------------------------------------------------------
# 9. Calculate abstract word count
# ------------------------------------------------------------

df["abstract_word_count"] = (
    df["abstract"]
    .apply(count_words)
    .astype("Int64")
)

print("\nAbstract word count created.")

display(
    df[
        [
            col for col in
            [
                "title",
                "abstract",
                "abstract_word_count"
            ]
            if col in df.columns
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 10. Identify failed abstract extraction records
# ------------------------------------------------------------

failed_abstract_extraction = (
    df[df["abstract"].isna()]
    .copy()
)

failed_display_columns = [
    col for col in
    [
        "file",
        "title",
        "source",
        "date",
        "content"
    ]
    if col in failed_abstract_extraction.columns
]

if len(failed_abstract_extraction) > 0:

    print("\nFailed abstract extraction examples:")

    display(
        failed_abstract_extraction[
            failed_display_columns
        ].head(20)
    )

    failed_abstract_extraction.to_excel(
        OUTPUT_DIR /
        "WSJ_failed_abstract_extraction.xlsx",
        index=False
    )

    print(
        "\nFailed extraction records saved to:",
        OUTPUT_DIR /
        "WSJ_failed_abstract_extraction.xlsx"
    )

else:

    print(
        "\nAll articles had an abstract section."
    )


# ------------------------------------------------------------
# 11. Check zero-word or extremely short abstracts
# ------------------------------------------------------------

short_abstracts = df[
    df["abstract_word_count"].notna()
    & (df["abstract_word_count"] < 10)
].copy()

print("\nAbstracts with fewer than 10 words:")
print(len(short_abstracts))

if len(short_abstracts) > 0:

    display(
        short_abstracts[
            [
                col for col in
                [
                    "file",
                    "title",
                    "abstract",
                    "abstract_word_count"
                ]
                if col in short_abstracts.columns
            ]
        ].head(20)
    )

    short_abstracts.to_excel(
        OUTPUT_DIR /
        "WSJ_abstracts_under_10_words.xlsx",
        index=False
    )


# ------------------------------------------------------------
# 12. Calculate descriptive statistics
# Only use successfully extracted abstracts
# ------------------------------------------------------------

valid_word_counts = (
    df["abstract_word_count"]
    .dropna()
    .astype(float)
)

if len(valid_word_counts) == 0:
    raise ValueError(
        "No valid abstract word counts were found. "
        "Please check the ABSTRACT and End of Document markers."
    )


# Basic statistics
word_count_statistics = pd.DataFrame(
    {
        "statistic": [
            "Total dataset articles",
            "Successfully extracted abstracts",
            "Failed abstract extractions",
            "Extraction success rate (%)",
            "Count",
            "Mean",
            "Median",
            "Standard deviation",
            "Variance",
            "Minimum",
            "25th percentile",
            "50th percentile",
            "75th percentile",
            "Maximum",
            "Interquartile range",
            "Range",
            "Skewness"
        ],

        "value": [
            n_total,
            n_abstract_success,
            n_abstract_failed,
            abstract_success_rate,
            valid_word_counts.count(),
            valid_word_counts.mean(),
            valid_word_counts.median(),
            valid_word_counts.std(),
            valid_word_counts.var(),
            valid_word_counts.min(),
            valid_word_counts.quantile(0.25),
            valid_word_counts.quantile(0.50),
            valid_word_counts.quantile(0.75),
            valid_word_counts.max(),
            (
                valid_word_counts.quantile(0.75)
                - valid_word_counts.quantile(0.25)
            ),
            (
                valid_word_counts.max()
                - valid_word_counts.min()
            ),
            valid_word_counts.skew()
        ]
    }
)

# Round statistical values
word_count_statistics["value"] = (
    pd.to_numeric(
        word_count_statistics["value"],
        errors="coerce"
    )
    .round(2)
)

print("\nAbstract word-count descriptive statistics:")

display(word_count_statistics)


# ------------------------------------------------------------
# 13. Generate standard pandas describe table
# ------------------------------------------------------------

word_count_describe = (
    valid_word_counts
    .describe(
        percentiles=[
            0.10,
            0.25,
            0.50,
            0.75,
            0.90
        ]
    )
    .reset_index()
)

word_count_describe.columns = [
    "statistic",
    "abstract_word_count"
]

word_count_describe[
    "abstract_word_count"
] = word_count_describe[
    "abstract_word_count"
].round(2)

print("\nDetailed describe table:")

display(word_count_describe)


# ------------------------------------------------------------
# 14. Create extraction summary
# ------------------------------------------------------------

extraction_summary = pd.DataFrame(
    [
        {
            "measure": "Total articles",
            "value": n_total
        },
        {
            "measure": "Successfully extracted abstracts",
            "value": n_abstract_success
        },
        {
            "measure": "Failed abstract extractions",
            "value": n_abstract_failed
        },
        {
            "measure": "Extraction success rate (%)",
            "value": round(
                abstract_success_rate,
                2
            )
        },
        {
            "measure": "Mean abstract word count",
            "value": round(
                valid_word_counts.mean(),
                2
            )
        },
        {
            "measure": "Median abstract word count",
            "value": round(
                valid_word_counts.median(),
                2
            )
        },
        {
            "measure": "Minimum abstract word count",
            "value": round(
                valid_word_counts.min(),
                2
            )
        },
        {
            "measure": "Maximum abstract word count",
            "value": round(
                valid_word_counts.max(),
                2
            )
        }
    ]
)

print("\nExtraction summary:")

display(extraction_summary)


# ------------------------------------------------------------
# 15. Save dataset with abstract and word count
# ------------------------------------------------------------

FINAL_DATA_PATH = (
    OUTPUT_DIR /
    "WSJ_dataset_with_abstract_and_word_count.xlsx"
)

df.to_excel(
    FINAL_DATA_PATH,
    index=False
)

print("\nFinal dataset saved successfully:")
print(FINAL_DATA_PATH)


# ------------------------------------------------------------
# 16. Save descriptive statistics
# ------------------------------------------------------------

STATISTICS_PATH = (
    OUTPUT_DIR /
    "WSJ_abstract_word_count_statistics.xlsx"
)

with pd.ExcelWriter(
    STATISTICS_PATH,
    engine="openpyxl"
) as writer:

    extraction_summary.to_excel(
        writer,
        sheet_name="Extraction Summary",
        index=False
    )

    word_count_statistics.to_excel(
        writer,
        sheet_name="Word Count Statistics",
        index=False
    )

    word_count_describe.to_excel(
        writer,
        sheet_name="Detailed Describe",
        index=False
    )

print("\nStatistics saved successfully:")
print(STATISTICS_PATH)


# ------------------------------------------------------------
# 17. Plot abstract word-count histogram
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    valid_word_counts,
    bins=20,
    edgecolor="black"
)

plt.axvline(
    valid_word_counts.mean(),
    linestyle="--",
    linewidth=2,
    label=(
        f"Mean = "
        f"{valid_word_counts.mean():.1f}"
    )
)

plt.axvline(
    valid_word_counts.median(),
    linestyle=":",
    linewidth=2,
    label=(
        f"Median = "
        f"{valid_word_counts.median():.1f}"
    )
)

plt.xlabel("Abstract Word Count")
plt.ylabel("Number of Articles")
plt.title(
    "Distribution of WSJ Abstract Word Counts"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR /
    "WSJ_abstract_word_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ------------------------------------------------------------
# 18. Plot abstract word count by year
# Optional descriptive figure
# ------------------------------------------------------------

if "year" in df.columns:

    yearly_word_count = (
        df
        .dropna(
            subset=[
                "year",
                "abstract_word_count"
            ]
        )
        .groupby("year")[
            "abstract_word_count"
        ]
        .agg(
            article_count="count",
            mean_word_count="mean",
            median_word_count="median",
            minimum_word_count="min",
            maximum_word_count="max"
        )
        .reset_index()
    )

    yearly_word_count[
        [
            "mean_word_count",
            "median_word_count"
        ]
    ] = yearly_word_count[
        [
            "mean_word_count",
            "median_word_count"
        ]
    ].round(2)

    print(
        "\nAbstract word-count statistics by year:"
    )

    display(yearly_word_count)

    yearly_word_count.to_excel(
        OUTPUT_DIR /
        "WSJ_abstract_word_count_by_year.xlsx",
        index=False
    )

    plt.figure(figsize=(10, 6))

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["mean_word_count"],
        marker="o",
        label="Mean"
    )

    plt.plot(
        yearly_word_count["year"].astype(str),
        yearly_word_count["median_word_count"],
        marker="o",
        label="Median"
    )

    plt.xlabel("Year")
    plt.ylabel("Abstract Word Count")
    plt.title(
        "Mean and Median WSJ Abstract Word Count by Year"
    )

    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "WSJ_abstract_word_count_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ------------------------------------------------------------
# 19. Final validation
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("WSJ abstract extraction completed.")
print("=" * 65)

print("Total articles:", n_total)
print(
    "Successfully extracted abstracts:",
    n_abstract_success
)
print(
    "Failed abstract extractions:",
    n_abstract_failed
)
print(
    f"Success rate: "
    f"{abstract_success_rate:.2f}%"
)

print(
    f"Mean word count: "
    f"{valid_word_counts.mean():.2f}"
)

print(
    f"Median word count: "
    f"{valid_word_counts.median():.2f}"
)

print(
    f"Minimum word count: "
    f"{valid_word_counts.min():.0f}"
)

print(
    f"Maximum word count: "
    f"{valid_word_counts.max():.0f}"
)

print("\nFinal dataset:")
print(FINAL_DATA_PATH)

print("\nStatistics file:")
print(STATISTICS_PATH)

print("\nOutput folder:")
print(OUTPUT_DIR)
```
