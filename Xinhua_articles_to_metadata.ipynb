```python
from docx import Document
import pandas as pd
import os

folder = "/Users/yurujia/Desktop/Dissertation Data/China/1-200"

data = []

files = os.listdir(folder)

docx_files = [
    f for f in files
    if f.lower().endswith(".docx") and not f.startswith("~$")
]

print("Found docx files:", len(docx_files))
print(docx_files[:10])

for file in docx_files:
    path = os.path.join(folder, file)

    doc = Document(path)

    paragraphs = []
    for p in doc.paragraphs:
        text = p.text.strip()
        if text:
            paragraphs.append(text)

    full_text = "\n".join(paragraphs)

    title = paragraphs[0] if len(paragraphs) > 0 else None
    date = paragraphs[1] if len(paragraphs) > 1 else None
    source = paragraphs[2] if len(paragraphs) > 2 else None

    data.append({
        "file": file,
        "title": title,
        "date": date,
        "source": source,
        "content": full_text
    })

df = pd.DataFrame(data)

print("Final dataframe shape:", df.shape)
print(df.head())

output_path = os.path.join(folder, "lexis_structured1-200.xlsx")
df.to_excel(output_path, index=False)

print("Saved to:", output_path)
```

```python
import pandas as pd
import re
import os


folder = "/Users/yurujia/Desktop/Dissertation Data/China/excel"

input_path = os.path.join(folder, "merged_output.xlsx")
output_path = os.path.join(folder, "merged_output_with_clean_content.xlsx")
deleted_path = os.path.join(folder, "merged_output_deleted_no_body_rows.xlsx")
problem_path = os.path.join(folder, "merged_output_problem_rows.xlsx")


df = pd.read_excel(input_path)

print("Original shape:", df.shape)
print("Columns:", df.columns.tolist())

if "content" not in df.columns:
    raise ValueError("没有找到 content 列，请检查你的 Excel 列名。")


df["has_body_marker"] = df["content"].astype(str).str.contains(
    r"(?im)^\s*Body\s*$",
    regex=True,
    na=False
)

df["has_wan_marker"] = df["content"].astype(str).str.contains(
    r"[（(]\s*完\s*[）)]",
    regex=True,
    na=False
)

df["has_load_date_marker"] = df["content"].astype(str).str.contains(
    r"(?im)^\s*Load-Date\s*:",
    regex=True,
    na=False
)

print("\nBefore deleting rows without Body:")
print(df["has_body_marker"].value_counts(dropna=False))


deleted_rows = df[df["has_body_marker"] == False].copy()
deleted_rows.to_excel(deleted_path, index=False)

df = df[df["has_body_marker"] == True].copy()

print("\nAfter deleting rows without Body:")
print("Remaining shape:", df.shape)
print("Deleted rows:", len(deleted_rows))


def extract_article_content(text):
    if pd.isna(text):
        return None, "missing_content"

    text = str(text)

    body_match = re.search(
        r"(?im)^\s*Body\s*$",
        text
    )

    if not body_match:
        return None, "no_body_deleted_should_not_happen"

    after_body = text[body_match.end():].strip()

    wan_match = re.search(
        r"[（(]\s*完\s*[）)]",
        after_body
    )

    if wan_match:
        clean_content = after_body[:wan_match.start()].strip()
        return clean_content, "body_to_wan"

    load_date_match = re.search(
        r"(?im)^\s*Load-Date\s*:",
        after_body
    )

    if load_date_match:
        clean_content = after_body[:load_date_match.start()].strip()
        return clean_content, "body_to_load_date"

    # 规则 3：如果没有（完），也没有 Load-Date，直接提取到最后
    clean_content = after_body.strip()
    return clean_content, "body_to_end"


results = df["content"].apply(extract_article_content)

df["article_content_clean"] = results.apply(lambda x: x[0])
df["extraction_method"] = results.apply(lambda x: x[1])
df["extract_success"] = df["article_content_clean"].notna() & (df["article_content_clean"].str.len() > 0)

print("\nHas Body marker:")
print(df["has_body_marker"].value_counts(dropna=False))

print("\nHas 完 marker:")
print(df["has_wan_marker"].value_counts(dropna=False))

print("\nHas Load-Date marker:")
print(df["has_load_date_marker"].value_counts(dropna=False))

print("\nExtraction method:")
print(df["extraction_method"].value_counts(dropna=False))

print("\nExtraction success:")
print(df["extract_success"].value_counts(dropna=False))

print("\nAll remaining rows extracted successfully:", df["extract_success"].all())


df.to_excel(output_path, index=False)

problem_rows = df[df["extract_success"] == False].copy()
problem_rows.to_excel(problem_path, index=False)

print("\nSaved cleaned Excel to:")
print(output_path)

print("\nSaved deleted rows without Body to:")
print(deleted_path)

print("\nSaved problem rows to:")
print(problem_path)

print("\nProblem rows count:", len(problem_rows))
```

```python
import pandas as pd
import os

folder = "/Users/yurujia/Desktop/Dissertation Data/China/excel"

input_path = os.path.join(folder, "merged_output_with_clean_content.xlsx")
output_path = os.path.join(folder, "manual_sentiment_sample_220.xlsx")


df = pd.read_excel(input_path)

print("Original shape:", df.shape)
print("Columns:", df.columns.tolist())

text_col = "article_content_clean"

if text_col not in df.columns:
    raise ValueError(f"没有找到 {text_col} 这一列，请检查列名。")

df = df[df[text_col].notna()].copy()
df = df[df[text_col].astype(str).str.strip() != ""].copy()

print("After removing empty content:", df.shape)


N_SAMPLE = 220

sample_df = df.sample(
    n=N_SAMPLE,
    random_state=42
).copy()



sample_df["manual_sentiment"] = ""
sample_df["manual_sentiment_note"] = ""

# 可选：加一个短文本预览，方便你快速看
sample_df["content_preview"] = sample_df[text_col].astype(str).str[:300]



preferred_cols = [
    "manual_sentiment",
    "manual_sentiment_note",
    "content_preview",
    text_col,
    "title",
    "date",
    "source",
    "file"
]

existing_preferred_cols = [col for col in preferred_cols if col in sample_df.columns]
other_cols = [col for col in sample_df.columns if col not in existing_preferred_cols]

sample_df = sample_df[existing_preferred_cols + other_cols]


sample_df.to_excel(output_path, index=False)

print("Sample saved to:")
print(output_path)
print("Sample shape:", sample_df.shape)
```
