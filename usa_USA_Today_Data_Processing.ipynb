```python
# %%
# ============================================================
# USA Today AV News Data Cleaning, Scope Screening,
# Content-Hash Deduplication, and First-Paragraph Extraction
#
# Moderately relaxed screening version:
#
# 1. Load and inspect the original LexisNexis dataset
# 2. Standardise publication dates
# 3. Extract article body from Body to End of Document
# 4. Remove unusable records
# 5. Apply automated AV scope screening
# 6. Retain ambiguous ADAS/company-context cases for review
# 7. Deduplicate using normalised full-text SHA-256 hashes
# 8. Extract the first substantive paragraph
# 9. Produce audit files and descriptive statistics
# 10. Save the final analytical dataset
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import hashlib
import unicodedata


# %%
# ============================================================
# 1. File paths
# ============================================================

INPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/excel/"
    "lexis_structured_USA_Today_1-442.xlsx"
)

OUTPUT_DIR = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_USA_Today"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FINAL_DATA_PATH = (
    OUTPUT_DIR /
    "USA_Today_AV_final_analytical_dataset.xlsx"
)

FINAL_WITH_LEAD_PATH = (
    OUTPUT_DIR /
    "USA_Today_AV_final_with_first_paragraph.xlsx"
)

# Completed manual scope-review decisions
MANUAL_REVIEW_COMPLETED_PATH = (
    OUTPUT_DIR /
    "USA_Today_manual_scope_review_completed.xlsx"
)


# %%
# ============================================================
# 2. Load data
# ============================================================

df = pd.read_excel(INPUT_PATH)

n_original = len(df)

print("=" * 75)
print("USA TODAY AV DATA CLEANING")
print("=" * 75)

print("\nData loaded successfully.")
print("Original shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# %%
# ============================================================
# 3. Define key columns
# ============================================================

DATE_COL = "date"
TITLE_COL = "title"
CONTENT_COL = "content"
SOURCE_COL = "source"

required_columns = [
    DATE_COL,
    TITLE_COL,
    CONTENT_COL
]

missing_required_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_required_columns:

    raise KeyError(
        "The following required columns are missing: "
        f"{missing_required_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )


if SOURCE_COL not in df.columns:

    print(
        f"\nWARNING: '{SOURCE_COL}' column was not found. "
        "A default source value will be created."
    )

    df[SOURCE_COL] = "USA Today"


# Preserve original row order for reproducible tie-breaking
df["_original_order"] = range(len(df))


# Ensure text columns are strings
df[TITLE_COL] = (
    df[TITLE_COL]
    .fillna("")
    .astype(str)
)

df[CONTENT_COL] = (
    df[CONTENT_COL]
    .fillna("")
    .astype(str)
)


# %%
# ============================================================
# 4. Missing-value and column summary
# ============================================================

column_summary = pd.DataFrame({
    "column": df.columns,
    "non_null_count": df.notna().sum().values,
    "missing_count": df.isna().sum().values,
    "missing_percent": (
        df.isna().mean().values * 100
    ).round(2),
    "dtype": df.dtypes.astype(str).values
})

display(column_summary)

column_summary.to_excel(
    OUTPUT_DIR /
    "USA_Today_column_missing_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 5. Standardise publication dates
#
# Example:
# December 14, 2023 Thursday
# -> 2023-12-14
# ============================================================

MONTH_PATTERN = (
    r"(?:January|February|March|April|May|June|"
    r"July|August|September|October|November|December)"
)

DATE_PATTERN = (
    rf"{MONTH_PATTERN}\s+"
    r"\d{1,2},\s+\d{4}"
)


def extract_english_date(value):
    """
    Extract an English-language date from a messy
    LexisNexis-style date string.
    """

    if pd.isna(value):
        return pd.NaT

    text = str(value).strip()

    match = re.search(
        DATE_PATTERN,
        text,
        flags=re.IGNORECASE
    )

    if match is None:
        return pd.NaT

    return pd.to_datetime(
        match.group(0),
        format="%B %d, %Y",
        errors="coerce"
    )


df["date_original"] = df[DATE_COL]

df["date_parsed_direct"] = pd.to_datetime(
    df[DATE_COL],
    errors="coerce"
)

df["date_fixed"] = (
    df[DATE_COL]
    .apply(extract_english_date)
)


date_summary = pd.DataFrame([
    {
        "metric": "Total original articles",
        "value": len(df)
    },
    {
        "metric": "Dates parsed directly by pandas",
        "value": int(
            df["date_parsed_direct"].notna().sum()
        )
    },
    {
        "metric": "Dates recovered using date extraction",
        "value": int(
            df["date_fixed"].notna().sum()
        )
    },
    {
        "metric": "Dates still missing after extraction",
        "value": int(
            df["date_fixed"].isna().sum()
        )
    }
])

print("\nDate parsing summary:")
display(date_summary)

date_summary.to_excel(
    OUTPUT_DIR /
    "USA_Today_date_parsing_summary.xlsx",
    index=False
)


failed_date_rows = df[
    df["date_fixed"].isna()
].copy()

failed_date_rows.to_excel(
    OUTPUT_DIR /
    "USA_Today_invalid_date_rows_check.xlsx",
    index=False
)


# %%
# ============================================================
# 6. Normalise line breaks
# ============================================================

def normalize_line_breaks(text):
    """
    Standardise line breaks while preserving paragraph structure.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Convert literal newline sequences only when no real
    # newline characters exist
    if "\\n" in text and "\n" not in text:
        text = text.replace("\\n", "\n")

    text = re.sub(
        r"[ \t]+\n",
        "\n",
        text
    )

    text = re.sub(
        r"\n[ \t]+",
        "\n",
        text
    )

    text = re.sub(
        r"\n{3,}",
        "\n\n",
        text
    )

    return text.strip()


# %%
# ============================================================
# 7. Extract full article body
#
# Extract text between:
# Body
# ...
# End of Document
# ============================================================

def extract_full_body(text):
    """
    Extract the full article body from a LexisNexis record.
    """

    text = normalize_line_breaks(text)

    if text == "":
        return ""

    # Preferred pattern:
    # Body and End of Document occur as standalone lines
    main_pattern = re.compile(
        r"(?ims)"
        r"^[ \t]*Body[ \t]*\n"
        r"(.*?)"
        r"(?=^[ \t]*End\s+of\s+Document[ \t]*$)"
    )

    match = main_pattern.search(text)

    # Fallback for unusual line formatting
    if match is None:

        fallback_pattern = re.compile(
            r"(?is)"
            r"(?:^|\n|\s)Body\s*"
            r"(.*?)"
            r"\s*End\s+of\s+Document"
        )

        match = fallback_pattern.search(text)

    if match is None:
        return ""

    body = match.group(1).strip()

    body = re.sub(
        r"[ \t]+\n",
        "\n",
        body
    )

    body = re.sub(
        r"\n[ \t]+",
        "\n",
        body
    )

    body = re.sub(
        r"\n{3,}",
        "\n\n",
        body
    )

    return body.strip()


df["article_body"] = (
    df[CONTENT_COL]
    .apply(extract_full_body)
)

df["full_text_extracted"] = (
    df["article_body"]
    .fillna("")
    .str
    .strip()
    .ne("")
)


print("\nFull-body extraction status:")

print(
    df["full_text_extracted"]
    .value_counts(dropna=False)
)


failed_body_rows = df[
    ~df["full_text_extracted"]
].copy()

failed_body_rows.to_excel(
    OUTPUT_DIR /
    "USA_Today_failed_full_body_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 8. Basic text measures
# ============================================================

def count_english_words(text):
    """
    Approximate English word count.

    Expressions such as:
    self-driving
    company's
    COVID-19
    U.S.
    are generally treated as single word units.
    """

    if pd.isna(text):
        return 0

    text = str(text).strip()

    if text == "":
        return 0

    words = re.findall(
        r"\b[A-Za-z0-9]+"
        r"(?:['’.-][A-Za-z0-9]+)*\b",
        text
    )

    return len(words)


df["content_char_length"] = (
    df["article_body"]
    .fillna("")
    .str
    .len()
)

df["full_text_word_count"] = (
    df["article_body"]
    .apply(count_english_words)
)


# %%
# ============================================================
# 9. Initial usability exclusions
#
# Exclude:
# - missing or unrecoverable publication dates
# - failed or empty article-body extraction
#
# Very long articles are audited but not automatically removed.
# ============================================================

df["exclude_invalid_date"] = (
    df["date_fixed"].isna()
)

df["exclude_missing_article_body"] = (
    ~df["full_text_extracted"]
)

df["exclude_initial_unusable_record"] = (
    df["exclude_invalid_date"]
    | df["exclude_missing_article_body"]
)


def create_initial_exclusion_reason(row):

    reasons = []

    if row["exclude_invalid_date"]:
        reasons.append(
            "Missing or invalid publication date"
        )

    if row["exclude_missing_article_body"]:
        reasons.append(
            "Article body could not be extracted"
        )

    return "; ".join(reasons)


df["initial_exclusion_reason"] = (
    df.apply(
        create_initial_exclusion_reason,
        axis=1
    )
)


initial_exclusion_summary = pd.DataFrame([
    {
        "rule": "Missing or invalid publication date",
        "article_count": int(
            df["exclude_invalid_date"].sum()
        )
    },
    {
        "rule": "Article body could not be extracted",
        "article_count": int(
            df["exclude_missing_article_body"].sum()
        )
    },
    {
        "rule": "Excluded by one or more initial rules",
        "article_count": int(
            df[
                "exclude_initial_unusable_record"
            ].sum()
        )
    }
])

initial_exclusion_summary["percent_of_original"] = (
    initial_exclusion_summary["article_count"]
    / len(df)
    * 100
).round(2)

display(initial_exclusion_summary)

initial_exclusion_summary.to_excel(
    OUTPUT_DIR /
    "USA_Today_initial_exclusion_summary.xlsx",
    index=False
)


initially_excluded = df[
    df["exclude_initial_unusable_record"]
].copy()

initially_excluded.to_excel(
    OUTPUT_DIR /
    "USA_Today_initially_excluded_records_check.xlsx",
    index=False
)


usable_data = df[
    ~df["exclude_initial_unusable_record"]
].copy()


n_after_initial_cleaning = len(usable_data)

n_removed_initial_cleaning = (
    len(df) - len(usable_data)
)


# Audit long articles without excluding them
very_long_articles = usable_data[
    usable_data["full_text_word_count"] > 2500
].copy()

very_long_articles.to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_over_2500_words_check.xlsx",
    index=False
)

print(
    "\nArticles longer than 2,500 words, retained for audit:",
    len(very_long_articles)
)


# %%
# ============================================================
# 10. Create automated AV-screening text
#
# Use title plus first 1,500 characters of article body.
# ============================================================

df_scope = usable_data.copy()

df_scope["automated_screen_text"] = (
    df_scope[TITLE_COL]
    + " "
    + df_scope["article_body"]
    .fillna("")
    .str[:1500]
)


def contains_any_pattern(text, patterns):
    """
    Return True if at least one regex pattern appears in text.
    """

    if pd.isna(text):
        return False

    text = str(text)

    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        is not None
        for pattern in patterns
    )


# %%
# ============================================================
# 11. Define English AV-scope patterns
# ============================================================

# ------------------------------------------------------------
# A. General road autonomous-vehicle terms
# ------------------------------------------------------------

general_road_av_patterns = [
    # Autonomous vehicles and cars
    r"\bautonomous\s+vehicles?\b",
    r"\bautonomous\s+cars?\b",
    r"\bautomated\s+vehicles?\b",
    r"\bautomated\s+cars?\b",

    # Autonomous driving
    r"\bautonomous\s+driving\b",
    r"\bautomated\s+driving\b",
    r"\bautomated\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+systems?\b",
    r"\bautonomous\s+driving\s+technology\b",
    r"\bself[-\s]?driving\s+technology\b",

    # Self-driving expressions
    r"\bself[-\s]?driving\b",
    r"\bself[-\s]?driving\s+vehicles?\b",
    r"\bself[-\s]?driving\s+cars?\b",

    # Driverless expressions
    r"\bdriverless\b",
    r"\bdriverless\s+vehicles?\b",
    r"\bdriverless\s+cars?\b",

    # Robotaxi expressions
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bself[-\s]?driving\s+taxis?\b",

    # Broader expressions
    r"\bvehicle\s+automation\b",
    r"\bautomotive\s+automation\b",
    r"\bautomated\s+mobility\b",
    r"\bautonomous\s+mobility\b",
    r"\bfull\s+self[-\s]?driving\b",
]


# ------------------------------------------------------------
# B. Explicit higher-level automation indicators
# ------------------------------------------------------------

higher_automation_patterns = [
    r"\bSAE\s+(?:Level\s*)?[345]\b",
    r"\bLevel[-\s]?[345]\b",
    r"\bL[345]\b",
    r"\bfully\s+autonomous\b",
    r"\bfull\s+autonomy\b",
    r"\bhighly\s+automated\b",
    r"\bhigh[-\s]?automation\b",
    r"\bdriverless\b",
    r"\bwithout\s+(?:a\s+)?safety\s+driver\b",
    r"\bno\s+safety\s+driver\b",
    r"\bremove(?:d|s|ing)?\s+(?:the\s+)?safety\s+driver\b",
    r"\bcommercial\s+autonomous\s+service\b",
    r"\bcommercial\s+driverless\s+service\b",
    r"\bcommercial\s+robotaxi\s+service\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+ride[-\s]?hailing\b",
]


# ------------------------------------------------------------
# C. Passenger-vehicle and robotaxi indicators
# ------------------------------------------------------------

passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bpersonal\s+vehicles?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\bminivans?\b",
    r"\btaxis?\b",
    r"\bcabs?\b",
    r"\bride[-\s]?hailing\b",
    r"\bride[-\s]?sharing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]


strong_passenger_robotaxi_patterns = [
    r"\bpassenger\s+vehicles?\b",
    r"\bpassenger\s+cars?\b",
    r"\bprivate\s+cars?\b",
    r"\bsedans?\b",
    r"\bSUVs?\b",
    r"\btaxis?\b",
    r"\bride[-\s]?hailing\b",
    r"\brobo[-\s]?taxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
]


# ------------------------------------------------------------
# D. Bus, shuttle and public-transport indicators
# ------------------------------------------------------------

bus_public_transport_patterns = [
    r"\bautonomous\s+buses?\b",
    r"\bdriverless\s+buses?\b",
    r"\bself[-\s]?driving\s+buses?\b",
    r"\bautomated\s+buses?\b",
    r"\bautonomous\s+shuttles?\b",
    r"\bdriverless\s+shuttles?\b",
    r"\bself[-\s]?driving\s+shuttles?\b",
    r"\bautomated\s+shuttles?\b",
    r"\bautonomous\s+minibuses?\b",
    r"\bdriverless\s+minibuses?\b",
    r"\bpublic\s+transit\b",
    r"\bpublic\s+transportation\b",
    r"\btransit\s+service\b",
    r"\bpeople\s+movers?\b",
]


# ------------------------------------------------------------
# E. ADAS and driver-assistance indicators
# ------------------------------------------------------------

adas_patterns = [
    r"\bADAS\b",
    r"\badvanced\s+driver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assistance\s+systems?\b",
    r"\bdriver[-\s]?assist(?:ance)?\b",
    r"\bautomated\s+parking\b",
    r"\bautomatic\s+parking\b",
    r"\bself[-\s]?parking\b",
    r"\badaptive\s+cruise\s+control\b",
    r"\blane[-\s]?keeping\b",
    r"\blane[-\s]?centering\b",
    r"\blane[-\s]?departure\s+warning\b",
    r"\bautomatic\s+emergency\s+braking\b",
    r"\bblind[-\s]?spot\s+monitoring\b",
    r"\bcollision\s+warning\b",
    r"\bparking\s+assist\b",
    r"\bSuper\s+Cruise\b",
    r"\bBlueCruise\b",
    r"\bAutopilot\b",
]


# ------------------------------------------------------------
# F. Road-vehicle context indicators
# ------------------------------------------------------------

road_vehicle_context_patterns = [
    r"\bvehicles?\b",
    r"\bcars?\b",
    r"\bautomobiles?\b",
    r"\bpassenger\s+vehicles?\b",
    r"\btaxis?\b",
    r"\broads?\b",
    r"\bhighways?\b",
    r"\bstreets?\b",
    r"\bautomakers?\b",
    r"\bcar\s+companies\b",
    r"\broad\s+testing\b",
    r"\bpublic\s+roads?\b",
    r"\btest\s+vehicles?\b",
]


# ------------------------------------------------------------
# G. Explicit non-road autonomous-system indicators
# ------------------------------------------------------------

strong_non_road_patterns = [
    # Aviation and drones
    r"\bautonomous\s+drones?\b",
    r"\bdriverless\s+drones?\b",
    r"\bunmanned\s+aerial\s+vehicles?\b",
    r"\bUAVs?\b",
    r"\bautonomous\s+aircraft\b",
    r"\bpilotless\s+aircraft\b",

    # Ships and water transport
    r"\bautonomous\s+ships?\b",
    r"\bunmanned\s+ships?\b",
    r"\bdriverless\s+ships?\b",
    r"\bautonomous\s+boats?\b",
    r"\bunmanned\s+boats?\b",
    r"\bautonomous\s+vessels?\b",

    # Railway and metro
    r"\bdriverless\s+trains?\b",
    r"\bautonomous\s+trains?\b",
    r"\bautomated\s+trains?\b",
    r"\bdriverless\s+metros?\b",
    r"\bautomated\s+metros?\b",

    # Mining
    r"\bautonomous\s+mining\s+trucks?\b",
    r"\bdriverless\s+mining\s+trucks?\b",
    r"\bautonomous\s+mine\s+vehicles?\b",
    r"\bdriverless\s+mine\s+vehicles?\b",

    # Ports
    r"\bautonomous\s+port\s+vehicles?\b",
    r"\bdriverless\s+port\s+vehicles?\b",
    r"\bautomated\s+container\s+vehicles?\b",

    # Agriculture
    r"\bautonomous\s+tractors?\b",
    r"\bdriverless\s+tractors?\b",
    r"\bautonomous\s+farm\s+equipment\b",
    r"\bunmanned\s+farm\s+equipment\b",
]


general_non_road_context_patterns = [
    r"\bdrones?\b",
    r"\baircraft\b",
    r"\baviation\b",
    r"\bships?\b",
    r"\bboats?\b",
    r"\bvessels?\b",
    r"\brailways?\b",
    r"\btrains?\b",
    r"\bmetros?\b",
    r"\bsubways?\b",
    r"\bmines?\b",
    r"\bmining\b",
    r"\bports?\b",
    r"\bharbors?\b",
    r"\btractors?\b",
    r"\bagricultural\s+machinery\b",
]


# ------------------------------------------------------------
# H. AV company, programme and platform indicators
#
# Company terms alone are not sufficient for relevance.
# They must occur with road-vehicle, ADAS or higher-automation
# context.
# ------------------------------------------------------------

av_company_patterns = [
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
    r"\bMotional\b",
    r"\bArgo\s+AI\b",
    r"\bAurora\b",
    r"\bNuro\b",
    r"\bMay\s+Mobility\b",
    r"\bMobileye\b",

    # Tesla systems
    r"\bTesla\s+Autopilot\b",
    r"\bFull\s+Self[-\s]?Driving\b",
    r"\bTesla\s+FSD\b",

    # Other AV programmes
    r"\bGoogle\s+self[-\s]?driving\b",
    r"\bUber\s+autonomous\b",
    r"\bUber\s+self[-\s]?driving\b",
    r"\bGM\s+Cruise\b",
]


# %%
# ============================================================
# 12. Generate automated screening indicators
# ============================================================

pattern_groups = {
    "has_general_road_av_term":
        general_road_av_patterns,

    "has_higher_automation_term":
        higher_automation_patterns,

    "has_passenger_robotaxi_term":
        passenger_robotaxi_patterns,

    "has_strong_passenger_robotaxi_term":
        strong_passenger_robotaxi_patterns,

    "has_bus_public_transport_term":
        bus_public_transport_patterns,

    "has_adas_term":
        adas_patterns,

    "has_road_vehicle_context":
        road_vehicle_context_patterns,

    "has_strong_non_road_term":
        strong_non_road_patterns,

    "has_general_non_road_context":
        general_non_road_context_patterns,

    "has_av_company_term":
        av_company_patterns
}


for output_column, patterns in pattern_groups.items():

    df_scope[output_column] = (
        df_scope["automated_screen_text"]
        .apply(
            lambda text: contains_any_pattern(
                text,
                patterns
            )
        )
    )


df_scope["has_non_road_term"] = (
    df_scope["has_strong_non_road_term"]
    | df_scope["has_general_non_road_context"]
)


# %%
# ============================================================
# 13. Create broader AV-relevance indicators
#
# These variables must be created before the exclusion rules.
# ============================================================

# Strong explicit AV evidence
df_scope["has_explicit_av_signal"] = (
    df_scope["has_general_road_av_term"]
    | df_scope["has_higher_automation_term"]
    | df_scope["has_passenger_robotaxi_term"]
)


# Company/programme evidence is accepted only when accompanied
# by vehicle, ADAS or higher-automation context
df_scope["has_company_supported_av_signal"] = (
    df_scope["has_av_company_term"]
    & (
        df_scope["has_road_vehicle_context"]
        | df_scope["has_adas_term"]
        | df_scope["has_higher_automation_term"]
    )
)


# Broader relevance indicator
df_scope["has_broad_road_av_relevance"] = (
    df_scope["has_explicit_av_signal"]
    | df_scope["has_company_supported_av_signal"]
)


# %%
# ============================================================
# 14. Define moderately relaxed exclusion rules
#
# Automatically exclude only clear out-of-scope cases.
# Ambiguous ADAS and company-context articles are retained for
# manual review.
# ============================================================

# Rule 1:
# No broad road-AV relevance signal
df_scope["remove_no_relevant_road_av_term"] = (
    ~df_scope["has_broad_road_av_relevance"]
)


# Rule 2:
# Explicit non-road autonomous application without meaningful
# passenger-car or robotaxi relevance
df_scope["remove_pure_non_road"] = (
    df_scope["has_strong_non_road_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)


# Rule 3:
# Bus/shuttle/public-transport-only article
df_scope["remove_bus_public_transport_only"] = (
    df_scope["has_bus_public_transport_term"]
    & ~df_scope["has_strong_passenger_robotaxi_term"]
)


# ADAS-only cases are reviewed rather than automatically removed
df_scope["review_adas_only"] = (
    df_scope["has_adas_term"]
    & ~df_scope["has_higher_automation_term"]
)


# Compatibility variable:
# always False because ADAS-only is no longer an automatic rule
df_scope["remove_adas_only"] = False


# Overall automatic exclusion decision
df_scope["exclude_by_automated_scope_screening"] = (
    df_scope["remove_no_relevant_road_av_term"]
    | df_scope["remove_pure_non_road"]
    | df_scope["remove_bus_public_transport_only"]
)


# Retained articles requiring manual scope review
df_scope["requires_manual_scope_review"] = (
    df_scope["review_adas_only"]
    | (
        ~df_scope["has_explicit_av_signal"]
        & df_scope["has_company_supported_av_signal"]
    )
)


# %%
# ============================================================
# 15. Create exclusion and manual-review reasons
# ============================================================

def create_scope_exclusion_reason(row):

    reasons = []

    if row["remove_no_relevant_road_av_term"]:
        reasons.append(
            "No broad road autonomous-vehicle relevance signal"
        )

    if row["remove_pure_non_road"]:
        reasons.append(
            "Pure non-road autonomous-system article"
        )

    if row["remove_bus_public_transport_only"]:
        reasons.append(
            "Bus, shuttle or public-transport-only article"
        )

    return "; ".join(reasons)


df_scope["automated_scope_exclusion_reason"] = (
    df_scope.apply(
        create_scope_exclusion_reason,
        axis=1
    )
)


def create_manual_review_reason(row):

    reasons = []

    if row["review_adas_only"]:
        reasons.append(
            "ADAS terminology without explicit "
            "higher-level automation indicator"
        )

    if (
        not row["has_explicit_av_signal"]
        and row["has_company_supported_av_signal"]
    ):
        reasons.append(
            "AV relevance inferred from company/programme "
            "and road-vehicle context"
        )

    return "; ".join(reasons)


df_scope["manual_scope_review_reason"] = (
    df_scope.apply(
        create_manual_review_reason,
        axis=1
    )
)


# %%
# ============================================================
# 16. Automated scope-screening summary
# ============================================================

scope_summary = pd.DataFrame([
    {
        "screening_rule":
            "No broad road autonomous-vehicle relevance signal",
        "article_count":
            int(
                df_scope[
                    "remove_no_relevant_road_av_term"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Pure non-road autonomous-system article",
        "article_count":
            int(
                df_scope[
                    "remove_pure_non_road"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Bus, shuttle or public-transport-only article",
        "article_count":
            int(
                df_scope[
                    "remove_bus_public_transport_only"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "ADAS case without explicit higher-level automation",
        "article_count":
            int(
                df_scope[
                    "review_adas_only"
                ].sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Company/context-supported borderline AV case",
        "article_count":
            int(
                (
                    ~df_scope["has_explicit_av_signal"]
                    & df_scope[
                        "has_company_supported_av_signal"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    },
    {
        "screening_rule":
            "Excluded by one or more automatic scope rules",
        "article_count":
            int(
                df_scope[
                    "exclude_by_automated_scope_screening"
                ].sum()
            ),
        "decision_type":
            "Automatic exclusion"
    },
    {
        "screening_rule":
            "Retained articles requiring manual scope review",
        "article_count":
            int(
                (
                    df_scope[
                        "requires_manual_scope_review"
                    ]
                    & ~df_scope[
                        "exclude_by_automated_scope_screening"
                    ]
                ).sum()
            ),
        "decision_type":
            "Retained for manual review"
    }
])


scope_summary["percent_of_scope_input"] = (
    scope_summary["article_count"]
    / len(df_scope)
    * 100
).round(2)


print("\nAutomated AV-scope screening summary:")

display(scope_summary)

scope_summary.to_excel(
    OUTPUT_DIR /
    "USA_Today_automated_AV_scope_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 17. Export scope-screening audit files
# ============================================================

scope_audit_columns = [
    "date_fixed",
    DATE_COL,
    TITLE_COL,

    "has_general_road_av_term",
    "has_higher_automation_term",
    "has_passenger_robotaxi_term",
    "has_strong_passenger_robotaxi_term",
    "has_bus_public_transport_term",
    "has_adas_term",
    "has_road_vehicle_context",
    "has_strong_non_road_term",
    "has_general_non_road_context",
    "has_non_road_term",
    "has_av_company_term",

    "has_explicit_av_signal",
    "has_company_supported_av_signal",
    "has_broad_road_av_relevance",

    "remove_no_relevant_road_av_term",
    "remove_pure_non_road",
    "remove_bus_public_transport_only",
    "remove_adas_only",

    "review_adas_only",
    "requires_manual_scope_review",

    "automated_scope_exclusion_reason",
    "manual_scope_review_reason",

    "automated_screen_text",
    "article_body"
]


scope_audit_columns = [
    column
    for column in scope_audit_columns
    if column in df_scope.columns
]


automatically_excluded_scope = df_scope[
    df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


automatically_retained_scope = df_scope[
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


automatically_excluded_scope[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_excluded_by_AV_scope.xlsx",
    index=False
)


manual_scope_review_articles[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_requiring_manual_scope_review.xlsx",
    index=False
)


# Rule 1 exclusions
df_scope[
    df_scope[
        "remove_no_relevant_road_av_term"
    ]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "USA_Today_no_broad_AV_signal_excluded_check.xlsx",
    index=False
)


# Non-road exclusions
df_scope[
    df_scope["remove_pure_non_road"]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "USA_Today_non_road_articles_excluded_check.xlsx",
    index=False
)


# Bus/public-transport exclusions
df_scope[
    df_scope[
        "remove_bus_public_transport_only"
    ]
][scope_audit_columns].to_excel(
    OUTPUT_DIR /
    "USA_Today_bus_only_articles_excluded_check.xlsx",
    index=False
)


# ADAS-only retained review cases
adas_review_cases = df_scope[
    df_scope["review_adas_only"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

adas_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_ADAS_cases_retained_for_review.xlsx",
    index=False
)


# Company/context-supported cases
company_context_review_cases = df_scope[
    ~df_scope["has_explicit_av_signal"]
    & df_scope[
        "has_company_supported_av_signal"
    ]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

company_context_review_cases[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_company_context_AV_cases_retained_for_review.xlsx",
    index=False
)


# Mixed bus/passenger cases retained
mixed_bus_retained = df_scope[
    df_scope["has_bus_public_transport_term"]
    & df_scope[
        "has_strong_passenger_robotaxi_term"
    ]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_bus_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_mixed_bus_passenger_articles_retained_check.xlsx",
    index=False
)


# Mixed ADAS/higher-automation cases retained
mixed_adas_retained = df_scope[
    df_scope["has_adas_term"]
    & df_scope["has_higher_automation_term"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()

mixed_adas_retained[
    scope_audit_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_mixed_ADAS_higher_automation_retained_check.xlsx",
    index=False
)

# %%
# ============================================================
# 17B. Apply completed manual scope-review decisions
#
# The automated screen retained ambiguous articles for manual
# review. The completed review file now determines which of
# those articles remain in the analytical corpus.
#
# Decision rules:
# - INCLUDE: retain in the analytical corpus
# - EXCLUDE: remove before content-hash deduplication
# - Articles not requiring manual review remain retained
# ============================================================

if not MANUAL_REVIEW_COMPLETED_PATH.exists():

    raise FileNotFoundError(
        "\nThe completed manual-review file could not be found:\n"
        f"{MANUAL_REVIEW_COMPLETED_PATH}\n\n"
        "Place USA_Today_manual_scope_review_completed.xlsx "
        "inside OUTPUT_DIR before running the code."
    )


manual_review_completed = pd.read_excel(
    MANUAL_REVIEW_COMPLETED_PATH,
    sheet_name="Sheet1"
)


print("\nCompleted manual-review file loaded:")
print(MANUAL_REVIEW_COMPLETED_PATH)

print("\nManual-review file shape:")
print(manual_review_completed.shape)

print("\nManual-review file columns:")
print(manual_review_completed.columns.tolist())


# ------------------------------------------------------------
# Check required manual-review columns
# ------------------------------------------------------------

required_manual_review_columns = [
    TITLE_COL,
    "article_body",
    "manual_review_decision",
    "manual_review_in_scope",
    "manual_review_reason_final",
    "manual_review_category"
]


missing_manual_review_columns = [
    column
    for column in required_manual_review_columns
    if column not in manual_review_completed.columns
]


if missing_manual_review_columns:

    raise KeyError(
        "The completed manual-review file is missing these columns: "
        f"{missing_manual_review_columns}\n"
        f"Available columns: "
        f"{manual_review_completed.columns.tolist()}"
    )


# ------------------------------------------------------------
# Create stable matching keys
#
# Title alone is not used because two records may share a title.
# We match using normalised title plus normalised article body.
# ------------------------------------------------------------

def normalize_manual_review_match_text(text):
    """
    Normalise title/body text for matching the manually reviewed
    file back to the main dataset.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = (
        text
        .replace("\u00a0", " ")
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip().lower()


def create_manual_review_match_key(title, body):
    """
    Create a SHA-256 matching key from the normalised title and
    article body.
    """

    normalized_title = (
        normalize_manual_review_match_text(title)
    )

    normalized_body = (
        normalize_manual_review_match_text(body)
    )

    combined_text = (
        normalized_title
        + "\n"
        + normalized_body
    )

    return hashlib.sha256(
        combined_text.encode("utf-8")
    ).hexdigest()


# Matching key for the main screened dataset
df_scope["_manual_review_match_key"] = (
    df_scope.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["article_body"]
        ),
        axis=1
    )
)


# Matching key for the completed manual-review file
manual_review_completed[
    "_manual_review_match_key"
] = (
    manual_review_completed.apply(
        lambda row: create_manual_review_match_key(
            row[TITLE_COL],
            row["article_body"]
        ),
        axis=1
    )
)


# ------------------------------------------------------------
# Standardise manual decisions
# ------------------------------------------------------------

manual_review_completed[
    "manual_review_decision"
] = (
    manual_review_completed[
        "manual_review_decision"
    ]
    .fillna("")
    .astype(str)
    .str
    .strip()
    .str
    .upper()
)


valid_manual_decisions = {
    "INCLUDE",
    "EXCLUDE"
}


invalid_manual_decision_rows = (
    manual_review_completed[
        ~manual_review_completed[
            "manual_review_decision"
        ].isin(valid_manual_decisions)
    ]
    .copy()
)


if len(invalid_manual_decision_rows) > 0:

    display_columns = [
        column
        for column in [
            "date_fixed",
            TITLE_COL,
            "manual_review_decision"
        ]
        if column
        in invalid_manual_decision_rows.columns
    ]

    print(
        "\nInvalid or missing manual-review decisions:"
    )

    display(
        invalid_manual_decision_rows[
            display_columns
        ]
    )

    raise ValueError(
        "Every reviewed article must have either INCLUDE "
        "or EXCLUDE in manual_review_decision."
    )


# ------------------------------------------------------------
# Check for conflicting decisions attached to the same article
# ------------------------------------------------------------

decision_conflicts = (
    manual_review_completed
    .groupby(
        "_manual_review_match_key"
    )["manual_review_decision"]
    .nunique()
)


conflicting_keys = set(
    decision_conflicts[
        decision_conflicts > 1
    ].index
)


if conflicting_keys:

    conflicting_rows = (
        manual_review_completed[
            manual_review_completed[
                "_manual_review_match_key"
            ].isin(conflicting_keys)
        ]
        .copy()
    )

    display(
        conflicting_rows[
            [
                TITLE_COL,
                "manual_review_decision",
                "manual_review_reason_final"
            ]
        ]
    )

    raise ValueError(
        "Conflicting manual decisions were found for "
        "the same title/body record."
    )


# ------------------------------------------------------------
# Retain one decision record per unique title/body key
#
# Duplicate LexisNexis copies may legitimately have the same
# title and article body. One manual decision therefore applies
# to all identical copies before content-hash deduplication.
# ------------------------------------------------------------

manual_decision_lookup = (
    manual_review_completed[
        [
            "_manual_review_match_key",
            "manual_review_decision",
            "manual_review_in_scope",
            "manual_review_reason_final",
            "manual_review_category"
        ]
    ]
    .drop_duplicates(
        subset=["_manual_review_match_key"],
        keep="first"
    )
    .copy()
)


manual_decision_lookup = (
    manual_decision_lookup
    .rename(
        columns={
            "manual_review_in_scope":
                "manual_review_in_scope_final"
        }
    )
)


# ------------------------------------------------------------
# Merge manual decisions into the complete scope-screened data
# ------------------------------------------------------------

df_scope = df_scope.merge(
    manual_decision_lookup,
    on="_manual_review_match_key",
    how="left",
    validate="many_to_one"
)


# ------------------------------------------------------------
# Verify that every article flagged for review was matched
# ------------------------------------------------------------

unmatched_manual_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & df_scope[
        "manual_review_decision"
    ].isna()
].copy()


if len(unmatched_manual_review_articles) > 0:

    unmatched_manual_review_articles.to_excel(
        OUTPUT_DIR /
        "USA_Today_unmatched_manual_review_articles_check.xlsx",
        index=False
    )

    print(
        "\nArticles requiring manual review but not matched "
        "to the completed review file:"
    )

    display(
        unmatched_manual_review_articles[
            [
                "date_fixed",
                TITLE_COL,
                "manual_scope_review_reason"
            ]
        ]
    )

    raise ValueError(
        f"{len(unmatched_manual_review_articles)} article(s) "
        "requiring manual review could not be matched. "
        "See USA_Today_unmatched_manual_review_articles_check.xlsx."
    )


# ------------------------------------------------------------
# Define final manual inclusion/exclusion variables
# ------------------------------------------------------------

df_scope["exclude_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "EXCLUDE"
    )
)


df_scope["include_by_manual_scope_review"] = (
    df_scope["requires_manual_scope_review"]
    & (
        df_scope["manual_review_decision"]
        == "INCLUDE"
    )
)


# Articles that were not flagged for review are automatically
# treated as not requiring a manual decision
df_scope["final_scope_inclusion"] = (
    ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
    & ~df_scope[
        "exclude_by_manual_scope_review"
    ]
)


# ------------------------------------------------------------
# Export manually included and manually excluded articles
# ------------------------------------------------------------

manually_included_articles = df_scope[
    df_scope["include_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


manually_excluded_articles = df_scope[
    df_scope["exclude_by_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


manually_included_articles.to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_included_after_manual_review.xlsx",
    index=False
)


manually_excluded_articles.to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_excluded_after_manual_review.xlsx",
    index=False
)


# ------------------------------------------------------------
# Recreate retained data after applying both stages
#
# This variable is used in Section 18 for SHA-256 deduplication.
# Therefore, the 25 manually excluded articles will not enter
# the deduplication or final analytical dataset.
# ------------------------------------------------------------

automatically_retained_scope = df_scope[
    df_scope["final_scope_inclusion"]
].copy()


# Keep this object as the complete pre-decision review set
manual_scope_review_articles = df_scope[
    df_scope["requires_manual_scope_review"]
    & ~df_scope[
        "exclude_by_automated_scope_screening"
    ]
].copy()


# ------------------------------------------------------------
# Counts for summaries
# ------------------------------------------------------------

n_before_scope_screening = len(df_scope)

n_removed_automatic_scope = int(
    df_scope[
        "exclude_by_automated_scope_screening"
    ].sum()
)

n_before_manual_review = int(
    (
        df_scope[
            "requires_manual_scope_review"
        ]
        & ~df_scope[
            "exclude_by_automated_scope_screening"
        ]
    ).sum()
)

n_manual_review_included = int(
    df_scope[
        "include_by_manual_scope_review"
    ].sum()
)

n_manual_review_excluded = int(
    df_scope[
        "exclude_by_manual_scope_review"
    ].sum()
)

n_after_scope_screening = len(
    automatically_retained_scope
)

# Total removed during the complete scope-screening stage:
# automatic exclusions plus manual exclusions
n_removed_scope_screening = (
    n_before_scope_screening
    - n_after_scope_screening
)

# For later summary tables, this now means articles included
# following manual review, rather than all articles originally
# flagged for review
n_manual_review_retained = (
    n_manual_review_included
)


# ------------------------------------------------------------
# Verification output
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MANUAL SCOPE REVIEW APPLIED")
print("=" * 75)

print("\nInput to automated scope screening:")
print(n_before_scope_screening)

print("\nExcluded by automatic scope screening:")
print(n_removed_automatic_scope)

print("\nArticles sent for manual review:")
print(n_before_manual_review)

print("\nIncluded after manual review:")
print(n_manual_review_included)

print("\nExcluded after manual review:")
print(n_manual_review_excluded)

print("\nTotal removed by automatic and manual scope screening:")
print(n_removed_scope_screening)

print("\nRetained after automatic and manual scope screening:")
print(n_after_scope_screening)

print("\nManually included article titles:")

display(
    manually_included_articles[
        [
            "date_fixed",
            TITLE_COL,
            "manual_review_category",
            "manual_review_reason_final"
        ]
    ]
    .sort_values(
        "date_fixed"
    )
)



# %%
# ============================================================
# 18. Normalise full article body for SHA-256 hashing
# ============================================================

df_hash = automatically_retained_scope.copy()


def normalize_content_for_hash(text):
    """
    Normalise article body before exact-content hashing.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"[\u200b\u200c\u200d\ufeff]",
        "",
        text
    )

    text = (
        text
        .replace("“", '"')
        .replace("”", '"')
        .replace("‘", "'")
        .replace("’", "'")
        .replace("—", "-")
        .replace("–", "-")
    )

    text = re.sub(
        r"\s+",
        "",
        text
    )

    return text.strip()


def generate_sha256_hash(normalized_text):
    """
    Generate SHA-256 hash for non-empty normalised text.
    """

    if pd.isna(normalized_text):
        return pd.NA

    normalized_text = str(normalized_text)

    if normalized_text == "":
        return pd.NA

    return hashlib.sha256(
        normalized_text.encode("utf-8")
    ).hexdigest()


df_hash["content_normalized_for_hash"] = (
    df_hash["article_body"]
    .apply(normalize_content_for_hash)
)

df_hash["content_sha256"] = (
    df_hash["content_normalized_for_hash"]
    .apply(generate_sha256_hash)
)

df_hash["normalized_content_length"] = (
    df_hash["content_normalized_for_hash"]
    .str
    .len()
)


# Empty rows receive unique temporary keys
df_hash["_content_dedup_key"] = (
    df_hash["content_sha256"]
    .astype("string")
)

empty_hash_mask = (
    df_hash["content_sha256"].isna()
)

df_hash.loc[
    empty_hash_mask,
    "_content_dedup_key"
] = (
    "EMPTY_CONTENT_ROW_"
    + df_hash.loc[
        empty_hash_mask,
        "_original_order"
    ].astype(str)
)


df_hash["is_content_duplicate_group"] = (
    df_hash[
        "_content_dedup_key"
    ].duplicated(
        keep=False
    )
)


duplicate_content_rows = df_hash[
    df_hash["is_content_duplicate_group"]
].copy()


print(
    "\nRows in duplicated-content groups:",
    len(duplicate_content_rows)
)

print(
    "Number of duplicated-content groups:",
    duplicate_content_rows[
        "_content_dedup_key"
    ].nunique()
)


duplicate_content_rows.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
).to_excel(
    OUTPUT_DIR /
    "USA_Today_duplicated_content_articles_check.xlsx",
    index=False
)


# %%
# ============================================================
# 19. Remove exact-content duplicates
#
# Retention rule:
# 1. Keep most recent publication date.
# 2. If dates are equal, keep earliest original row.
# ============================================================

df_hash_sorted = df_hash.sort_values(
    by=[
        "_content_dedup_key",
        "date_fixed",
        "_original_order"
    ],
    ascending=[
        True,
        False,
        True
    ],
    na_position="last"
)


df_no_content_duplicates = (
    df_hash_sorted
    .drop_duplicates(
        subset=["_content_dedup_key"],
        keep="first"
    )
    .copy()
)


kept_original_orders = set(
    df_no_content_duplicates[
        "_original_order"
    ]
)


removed_content_duplicates = df_hash[
    ~df_hash["_original_order"].isin(
        kept_original_orders
    )
].copy()


removed_content_duplicates.to_excel(
    OUTPUT_DIR /
    "USA_Today_articles_removed_by_content_hash.xlsx",
    index=False
)


n_before_hash_deduplication = len(df_hash)

n_after_hash_deduplication = len(
    df_no_content_duplicates
)

n_removed_content_duplicates = (
    n_before_hash_deduplication
    - n_after_hash_deduplication
)


print("\nBefore content-hash deduplication:")
print(n_before_hash_deduplication)

print("\nAfter content-hash deduplication:")
print(n_after_hash_deduplication)

print("\nRemoved by content-hash deduplication:")
print(n_removed_content_duplicates)


# %%
# ============================================================
# 20. Create final analytical corpus before lead extraction
# ============================================================

data = df_no_content_duplicates.copy()

data["date_fixed"] = pd.to_datetime(
    data["date_fixed"],
    errors="coerce"
)

data["year"] = (
    data["date_fixed"]
    .dt
    .year
)

data["month"] = (
    data["date_fixed"]
    .dt
    .month
)

data["month_name"] = (
    data["date_fixed"]
    .dt
    .month_name()
)

data["year_month"] = (
    data["date_fixed"]
    .dt
    .to_period("M")
    .astype(str)
)

data["content_char_length"] = (
    data["article_body"]
    .fillna("")
    .str
    .len()
)

data["full_text_word_count"] = (
    data["article_body"]
    .apply(count_english_words)
)


data = (
    data
    .sort_values(
        "date_fixed",
        na_position="last"
    )
    .reset_index(drop=True)
)


data.to_excel(
    FINAL_DATA_PATH,
    index=False
)


print("\nFinal corpus before lead extraction:")
print(len(data))

print("\nSaved to:")
print(FINAL_DATA_PATH)


# %%
# ============================================================
# 21. First-substantive-paragraph extraction helpers
# ============================================================

def clean_candidate_line(line):
    """
    Clean one candidate paragraph or line.
    """

    if pd.isna(line):
        return ""

    line = str(line)

    line = line.replace(
        "\u00a0",
        " "
    )

    line = re.sub(
        r"[ \t]+",
        " ",
        line
    )

    return (
        line
        .strip()
        .strip('"')
        .strip()
    )


def normalize_for_comparison(text):
    """
    Normalise text for title and metadata comparison.
    """

    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = unicodedata.normalize(
        "NFKC",
        text
    )

    text = re.sub(
        r"\s+",
        "",
        text
    )

    text = re.sub(
        r"""[.,!?;:'"“”‘’()\-—–]""",
        "",
        text
    )

    return text


def is_same_as_title(line, title):
    """
    Check whether a candidate repeats the article title.
    """

    line_key = normalize_for_comparison(line)
    title_key = normalize_for_comparison(title)

    if not line_key or not title_key:
        return False

    return (
        line_key == title_key
        and len(title_key) >= 8
    )


def is_usa_today_metadata_line(line):
    """
    Identify common standalone bylines, publisher labels,
    photo credits and editorial notes.
    """

    compact = clean_candidate_line(line)

    if compact == "":
        return True

    metadata_patterns = [
        r"^USA\s+TODAY$",
        r"^Associated\s+Press$",
        r"^Reuters$",

        # Standalone byline
        r"^By\s+[A-Z][A-Za-z.'’\-]+"
        r"(?:\s+[A-Z][A-Za-z.'’\-]+){0,5}$",

        # Name followed by USA TODAY
        r"^[A-Z][A-Za-z.'’\-]+"
        r"(?:\s+[A-Z][A-Za-z.'’\-]+){1,5}"
        r",?\s+USA\s+TODAY$",

        # Photo and visual metadata
        r"^(?:Photo|Photos|Video|Graphic|Illustration)"
        r"\s*(?:by|:).*$",

        r"^Credit:\s*.*$",
        r"^Caption:\s*.*$",

        # Editorial notes
        r"^Correction$",
        r"^Corrections$",
        r"^Editor's\s+note:?.*$",
        r"^Editor’s\s+note:?.*$",
    ]

    return any(
        re.fullmatch(
            pattern,
            compact,
            flags=re.IGNORECASE
        )
        is not None
        for pattern in metadata_patterns
    )


def remove_english_dateline_prefix(paragraph):
    """
    Conservatively remove an English news dateline.

    Examples:
    WASHINGTON — The agency announced...
    DETROIT – Automakers are...
    SAN FRANCISCO (AP) — A company...
    """

    original = clean_candidate_line(paragraph)

    if original == "":
        return "", False

    text = original

    # More specific patterns must be checked first
    dateline_patterns = [
        # CITY (AP) — text
        r"^[A-Z][A-Za-z .'\-]{1,40}"
        r"\s*\((?:AP|Reuters|USA TODAY)\)"
        r"\s*[—–-]\s+",

        # CITY, State — text
        r"^[A-Z][A-Za-z .'\-]{1,30},\s*"
        r"[A-Z][A-Za-z .'\-]{1,20}"
        r"\s*[—–-]\s+",

        # ALL-CAPS CITY — text
        r"^[A-Z][A-Z .'\-]{1,40}"
        r"\s*[—–-]\s+",
    ]

    for pattern in dateline_patterns:

        new_text = re.sub(
            pattern,
            "",
            text,
            count=1
        )

        if new_text != text:

            text = new_text.strip()
            break

    prefix_removed = (
        normalize_for_comparison(text)
        != normalize_for_comparison(original)
    )

    return text, prefix_removed


# %%
# ============================================================
# 22. Extract first substantive paragraph
# ============================================================

def extract_first_substantive_paragraph(
    full_text,
    title=""
):
    """
    Extract the first substantive paragraph from a USA Today
    article body.

    Skip:
    - repeated titles;
    - standalone bylines;
    - publisher labels;
    - photo or caption metadata;
    - editorial notes.
    """

    result = {
        "first_paragraph_before_dateline_removal": "",
        "first_paragraph": "",
        "english_dateline_removed": False,
        "extraction_candidate_position": pd.NA,
        "skipped_metadata_count": 0,
        "skipped_metadata_text": ""
    }

    if pd.isna(full_text):
        return result

    text = normalize_line_breaks(full_text)

    if text == "":
        return result

    raw_lines = re.split(
        r"\n+",
        text
    )

    candidates = []

    for raw_line in raw_lines:

        candidate = clean_candidate_line(
            raw_line
        )

        if candidate != "":
            candidates.append(candidate)

    skipped_metadata = []

    for position, candidate in enumerate(
        candidates,
        start=1
    ):

        if is_same_as_title(
            candidate,
            title
        ):
            skipped_metadata.append(candidate)
            continue

        if is_usa_today_metadata_line(
            candidate
        ):
            skipped_metadata.append(candidate)
            continue

        cleaned_candidate, dateline_removed = (
            remove_english_dateline_prefix(
                candidate
            )
        )

        if cleaned_candidate == "":
            skipped_metadata.append(candidate)
            continue

        if is_same_as_title(
            cleaned_candidate,
            title
        ):
            skipped_metadata.append(candidate)
            continue

        if is_usa_today_metadata_line(
            cleaned_candidate
        ):
            skipped_metadata.append(candidate)
            continue

        result[
            "first_paragraph_before_dateline_removal"
        ] = candidate

        result[
            "first_paragraph"
        ] = cleaned_candidate

        result[
            "english_dateline_removed"
        ] = dateline_removed

        result[
            "extraction_candidate_position"
        ] = position

        result[
            "skipped_metadata_count"
        ] = len(skipped_metadata)

        result[
            "skipped_metadata_text"
        ] = " | ".join(skipped_metadata)

        return result

    result[
        "skipped_metadata_count"
    ] = len(skipped_metadata)

    result[
        "skipped_metadata_text"
    ] = " | ".join(skipped_metadata)

    return result


extraction_results = data.apply(
    lambda row: extract_first_substantive_paragraph(
        row["article_body"],
        row[TITLE_COL]
    ),
    axis=1
)


extraction_results_df = pd.DataFrame(
    extraction_results.tolist(),
    index=data.index
)


for column in extraction_results_df.columns:

    data[column] = (
        extraction_results_df[column]
    )


# %%
# ============================================================
# 23. Calculate first-paragraph measures
# ============================================================

data["first_paragraph_word_count"] = (
    data["first_paragraph"]
    .apply(count_english_words)
)

data["first_paragraph_char_count"] = (
    data["first_paragraph"]
    .fillna("")
    .apply(
        lambda text: len(
            re.sub(
                r"\s+",
                "",
                str(text)
            )
        )
    )
)

data["first_paragraph_extracted"] = (
    data["first_paragraph"]
    .fillna("")
    .str
    .strip()
    .ne("")
)

data["first_paragraph_equals_full_text"] = (
    data["first_paragraph"]
    .fillna("")
    .str
    .strip()
    ==
    data["article_body"]
    .fillna("")
    .str
    .strip()
)


# %%
# ============================================================
# 23B. Extract First AV-Relevant Substantive Paragraph
#
# Purpose:
# - Preserve the original first substantive paragraph
# - Identify the earliest substantive paragraph that actually
#   discusses autonomous vehicles / autonomous driving
# - Use this paragraph as the main text unit for AV-specific
#   sentiment classification
#
# Main logic:
# 1. Split article_body into paragraph-like lines
# 2. Skip title and USA Today metadata
# 3. Remove English dateline prefixes
# 4. Require a minimally substantive paragraph
# 5. Test paragraph-level AV relevance
# 6. Return the earliest qualifying paragraph
#
# IMPORTANT:
# The original first_paragraph is NOT deleted.
# ============================================================


# ============================================================
# 23B.1 Output paths
# ============================================================

FINAL_WITH_AV_PARAGRAPH_PATH = (
    OUTPUT_DIR /
    "USA_Today_AV_final_with_first_av_relevant_paragraph.xlsx"
)

AV_PARAGRAPH_CHECK_PATH = (
    OUTPUT_DIR /
    "USA_Today_first_av_relevant_paragraph_extraction_check.xlsx"
)

AV_PARAGRAPH_FAILED_PATH = (
    OUTPUT_DIR /
    "USA_Today_first_av_relevant_paragraph_failed_check.xlsx"
)

AV_PARAGRAPH_POSITION_PATH = (
    OUTPUT_DIR /
    "USA_Today_first_av_relevant_paragraph_position_distribution.xlsx"
)


# %%
# ============================================================
# 23B.2 Strong paragraph-level AV patterns
# ============================================================

STRONG_AV_PARAGRAPH_PATTERNS_EN = [

    # Autonomous vehicle / car
    r"\bautonomous\s+vehicles?\b",
    r"\bautonomous\s+cars?\b",
    r"\bautomated\s+vehicles?\b",
    r"\bautomated\s+cars?\b",

    # Autonomous / automated driving
    r"\bautonomous\s+driving\b",
    r"\bautomated\s+driving\b",
    r"\bautonomous\s+driving\s+systems?\b",
    r"\bautomated\s+driving\s+systems?\b",

    # Self-driving
    r"\bself[-\s]?driving\b",
    r"\bself[-\s]?driving\s+cars?\b",
    r"\bself[-\s]?driving\s+vehicles?\b",

    # Driverless
    r"\bdriverless\b",
    r"\bdriverless\s+cars?\b",
    r"\bdriverless\s+vehicles?\b",

    # Robotaxi
    r"\brobo[-\s]?taxis?\b",
    r"\brobotaxis?\b",
    r"\bautonomous\s+taxis?\b",
    r"\bdriverless\s+taxis?\b",
    r"\bself[-\s]?driving\s+taxis?\b",

    # High automation
    r"\bfully\s+autonomous\b",
    r"\bfull\s+autonomy\b",
    r"\bfull\s+self[-\s]?driving\b",
]


# %%
# ============================================================
# 23B.3 Higher-automation context
# ============================================================

HIGH_AUTOMATION_PARAGRAPH_PATTERNS_EN = [

    r"\bSAE\s+(?:Level\s*)?[345]\b",
    r"\bLevel[-\s]?[345]\b",
    r"\bL[345]\b",

    r"\bfully\s+autonomous\b",
    r"\bfully\s+driverless\b",

    r"\bwithout\s+(?:a\s+)?safety\s+driver\b",
    r"\bno\s+safety\s+driver\b",

    r"\bremove(?:d|s|ing)?\s+"
    r"(?:the\s+)?safety\s+driver\b",

    r"\bcommercial\s+driverless\s+service\b",
    r"\bcommercial\s+autonomous\s+service\b",
    r"\bcommercial\s+robotaxi\s+service\b",

    r"\bpaid\s+driverless\s+rides?\b",
    r"\bpaid\s+rides?\b",

    r"\bdriverless\s+ride[-\s]?hailing\b",
    r"\bautonomous\s+ride[-\s]?hailing\b",
]


# %%
# ============================================================
# 23B.4 AV company / programme terms
#
# Company names alone are not automatically treated as enough.
# They require vehicle, ride, deployment, test or automation
# context in the same paragraph.
# ============================================================

AV_COMPANY_PARAGRAPH_PATTERNS_EN = [
    r"\bWaymo\b",
    r"\bCruise\b",
    r"\bZoox\b",
    r"\bMotional\b",
    r"\bArgo\s+AI\b",
    r"\bAurora\b",
    r"\bNuro\b",
    r"\bMay\s+Mobility\b",
    r"\bMobileye\b",
]


AV_COMPANY_CONTEXT_PATTERNS_EN = [

    # Vehicle context
    r"\bvehicles?\b",
    r"\bcars?\b",
    r"\bfleet\b",
    r"\bfleets\b",

    # Passenger / taxi / ride context
    r"\btaxis?\b",
    r"\bcabs?\b",
    r"\brides?\b",
    r"\bpassengers?\b",
    r"\bride[-\s]?hailing\b",
    r"\bride[-\s]?sharing\b",

    # Testing
    r"\broad\s+testing\b",
    r"\btest(?:ing|ed)?\b",
    r"\bpublic\s+roads?\b",

    # Deployment / operation
    r"\bdeploy(?:ed|ing|ment)?\b",
    r"\blaunch(?:ed|ing)?\b",
    r"\bservice\b",
    r"\boperations?\b",
    r"\bcommercial(?:ize|ized|ization)?\b",

    # Driving context
    r"\bdriving\b",
    r"\bdriverless\b",
    r"\bautonomous\b",
    r"\bself[-\s]?driving\b",
]


# %%
# ============================================================
# 23B.5 Tesla-specific AV programme patterns
#
# Tesla itself is not enough because USA Today may discuss Tesla
# for many unrelated reasons.
# ============================================================

TESLA_AV_PARAGRAPH_PATTERNS_EN = [
    r"\bTesla\s+Autopilot\b",
    r"\bTesla\s+FSD\b",
    r"\bFull\s+Self[-\s]?Driving\b",
    r"\bAutopilot\b",
]


# %%
# ============================================================
# 23B.6 Paragraph relevance helper
# ============================================================

def paragraph_contains_pattern_group(
    text,
    patterns
):
    """
    Return True when at least one regex pattern is present.
    """

    if pd.isna(text):
        return False

    text = str(text)

    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        is not None
        for pattern in patterns
    )


def is_paragraph_av_relevant_en(
    paragraph
):
    """
    Determine whether a paragraph substantively refers to
    autonomous road vehicles.

    A paragraph is relevant when:

    1. It contains an explicit strong AV expression; OR

    2. It contains an explicit higher-automation indicator; OR

    3. It contains a known AV company together with vehicle,
       passenger-service, testing, deployment or automation
       context; OR

    4. It explicitly mentions a Tesla AV programme such as
       Autopilot or Full Self-Driving.

    Company names alone are deliberately insufficient.
    """

    if pd.isna(paragraph):
        return False

    paragraph = str(paragraph).strip()

    if paragraph == "":
        return False


    # --------------------------------------------------------
    # Rule 1: explicit road-AV language
    # --------------------------------------------------------

    if paragraph_contains_pattern_group(
        paragraph,
        STRONG_AV_PARAGRAPH_PATTERNS_EN
    ):
        return True


    # --------------------------------------------------------
    # Rule 2: explicit higher automation
    # --------------------------------------------------------

    if paragraph_contains_pattern_group(
        paragraph,
        HIGH_AUTOMATION_PARAGRAPH_PATTERNS_EN
    ):
        return True


    # --------------------------------------------------------
    # Rule 3: AV company + relevant context
    # --------------------------------------------------------

    has_av_company = (
        paragraph_contains_pattern_group(
            paragraph,
            AV_COMPANY_PARAGRAPH_PATTERNS_EN
        )
    )

    has_company_context = (
        paragraph_contains_pattern_group(
            paragraph,
            AV_COMPANY_CONTEXT_PATTERNS_EN
        )
    )

    if (
        has_av_company
        and has_company_context
    ):
        return True


    # --------------------------------------------------------
    # Rule 4: explicit Tesla AV programme
    # --------------------------------------------------------

    if paragraph_contains_pattern_group(
        paragraph,
        TESLA_AV_PARAGRAPH_PATTERNS_EN
    ):
        return True


    return False


# %%
# ============================================================
# 23B.7 Define substantive paragraph criterion
# ============================================================

def is_substantive_av_candidate_en(
    paragraph,
    min_words=10
):
    """
    Require a minimum amount of substantive English text.

    Ten words is used here rather than fifteen because some
    legitimate USA Today leads are concise. Paragraphs shorter
    than 15 words can still be separately audited.
    """

    if pd.isna(paragraph):
        return False

    paragraph = str(paragraph).strip()

    if paragraph == "":
        return False

    return (
        count_english_words(
            paragraph
        )
        >= min_words
    )


# %%
# ============================================================
# 23B.8 Extract first AV-relevant substantive paragraph
# ============================================================

def extract_first_av_relevant_substantive_paragraph_en(
    full_text,
    title="",
    min_words=10
):
    """
    Extract the earliest substantive paragraph that actually
    discusses autonomous road vehicles.

    Existing USA Today cleaning functions are reused:
    - normalize_line_breaks
    - clean_candidate_line
    - is_same_as_title
    - is_usa_today_metadata_line
    - remove_english_dateline_prefix
    """

    result = {

        "first_av_relevant_paragraph_before_dateline_removal":
            "",

        "first_av_relevant_paragraph":
            "",

        "first_av_relevant_paragraph_found":
            False,

        "first_av_relevant_paragraph_position":
            pd.NA,

        "first_av_relevant_substantive_position":
            pd.NA,

        "av_relevant_english_dateline_removed":
            False,

        "av_nonrelevant_substantive_paragraphs_skipped":
            0,

        "av_skipped_nonrelevant_text":
            ""
    }


    if pd.isna(full_text):
        return result


    text = normalize_line_breaks(
        full_text
    )


    if text == "":
        return result


    # --------------------------------------------------------
    # Split article body into paragraph-like lines
    # --------------------------------------------------------

    raw_candidates = re.split(
        r"\n+",
        text
    )


    candidates = []

    for raw_candidate in raw_candidates:

        candidate = clean_candidate_line(
            raw_candidate
        )

        if candidate != "":
            candidates.append(
                candidate
            )


    substantive_position = 0

    skipped_nonrelevant = []


    # --------------------------------------------------------
    # Search from beginning of article
    # --------------------------------------------------------

    for raw_position, candidate in enumerate(
        candidates,
        start=1
    ):


        # ----------------------------------------------------
        # Skip repeated title
        # ----------------------------------------------------

        if is_same_as_title(
            candidate,
            title
        ):
            continue


        # ----------------------------------------------------
        # Skip publisher/byline/photo/editorial metadata
        # ----------------------------------------------------

        if is_usa_today_metadata_line(
            candidate
        ):
            continue


        # ----------------------------------------------------
        # Remove dateline
        # ----------------------------------------------------

        (
            cleaned_candidate,
            dateline_removed
        ) = remove_english_dateline_prefix(
            candidate
        )


        if cleaned_candidate == "":
            continue


        # ----------------------------------------------------
        # Recheck after dateline removal
        # ----------------------------------------------------

        if is_same_as_title(
            cleaned_candidate,
            title
        ):
            continue


        if is_usa_today_metadata_line(
            cleaned_candidate
        ):
            continue


        # ----------------------------------------------------
        # Require substantive text
        # ----------------------------------------------------

        if not is_substantive_av_candidate_en(
            cleaned_candidate,
            min_words=min_words
        ):
            continue


        substantive_position += 1


        # ----------------------------------------------------
        # Test AV relevance
        # ----------------------------------------------------

        if is_paragraph_av_relevant_en(
            cleaned_candidate
        ):

            result[
                "first_av_relevant_paragraph_before_dateline_removal"
            ] = candidate

            result[
                "first_av_relevant_paragraph"
            ] = cleaned_candidate

            result[
                "first_av_relevant_paragraph_found"
            ] = True

            result[
                "first_av_relevant_paragraph_position"
            ] = raw_position

            result[
                "first_av_relevant_substantive_position"
            ] = substantive_position

            result[
                "av_relevant_english_dateline_removed"
            ] = dateline_removed

            result[
                "av_nonrelevant_substantive_paragraphs_skipped"
            ] = len(
                skipped_nonrelevant
            )

            result[
                "av_skipped_nonrelevant_text"
            ] = " | ".join(
                skipped_nonrelevant
            )

            return result


        # Substantive, but not AV-relevant
        skipped_nonrelevant.append(
            cleaned_candidate
        )


    # --------------------------------------------------------
    # No qualifying paragraph found
    # --------------------------------------------------------

    result[
        "av_nonrelevant_substantive_paragraphs_skipped"
    ] = len(
        skipped_nonrelevant
    )

    result[
        "av_skipped_nonrelevant_text"
    ] = " | ".join(
        skipped_nonrelevant
    )


    return result


# %%
# ============================================================
# 23B.9 Apply extraction
# ============================================================

av_extraction_results = data.apply(

    lambda row:
        extract_first_av_relevant_substantive_paragraph_en(
            row["article_body"],
            row[TITLE_COL],
            min_words=10
        ),

    axis=1
)


av_extraction_results_df = pd.DataFrame(
    av_extraction_results.tolist(),
    index=data.index
)


for column in av_extraction_results_df.columns:

    data[column] = (
        av_extraction_results_df[
            column
        ]
    )


print(
    "\nFirst AV-relevant substantive "
    "paragraph extraction completed."
)


print(
    "\nExtraction status:"
)

print(
    data[
        "first_av_relevant_paragraph_found"
    ]
    .value_counts(
        dropna=False
    )
)


# %%
# ============================================================
# 23B.10 Calculate AV-paragraph measures
# ============================================================

data[
    "first_av_relevant_paragraph_word_count"
] = (
    data[
        "first_av_relevant_paragraph"
    ]
    .apply(
        count_english_words
    )
)


data[
    "first_av_relevant_paragraph_char_count"
] = (
    data[
        "first_av_relevant_paragraph"
    ]
    .fillna("")
    .apply(
        lambda text:
            len(
                re.sub(
                    r"\s+",
                    "",
                    str(text)
                )
            )
    )
)


# %%
# ============================================================
# 23B.11 Compare original first paragraph with AV paragraph
# ============================================================

data[
    "first_paragraph_is_av_relevant"
] = (
    data[
        "first_paragraph"
    ]
    .apply(
        is_paragraph_av_relevant_en
    )
)


data[
    "av_paragraph_same_as_original_first"
] = (

    data[
        "first_av_relevant_paragraph"
    ]
    .fillna("")
    .apply(
        normalize_for_comparison
    )

    ==

    data[
        "first_paragraph"
    ]
    .fillna("")
    .apply(
        normalize_for_comparison
    )
)


n_original_first_is_av = int(
    data[
        "first_paragraph_is_av_relevant"
    ].sum()
)


pct_original_first_is_av = (
    n_original_first_is_av
    /
    len(data)
    *
    100
)


print(
    "\nOriginal first substantive paragraph "
    "already AV-relevant:"
)

print(
    n_original_first_is_av
)


print(
    "\nPercentage of articles whose original "
    "first paragraph is AV-relevant:"
)

print(
    round(
        pct_original_first_is_av,
        2
    ),
    "%"
)


# %%
# ============================================================
# 23B.12 Position distribution
# ============================================================

av_position_distribution = (

    data.loc[
        data[
            "first_av_relevant_paragraph_found"
        ],
        "first_av_relevant_substantive_position"
    ]

    .value_counts()

    .sort_index()

    .reset_index()
)


av_position_distribution.columns = [
    "substantive_paragraph_position",
    "article_count"
]


av_position_distribution[
    "percentage"
] = (

    av_position_distribution[
        "article_count"
    ]

    /

    av_position_distribution[
        "article_count"
    ].sum()

    * 100

).round(2)


display(
    av_position_distribution.head(30)
)


av_position_distribution.to_excel(
    AV_PARAGRAPH_POSITION_PATH,
    index=False
)


# %%
# ============================================================
# 23B.13 Export failed cases
# ============================================================

failed_av_paragraphs = data[
    ~data[
        "first_av_relevant_paragraph_found"
    ]
].copy()


print(
    "\nArticles with no AV-relevant "
    "substantive paragraph:"
)

print(
    len(
        failed_av_paragraphs
    )
)


failed_av_columns = [
    "date_fixed",
    TITLE_COL,

    "first_paragraph",
    "first_paragraph_is_av_relevant",

    "automated_scope_exclusion_reason",
    "manual_review_decision",
    "manual_review_category",

    "article_body"
]


failed_av_columns = [
    column
    for column in failed_av_columns
    if column in failed_av_paragraphs.columns
]


failed_av_paragraphs[
    failed_av_columns
].to_excel(
    AV_PARAGRAPH_FAILED_PATH,
    index=False
)


# %%
# ============================================================
# 23B.14 Random 50-article extraction audit
# ============================================================

av_found_data = data[
    data[
        "first_av_relevant_paragraph_found"
    ]
].copy()


N_AV_EXTRACTION_CHECK = min(
    50,
    len(av_found_data)
)


if N_AV_EXTRACTION_CHECK > 0:

    av_extraction_check = (
        av_found_data
        .sample(
            n=N_AV_EXTRACTION_CHECK,
            random_state=42
        )
        .copy()
    )


    av_check_columns = [
        "date_fixed",
        TITLE_COL,

        # Original lead
        "first_paragraph",
        "first_paragraph_word_count",
        "first_paragraph_is_av_relevant",

        # AV-specific paragraph
        "first_av_relevant_paragraph",
        "first_av_relevant_paragraph_word_count",

        "first_av_relevant_paragraph_position",
        "first_av_relevant_substantive_position",

        "av_nonrelevant_substantive_paragraphs_skipped",
        "av_skipped_nonrelevant_text",

        "article_body"
    ]


    av_check_columns = [
        column
        for column in av_check_columns
        if column in av_extraction_check.columns
    ]


    av_extraction_check[
        av_check_columns
    ].to_excel(
        AV_PARAGRAPH_CHECK_PATH,
        index=False
    )


    display(
        av_extraction_check[
            av_check_columns
        ].head(10)
    )


# %%
# ============================================================
# 23B.15 Create standard sentiment-analysis text columns
# ============================================================

data[
    "sentiment_analysis_text"
] = (
    data[
        "first_av_relevant_paragraph"
    ]
)


data[
    "sentiment_analysis_text_preview"
] = (
    data[
        "first_av_relevant_paragraph"
    ]
    .fillna("")
    .astype(str)
    .str[:500]
)


data[
    "has_sentiment_analysis_text"
] = (
    data[
        "sentiment_analysis_text"
    ]
    .fillna("")
    .astype(str)
    .str
    .strip()
    .ne("")
)


print(
    "\nArticles with usable AV-specific "
    "sentiment-analysis text:"
)

print(
    int(
        data[
            "has_sentiment_analysis_text"
        ].sum()
    )
)


print(
    "\nArticles without usable AV-specific "
    "sentiment-analysis text:"
)

print(
    int(
        (
            ~data[
                "has_sentiment_analysis_text"
            ]
        ).sum()
    )
)

# %%
# ============================================================
# 24. Lead-extraction audit variables
# ============================================================

data["first_paragraph_starts_with_byline"] = (
    data["first_paragraph"]
    .fillna("")
    .str
    .contains(
        r"^\s*By\s+[A-Z]",
        regex=True,
        case=False,
        na=False
    )
)

data["first_paragraph_starts_with_usa_today"] = (
    data["first_paragraph"]
    .fillna("")
    .str
    .contains(
        r"^\s*USA\s+TODAY\b",
        regex=True,
        case=False,
        na=False
    )
)

data["first_paragraph_has_possible_dateline"] = (
    data["first_paragraph"]
    .fillna("")
    .str
    .contains(
        r"^\s*[A-Z][A-Z .'\-]{2,40}"
        r"\s*[—–-]\s+",
        regex=True,
        na=False
    )
)

data["first_paragraph_prefix_problem"] = (
    data["first_paragraph_starts_with_byline"]
    | data[
        "first_paragraph_starts_with_usa_today"
    ]
    | data[
        "first_paragraph_has_possible_dateline"
    ]
)

# %%
# ============================================================
# 34B. Save final dataset with AV-relevant paragraph
# ============================================================

data.to_excel(
    FINAL_WITH_AV_PARAGRAPH_PATH,
    index=False
)


print(
    "\nFinal dataset with AV-relevant paragraph saved to:"
)

print(
    FINAL_WITH_AV_PARAGRAPH_PATH
)

# %%
# ============================================================
# 25. Export lead-extraction audit files
# ============================================================

failed_first_paragraph = data[
    ~data["first_paragraph_extracted"]
].copy()

short_first_paragraph = data[
    data["first_paragraph_extracted"]
    & (
        data["first_paragraph_word_count"] < 15
    )
].copy()

long_first_paragraph = data[
    data["first_paragraph_word_count"] > 200
].copy()

remaining_prefix_problems = data[
    data["first_paragraph_prefix_problem"]
].copy()

first_equals_full = data[
    data["first_paragraph_extracted"]
    & data["first_paragraph_equals_full_text"]
].copy()

skipped_metadata_cases = data[
    data["skipped_metadata_count"] > 0
].copy()


failed_first_paragraph.to_excel(
    OUTPUT_DIR /
    "USA_Today_failed_first_paragraph_extraction.xlsx",
    index=False
)

short_first_paragraph.to_excel(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_under_15_words_check.xlsx",
    index=False
)

long_first_paragraph.to_excel(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_over_200_words_check.xlsx",
    index=False
)

remaining_prefix_problems.to_excel(
    OUTPUT_DIR /
    "USA_Today_remaining_lead_prefix_check.xlsx",
    index=False
)

first_equals_full.to_excel(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_equals_full_text_check.xlsx",
    index=False
)

skipped_metadata_cases.to_excel(
    OUTPUT_DIR /
    "USA_Today_skipped_metadata_check.xlsx",
    index=False
)


# %%
# ============================================================
# 26. First-paragraph descriptive statistics
# ============================================================

valid_leads = data[
    data["first_paragraph_extracted"]
].copy()


if len(valid_leads) == 0:

    raise ValueError(
        "No first paragraphs were successfully extracted."
    )


lead_statistics = pd.DataFrame([
    {
        "metric": "Total final analytical articles",
        "value": len(data)
    },
    {
        "metric": "Successfully extracted first paragraphs",
        "value": len(valid_leads)
    },
    {
        "metric": "Failed first-paragraph extractions",
        "value": len(failed_first_paragraph)
    },
    {
        "metric": "Extraction success rate (%)",
        "value": round(
            len(valid_leads)
            / len(data)
            * 100,
            2
        )
    },
    {
        "metric": "Articles with skipped metadata",
        "value": len(skipped_metadata_cases)
    },
    {
        "metric": "Dateline prefixes removed",
        "value": int(
            data[
                "english_dateline_removed"
            ].sum()
        )
    },
    {
        "metric": "Remaining possible prefix problems",
        "value": len(remaining_prefix_problems)
    },
    {
        "metric": "First paragraph equals full article",
        "value": len(first_equals_full)
    },
    {
        "metric": "First paragraphs under 15 words",
        "value": len(short_first_paragraph)
    },
    {
        "metric": "First paragraphs over 200 words",
        "value": len(long_first_paragraph)
    },
    {
        "metric": "Mean first-paragraph word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].mean()
    },
    {
        "metric": "Median first-paragraph word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].median()
    },
    {
        "metric": "Standard deviation",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].std()
    },
    {
        "metric": "Minimum word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].min()
    },
    {
        "metric": "25th percentile",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].quantile(0.25)
    },
    {
        "metric": "75th percentile",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].quantile(0.75)
    },
    {
        "metric": "90th percentile",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].quantile(0.90)
    },
    {
        "metric": "95th percentile",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].quantile(0.95)
    },
    {
        "metric": "Maximum word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].max()
    }
])


lead_statistics["value"] = (
    lead_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)


display(lead_statistics)

lead_statistics.to_excel(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 27. Cleaning-process summary
# ============================================================

cleaning_summary = pd.DataFrame([
    {
        "step": "Original USA Today dataset",
        "article_count": n_original,
        "removed_at_step": 0,
        "retained_for_manual_review": 0,
        "screening_type": "Input data"
    },
    {
        "step":
            "After removing invalid dates and "
            "failed body extractions",
        "article_count":
            n_after_initial_cleaning,
        "removed_at_step":
            n_removed_initial_cleaning,
        "retained_for_manual_review":
            0,
        "screening_type":
            "Source-format and usability cleaning"
    },
    {
        "step":
            "After automated AV scope screening",
        "article_count":
            n_before_scope_screening
            - n_removed_automatic_scope,
        "removed_at_step":
            n_removed_automatic_scope,
        "retained_for_manual_review":
            n_before_manual_review,
        "screening_type":
            (
                "Automatic exclusion of clearly non-relevant, "
                "non-road and public-transport-only articles"
            )
    },
    {
        "step":
            "After manual scope review",
        "article_count":
            n_after_scope_screening,
        "removed_at_step":
            n_manual_review_excluded,
        "retained_for_manual_review":
            n_manual_review_included,
        "screening_type":
            (
                "Manual adjudication of borderline ADAS and "
                "company-context cases: INCLUDE retained and "
                "EXCLUDE removed"
            )
    },

    {
        "step":
            "After full-content SHA-256 hash deduplication",
        "article_count":
            n_after_hash_deduplication,
        "removed_at_step":
            n_removed_content_duplicates,
        "retained_for_manual_review":
            int(
                df_no_content_duplicates[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Automated exact-content deduplication"
    },
    {
        "step":
            "Final analytical dataset",
        "article_count":
            len(data),
        "removed_at_step":
            0,
        "retained_for_manual_review":
            int(
                data[
                    "requires_manual_scope_review"
                ].sum()
            ),
        "screening_type":
            "Final retained sample"
    }
])


display(cleaning_summary)

cleaning_summary.to_excel(
    OUTPUT_DIR /
    "USA_Today_cleaning_process_summary.xlsx",
    index=False
)


# %%
# ============================================================
# 28. Overall descriptive statistics
# ============================================================

overall_statistics = pd.DataFrame([
    {
        "metric": "Total final articles",
        "value": len(data)
    },
    {
        "metric":
            "Articles included after manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & (
                        data[
                            "manual_review_decision"
                        ]
                        == "INCLUDE"
                    )
                ).sum()
            )
    },
    {
        "metric":
            "Articles still awaiting manual scope review",
        "value":
            int(
                (
                    data[
                        "requires_manual_scope_review"
                    ]
                    & data[
                        "manual_review_decision"
                    ].isna()
                ).sum()
            )
    },
    {
        "metric": "Unique titles",
        "value": data[TITLE_COL].nunique()
    },
    {
        "metric": "Repeated titles remaining",
        "value": int(
            data[TITLE_COL]
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Unique content hashes",
        "value": data[
            "content_sha256"
        ].nunique(
            dropna=True
        )
    },
    {
        "metric": "Duplicate content hashes remaining",
        "value": int(
            data[
                "content_sha256"
            ]
            .dropna()
            .duplicated()
            .sum()
        )
    },
    {
        "metric": "Missing dates",
        "value": int(
            data[
                "date_fixed"
            ].isna().sum()
        )
    },
    {
        "metric": "Mean full-text word count",
        "value": data[
            "full_text_word_count"
        ].mean()
    },
    {
        "metric": "Median full-text word count",
        "value": data[
            "full_text_word_count"
        ].median()
    },
    {
        "metric": "Minimum full-text word count",
        "value": data[
            "full_text_word_count"
        ].min()
    },
    {
        "metric": "Maximum full-text word count",
        "value": data[
            "full_text_word_count"
        ].max()
    },
    {
        "metric": "Mean first-paragraph word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].mean()
    },
    {
        "metric": "Median first-paragraph word count",
        "value": valid_leads[
            "first_paragraph_word_count"
        ].median()
    }
])


overall_statistics["value"] = (
    overall_statistics["value"]
    .apply(
        lambda value: round(value, 2)
        if isinstance(
            value,
            (
                int,
                float,
                np.integer,
                np.floating
            )
        )
        and not pd.isna(value)
        else value
    )
)


display(overall_statistics)

overall_statistics.to_excel(
    OUTPUT_DIR /
    "USA_Today_overall_descriptive_statistics.xlsx",
    index=False
)


# %%
# ============================================================
# 29. Article counts by year
# ============================================================

year_counts = (
    data.dropna(
        subset=["year"]
    )
    .groupby("year")
    .size()
    .reset_index(name="article_count")
    .sort_values("year")
)


display(year_counts)

year_counts.to_excel(
    OUTPUT_DIR /
    "USA_Today_final_article_counts_by_year.xlsx",
    index=False
)


if len(year_counts) > 0:

    plt.figure(figsize=(10, 6))

    plt.bar(
        year_counts["year"]
        .astype(int)
        .astype(str),

        year_counts["article_count"]
    )

    plt.xlabel("Year")
    plt.ylabel("Number of Articles")

    plt.title(
        "Number of USA Today Autonomous-Vehicle Articles by Year"
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "USA_Today_final_article_counts_by_year.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 30. Complete monthly article timeline
# ============================================================

valid_date_data = data[
    data["date_fixed"].notna()
].copy()


if len(valid_date_data) > 0:

    year_month_counts = (
        valid_date_data
        .groupby("year_month")
        .size()
        .reset_index(name="article_count")
        .sort_values("year_month")
    )


    full_month_range = pd.period_range(
        start=valid_date_data[
            "date_fixed"
        ].min().to_period("M"),

        end=valid_date_data[
            "date_fixed"
        ].max().to_period("M"),

        freq="M"
    )


    complete_month_table = pd.DataFrame({
        "year_month":
            full_month_range.astype(str)
    })


    complete_month_table = (
        complete_month_table
        .merge(
            year_month_counts,
            on="year_month",
            how="left"
        )
    )


    complete_month_table["article_count"] = (
        complete_month_table[
            "article_count"
        ]
        .fillna(0)
        .astype(int)
    )


    complete_month_table.to_excel(
        OUTPUT_DIR /
        "USA_Today_final_complete_monthly_counts.xlsx",
        index=False
    )


    plt.figure(figsize=(16, 6))

    plt.plot(
        complete_month_table["year_month"],
        complete_month_table["article_count"],
        marker="o",
        markersize=3
    )

    label_interval = max(
        1,
        len(complete_month_table) // 24
    )

    tick_positions = list(
        range(
            0,
            len(complete_month_table),
            label_interval
        )
    )

    plt.xticks(
        ticks=tick_positions,
        labels=complete_month_table.iloc[
            tick_positions
        ]["year_month"],
        rotation=90
    )

    plt.xlabel("Year-Month")
    plt.ylabel("Number of Articles")

    plt.title(
        "Monthly Number of USA Today "
        "Autonomous-Vehicle Articles"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "USA_Today_final_monthly_article_trend.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

else:

    print(
        "\nWARNING: No valid dates were available for "
        "the monthly timeline."
    )


# %%
# ============================================================
# 31. Word-count distributions
# ============================================================

if len(data) > 0:

    plt.figure(figsize=(10, 6))

    plt.hist(
        data[
            "full_text_word_count"
        ].dropna(),
        bins=30,
        edgecolor="black"
    )

    plt.xlabel("Full-Article Word Count")
    plt.ylabel("Number of Articles")

    plt.title(
        "Distribution of USA Today Full-Article Lengths"
    )

    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "USA_Today_final_full_text_word_distribution.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


if len(valid_leads) > 0:

    plt.figure(figsize=(10, 6))

    plt.hist(
        valid_leads[
            "first_paragraph_word_count"
        ],
        bins=30,
        edgecolor="black"
    )

    lead_mean = valid_leads[
        "first_paragraph_word_count"
    ].mean()

    lead_median = valid_leads[
        "first_paragraph_word_count"
    ].median()

    plt.axvline(
        lead_mean,
        linestyle="--",
        linewidth=2,
        label=f"Mean = {lead_mean:.1f}"
    )

    plt.axvline(
        lead_median,
        linestyle=":",
        linewidth=2,
        label=f"Median = {lead_median:.1f}"
    )

    plt.xlabel("First-Paragraph Word Count")
    plt.ylabel("Number of Articles")

    plt.title(
        "Distribution of USA Today First-Paragraph Lengths"
    )

    plt.legend()
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR /
        "USA_Today_final_first_paragraph_distribution.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# %%
# ============================================================
# 32. Create combined extraction-check file
# ============================================================

extraction_check = pd.concat(
    [
        data.head(100),

        skipped_metadata_cases,

        data[
            data["english_dateline_removed"]
        ],

        remaining_prefix_problems,

        short_first_paragraph,

        long_first_paragraph,

        first_equals_full,

        failed_first_paragraph
    ],
    ignore_index=True
)


# Deduplicate check rows
if "content_sha256" in extraction_check.columns:

    non_empty_hash = (
        extraction_check[
            "content_sha256"
        ]
        .fillna("")
        .ne("")
    )

    extraction_check_with_hash = (
        extraction_check[
            non_empty_hash
        ]
        .drop_duplicates(
            subset=["content_sha256"],
            keep="first"
        )
    )

    extraction_check_without_hash = (
        extraction_check[
            ~non_empty_hash
        ]
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )

    extraction_check = pd.concat(
        [
            extraction_check_with_hash,
            extraction_check_without_hash
        ],
        ignore_index=True
    )

else:

    extraction_check = (
        extraction_check
        .drop_duplicates(
            subset=[TITLE_COL],
            keep="first"
        )
    )


check_columns = [
    "date_fixed",
    TITLE_COL,

    "skipped_metadata_count",
    "skipped_metadata_text",

    "first_paragraph_before_dateline_removal",
    "first_paragraph",

    "english_dateline_removed",
    "extraction_candidate_position",

    "first_paragraph_word_count",
    "first_paragraph_char_count",
    "first_paragraph_extracted",
    "first_paragraph_equals_full_text",

    "first_paragraph_starts_with_byline",
    "first_paragraph_starts_with_usa_today",
    "first_paragraph_has_possible_dateline",
    "first_paragraph_prefix_problem",

    "article_body"
]


check_columns = [
    column
    for column in check_columns
    if column in extraction_check.columns
]


extraction_check[
    check_columns
].to_excel(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_extraction_check.xlsx",
    index=False
)


# %%
# ============================================================
# 33. Remove internal helper columns from final output
# ============================================================

temporary_columns_to_remove = [
    "_original_order",
    "_content_dedup_key"
]

data = data.drop(
    columns=[
        column
        for column in temporary_columns_to_remove
        if column in data.columns
    ]
)


# %%
# ============================================================
# 34. Save final dataset with first paragraph
# ============================================================

data.to_excel(
    FINAL_WITH_LEAD_PATH,
    index=False
)


# %%
# ============================================================
# 35. Final verification
# ============================================================

print("\n" + "=" * 75)
print("USA TODAY CLEANING COMPLETED")
print("=" * 75)

print("\nOriginal articles:")
print(n_original)

print(
    "\nRemoved because of invalid dates "
    "or failed body extraction:"
)
print(n_removed_initial_cleaning)

print("\nRemoved by automatic AV scope screening:")
print(n_removed_automatic_scope)

print("\nSent for manual scope review:")
print(n_before_manual_review)

print("\nIncluded after manual scope review:")
print(n_manual_review_included)

print("\nExcluded after manual scope review:")
print(n_manual_review_excluded)

print(
    "\nRetained after automatic and manual "
    "scope screening, before deduplication:"
)
print(n_after_scope_screening)

print("\nRemoved by full-content SHA-256 deduplication:")
print(n_removed_content_duplicates)

print("\nFinal analytical articles:")
print(len(data))

final_manually_included_count = int(
    (
        data["requires_manual_scope_review"]
        & (
            data["manual_review_decision"]
            == "INCLUDE"
        )
    ).sum()
)

final_pending_manual_review_count = int(
    (
        data["requires_manual_scope_review"]
        & data[
            "manual_review_decision"
        ].isna()
    ).sum()
)


print("\nFinal articles included after manual scope review:")
print(final_manually_included_count)

print("\nFinal articles still awaiting manual review:")
print(final_pending_manual_review_count)

print("\nMissing dates in final dataset:")
print(
    data["date_fixed"].isna().sum()
)

print("\nDuplicate content hashes remaining:")
print(
    data[
        "content_sha256"
    ]
    .dropna()
    .duplicated()
    .sum()
)

print("\nSuccessfully extracted first paragraphs:")
print(
    int(
        data[
            "first_paragraph_extracted"
        ].sum()
    )
)

print("\nFailed first-paragraph extractions:")
print(len(failed_first_paragraph))

print("\nFirst paragraphs under 15 words:")
print(len(short_first_paragraph))

print("\nFirst paragraphs over 200 words:")
print(len(long_first_paragraph))

print("\nRemaining possible lead-prefix problems:")
print(len(remaining_prefix_problems))

print("\nMean full-text word count:")
print(
    round(
        data[
            "full_text_word_count"
        ].mean(),
        2
    )
)

print("\nMedian full-text word count:")
print(
    round(
        data[
            "full_text_word_count"
        ].median(),
        2
    )
)

print("\nMean first-paragraph word count:")
print(
    round(
        valid_leads[
            "first_paragraph_word_count"
        ].mean(),
        2
    )
)

print("\nMedian first-paragraph word count:")
print(
    round(
        valid_leads[
            "first_paragraph_word_count"
        ].median(),
        2
    )
)

print("\nFinal analytical dataset saved to:")
print(FINAL_DATA_PATH)

print("\nFinal dataset with first paragraph saved to:")
print(FINAL_WITH_LEAD_PATH)

print("\nManual scope-review file saved to:")
print(
    OUTPUT_DIR /
    "USA_Today_articles_requiring_manual_scope_review.xlsx"
)

print("\nAutomatically excluded scope file saved to:")
print(
    OUTPUT_DIR /
    "USA_Today_articles_excluded_by_AV_scope.xlsx"
)

print("\nAll audit and descriptive files saved in:")
print(OUTPUT_DIR)


print(
    "\nSuccessfully extracted "
    "first AV-relevant paragraphs:"
)

print(
    int(
        data[
            "first_av_relevant_paragraph_found"
        ].sum()
    )
)


print(
    "\nFailed first AV-relevant paragraph extractions:"
)

print(
    len(
        failed_av_paragraphs
    )
)


print(
    "\nOriginal first paragraph already AV-relevant:"
)

print(
    n_original_first_is_av
)


print(
    "\nPercentage original first paragraph already AV-relevant:"
)

print(
    round(
        pct_original_first_is_av,
        2
    ),
    "%"
)


print(
    "\nMean first AV-relevant paragraph word count:"
)

print(
    round(
        data.loc[
            data[
                "first_av_relevant_paragraph_found"
            ],
            "first_av_relevant_paragraph_word_count"
        ].mean(),
        2
    )
)


print(
    "\nMedian first AV-relevant paragraph word count:"
)

print(
    round(
        data.loc[
            data[
                "first_av_relevant_paragraph_found"
            ],
            "first_av_relevant_paragraph_word_count"
        ].median(),
        2
    )
)


print(
    "\nFinal dataset with AV-relevant paragraph saved to:"
)

print(
    FINAL_WITH_AV_PARAGRAPH_PATH
)
```

```python
# %%
# ============================================================
# Create a 10% Prompt-Validation Sample for USA Today
#
# Purpose:
# 1. Load the final cleaned USA Today dataset
# 2. Retain only usable records
# 3. Randomly sample 10% of the dataset
# 4. Use first_paragraph as the prompt input text
# 5. Add blank human-coding and LLM-output columns
# 6. Save the validation sample as Excel
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
import math


# %%
# ============================================================
# 1. File paths
# ============================================================

INPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_USA_Today/"
    "USA_Today_AV_final_with_first_av_relevant_paragraph.xlsx"
)

OUTPUT_PATH = Path(
    "/Users/yurujia/Desktop/Dissertation Data/USA/"
    "descriptive_stats_USA_Today/"
    "USA_Today_10_percent_prompt_validation_sample.xlsx"
)

OUTPUT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# %%
# ============================================================
# 2. Sampling settings
# ============================================================

SAMPLE_FRACTION = 0.10

# Fixed seed ensures that the same records are selected
# whenever the code is rerun.
RANDOM_SEED = 42

TEXT_COLUMN = "sentiment_analysis_text"


# %%
# ============================================================
# 3. Helper function for Boolean conversion
# ============================================================

def convert_to_boolean(series):
    """
    Convert common True/False representations into Boolean values.
    """

    true_values = {
        "true",
        "1",
        "yes",
        "y",
        "include",
        "included"
    }

    return (
        series
        .astype(str)
        .str.strip()
        .str.lower()
        .isin(true_values)
    )


# %%
# ============================================================
# 4. Load data
# ============================================================

if not INPUT_PATH.exists():
    raise FileNotFoundError(
        "USA Today input file was not found:\n"
        f"{INPUT_PATH}\n\n"
        "Please update INPUT_PATH."
    )

df = pd.read_excel(INPUT_PATH)

print("USA Today data loaded successfully.")
print("Original shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# %%
# ============================================================
# 5. Retain manually included records if necessary
# ============================================================

df_final = df.copy()

if "manual_review_in_scope" in df_final.columns:

    reviewed_mask = df_final[
        "manual_review_in_scope"
    ].notna()

    if reviewed_mask.any():

        manual_include_mask = convert_to_boolean(
            df_final["manual_review_in_scope"]
        )

        n_before_manual_filter = len(df_final)

        df_final = df_final.loc[
            (~reviewed_mask) |
            manual_include_mask
        ].copy()

        print(
            "\nRows removed because manual review "
            "marked them out of scope:",
            n_before_manual_filter - len(df_final)
        )


elif "manual_include_in_final_dataset" in df_final.columns:

    reviewed_mask = df_final[
        "manual_include_in_final_dataset"
    ].notna()

    if reviewed_mask.any():

        manual_include_mask = convert_to_boolean(
            df_final["manual_include_in_final_dataset"]
        )

        n_before_manual_filter = len(df_final)

        df_final = df_final.loc[
            (~reviewed_mask) |
            manual_include_mask
        ].copy()

        print(
            "\nRows removed because manual review "
            "marked them out of scope:",
            n_before_manual_filter - len(df_final)
        )


print("\nRows available after manual-review filtering:", len(df_final))


# %%
# ============================================================
# 6. Check prompt text column
# ============================================================

if TEXT_COLUMN not in df_final.columns:
    raise ValueError(
        f"Required column '{TEXT_COLUMN}' was not found.\n"
        f"Available columns: {df_final.columns.tolist()}"
    )


# Remove rows where the AV-specific sentiment-analysis text
# is missing or empty.

usable_text_mask = (
    df_final[TEXT_COLUMN]
    .fillna("")
    .astype(str)
    .str.strip()
    .ne("")
)

n_before_text_filter = len(df_final)

df_usable = df_final.loc[
    usable_text_mask
].copy()

print(
    "\nRows removed because AV-specific "
    "sentiment-analysis text is missing or empty:",
    n_before_text_filter - len(df_usable)
)

print(
    "Rows with usable AV-specific sentiment text:",
    len(df_usable)
)


if len(df_usable) == 0:
    raise ValueError(
        "No usable USA Today records remain for sampling."
    )


# %%
# ============================================================
# 7. Calculate the 10% sample size
# ============================================================

sample_size = math.ceil(
    len(df_usable) * SAMPLE_FRACTION
)

sample_size = max(1, sample_size)

print("\nSampling fraction:", SAMPLE_FRACTION)
print("Population used for sampling:", len(df_usable))
print("Sample size:", sample_size)


# %%
# ============================================================
# 8. Draw a reproducible random sample
# ============================================================

sample_df = df_usable.sample(
    n=sample_size,
    random_state=RANDOM_SEED,
    replace=False
).copy()


# %%
# ============================================================
# 9. Sort sample for easier manual review
# ============================================================

if "date_fixed" in sample_df.columns:

    sample_df["date_fixed"] = pd.to_datetime(
        sample_df["date_fixed"],
        errors="coerce"
    )

    sample_df = sample_df.sort_values(
        by="date_fixed",
        ascending=True,
        na_position="last"
    )

elif "date" in sample_df.columns:

    sample_df["date"] = pd.to_datetime(
        sample_df["date"],
        errors="coerce"
    )

    sample_df = sample_df.sort_values(
        by="date",
        ascending=True,
        na_position="last"
    )


sample_df = sample_df.reset_index(drop=True)


# %%
# ============================================================
# 10. Add validation ID and prompt input text
# ============================================================

sample_df.insert(
    0,
    "prompt_validation_id",
    [
        f"USA_{i:03d}"
        for i in range(1, len(sample_df) + 1)
    ]
)

sample_df.insert(
    1,
    "news_source",
    "USA Today"
)

sample_df.insert(
    2,
    "prompt_text_column",
    TEXT_COLUMN
)

sample_df.insert(
    3,
    "prompt_input_text",
    sample_df[TEXT_COLUMN]
)

sample_df[
    "original_first_paragraph"
] = (
    sample_df[
        "first_paragraph"
    ]
)


sample_df[
    "av_relevant_paragraph"
] = (
    sample_df[
        "first_av_relevant_paragraph"
    ]
)

# %%
# ============================================================
# 11. Add blank human annotation columns
# ============================================================

sample_df["human_sentiment_label"] = ""
sample_df["human_sentiment_score"] = np.nan

sample_df["human_primary_frame"] = ""
sample_df["human_secondary_frame"] = ""

sample_df["human_issue_category"] = ""

sample_df["human_annotation_notes"] = ""


# %%
# ============================================================
# 12. Add blank LLM output columns
# ============================================================

sample_df["llm_sentiment_label"] = ""
sample_df["llm_sentiment_score"] = np.nan

sample_df["llm_primary_frame"] = ""
sample_df["llm_secondary_frame"] = ""

sample_df["llm_issue_category"] = ""

sample_df["llm_raw_response"] = ""


# %%
# ============================================================
# 13. Add blank accuracy comparison columns
# ============================================================

sample_df["sentiment_label_match"] = ""
sample_df["primary_frame_match"] = ""
sample_df["issue_category_match"] = ""

sample_df["review_completed"] = False


# %%
# ============================================================
# 14. Preview sample
# ============================================================

preview_columns = [
    "prompt_validation_id",
    "news_source",
    "title",
    "prompt_input_text"
]

if "date_fixed" in sample_df.columns:
    preview_columns.insert(2, "date_fixed")

elif "date" in sample_df.columns:
    preview_columns.insert(2, "date")


print("\nUSA Today validation sample preview:")

display(
    sample_df[
        preview_columns
    ].head(10)
)


# %%
# ============================================================
# 15. Save validation sample
# ============================================================

sample_df.to_excel(
    OUTPUT_PATH,
    index=False
)

print("\n" + "=" * 70)
print("USA TODAY PROMPT VALIDATION SAMPLE CREATED")
print("=" * 70)

print("Final cleaned population:", len(df_usable))
print("10% validation sample:", len(sample_df))
print("Random seed:", RANDOM_SEED)

print("\nSaved to:")
print(OUTPUT_PATH)
```

```python
# %%
# ============================================================
# Extract full article body and first paragraph
# USA Today LexisNexis dataset
#
# Full text:
# Text between "Body" and "End of Document"
#
# First paragraph:
# The first non-empty line after "Body"
# ============================================================

import pandas as pd
import numpy as np
import re
from pathlib import Path


# ------------------------------------------------------------
# 1. Basic settings
# ------------------------------------------------------------

CONTENT_COL = "content"

if CONTENT_COL not in df.columns:
    raise KeyError(
        f"Column '{CONTENT_COL}' was not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

print("=" * 70)
print("USA TODAY TEXT EXTRACTION")
print("=" * 70)

print("Total articles:", len(df))
print("Content column:", CONTENT_COL)


# ------------------------------------------------------------
# 2. Normalize line breaks
# ------------------------------------------------------------

def normalize_line_breaks(text):
    """
    Standardize line breaks while preserving paragraph boundaries.
    """

    if pd.isna(text):
        return pd.NA

    text = str(text)

    # Standardize different line-break formats
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove spaces and tabs around line breaks
    text = re.sub(r"[ \t]+\n", "\n", text)
    text = re.sub(r"\n[ \t]+", "\n", text)

    # Reduce excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


# ------------------------------------------------------------
# 3. Extract full article body
# ------------------------------------------------------------

def extract_full_body(text):
    """
    Extract text between the Body marker and End of Document marker.

    Expected structure:

    Body
    Article text...
    End of Document
    """

    text = normalize_line_breaks(text)

    if pd.isna(text):
        return pd.NA

    # Preferred pattern:
    # Body and End of Document appear as separate lines
    main_pattern = re.compile(
        r"(?ims)"
        r"^[ \t]*Body[ \t]*\n"
        r"(.*?)"
        r"(?=^[ \t]*End\s+of\s+Document[ \t]*$)"
    )

    match = main_pattern.search(text)

    # Fallback for records with unusual line formatting
    if match is None:
        fallback_pattern = re.compile(
            r"(?is)"
            r"\bBody\b\s*"
            r"(.*?)"
            r"\s*\bEnd\s+of\s+Document\b"
        )

        match = fallback_pattern.search(text)

    if match is None:
        return pd.NA

    full_body = match.group(1).strip()

    # Clean spaces around line breaks
    full_body = re.sub(r"[ \t]+\n", "\n", full_body)
    full_body = re.sub(r"\n[ \t]+", "\n", full_body)

    # Reduce excessive blank lines but preserve normal line structure
    full_body = re.sub(r"\n{3,}", "\n\n", full_body)

    if full_body == "":
        return pd.NA

    return full_body


# Create full-text column
df["全文"] = df[CONTENT_COL].apply(extract_full_body)

df["全文提取成功"] = df["全文"].notna()


# ------------------------------------------------------------
# 4. Full-text extraction summary
# ------------------------------------------------------------

full_text_success = df["全文"].notna().sum()
full_text_failure = df["全文"].isna().sum()

print("\nFull-text extraction summary:")
print("Successfully extracted:", full_text_success)
print("Failed extraction:", full_text_failure)

if len(df) > 0:
    print(
        "Success rate:",
        f"{full_text_success / len(df) * 100:.2f}%"
    )


# ------------------------------------------------------------
# 5. Save and inspect failed full-text extraction
# ------------------------------------------------------------

failed_full_text_rows = df[
    df["全文"].isna()
].copy()

if len(failed_full_text_rows) > 0:

    failed_columns = [
        col for col in [
            "file",
            "title",
            "source",
            "date",
            "content"
        ]
        if col in failed_full_text_rows.columns
    ]

    print("\nFailed full-text extraction examples:")

    display(
        failed_full_text_rows[
            failed_columns
        ].head(20)
    )

    failed_full_text_rows.to_excel(
        OUTPUT_DIR /
        "USA_Today_failed_full_text_extraction.xlsx",
        index=False
    )


# ------------------------------------------------------------
# 6. Extract first paragraph
# ------------------------------------------------------------

def extract_first_paragraph_usa_today(full_text):
    """
    Extract the first paragraph from USA Today LexisNexis articles.

    In this dataset, paragraphs are generally separated by a single
    line break rather than an empty line.

    Therefore, the first paragraph is defined as the first non-empty
    line in the extracted full article body.
    """

    if pd.isna(full_text):
        return pd.NA

    text = normalize_line_breaks(full_text)

    if pd.isna(text):
        return pd.NA

    # Split at every line break
    lines = text.split("\n")

    cleaned_lines = []

    for line in lines:

        line = line.strip()

        # Replace repeated internal spaces with one space
        line = re.sub(r"\s+", " ", line)

        if line != "":
            cleaned_lines.append(line)

    if len(cleaned_lines) == 0:
        return pd.NA

    # The first non-empty line is treated as the first paragraph
    first_paragraph = cleaned_lines[0]

    return first_paragraph


# Create first-paragraph column
df["首段"] = df["全文"].apply(
    extract_first_paragraph_usa_today
)

df["首段提取成功"] = df["首段"].notna()


# ------------------------------------------------------------
# 7. First-paragraph extraction summary
# ------------------------------------------------------------

first_paragraph_success = df["首段"].notna().sum()
first_paragraph_failure = df["首段"].isna().sum()

print("\nFirst-paragraph extraction summary:")
print("Successfully extracted:", first_paragraph_success)
print("Failed extraction:", first_paragraph_failure)

if len(df) > 0:
    print(
        "Success rate:",
        f"{first_paragraph_success / len(df) * 100:.2f}%"
    )


# ------------------------------------------------------------
# 8. Display extraction examples
# ------------------------------------------------------------

print("\nFull text and first-paragraph examples:")

display(
    df.loc[
        df["全文"].notna(),
        [
            "title",
            "全文",
            "首段"
        ]
    ].head(10)
)


# ------------------------------------------------------------
# 9. English word-count function
# ------------------------------------------------------------

def count_english_words(text):
    """
    Count English word tokens.

    Examples treated as one word:
    self-driving
    company's
    COVID-19
    U.S.
    """

    if pd.isna(text):
        return np.nan

    text = str(text).strip()

    if text == "":
        return np.nan

    words = re.findall(
        r"\b[A-Za-z0-9]+(?:['’.-][A-Za-z0-9]+)*\b",
        text
    )

    return len(words)


# ------------------------------------------------------------
# 10. Calculate full-text and first-paragraph word counts
# ------------------------------------------------------------

df["全文_word_count"] = df["全文"].apply(
    count_english_words
)

df["首段_word_count"] = df["首段"].apply(
    count_english_words
)

# Use nullable integer type
df["全文_word_count"] = (
    df["全文_word_count"]
    .round()
    .astype("Int64")
)

df["首段_word_count"] = (
    df["首段_word_count"]
    .round()
    .astype("Int64")
)


# ------------------------------------------------------------
# 11. Preview word-count results
# ------------------------------------------------------------

print("\nWord-count preview:")

display(
    df[
        [
            "title",
            "全文_word_count",
            "首段",
            "首段_word_count"
        ]
    ].head(20)
)


# ------------------------------------------------------------
# 12. Check whether first paragraph incorrectly equals full text
# ------------------------------------------------------------

df["首段等于全文"] = (
    df["首段"].fillna("").str.strip()
    ==
    df["全文"].fillna("").str.strip()
)

same_text_count = df["首段等于全文"].sum()

print("\nVerification:")
print(
    "Articles where first paragraph equals full text:",
    same_text_count
)

# Only inspect cases where both fields are non-missing
same_non_missing_rows = df[
    df["首段等于全文"]
    & df["全文"].notna()
    & df["首段"].notna()
].copy()

if len(same_non_missing_rows) > 0:

    print(
        "\nArticles where the extracted first paragraph "
        "is identical to the full text:"
    )

    display(
        same_non_missing_rows[
            [
                "title",
                "全文",
                "首段",
                "全文_word_count",
                "首段_word_count"
            ]
        ].head(20)
    )


# ------------------------------------------------------------
# 13. Check implausibly long first paragraphs
# ------------------------------------------------------------

long_first_paragraph_rows = df[
    df["首段_word_count"] > 200
].copy()

print(
    "\nFirst paragraphs longer than 200 words:",
    len(long_first_paragraph_rows)
)

if len(long_first_paragraph_rows) > 0:

    print(
        "\nLong first-paragraph examples requiring inspection:"
    )

    display(
        long_first_paragraph_rows[
            [
                "title",
                "首段",
                "首段_word_count",
                "全文_word_count"
            ]
        ].head(20)
    )

    long_first_paragraph_rows.to_excel(
        OUTPUT_DIR /
        "USA_Today_first_paragraphs_over_200_words.xlsx",
        index=False
    )


# ------------------------------------------------------------
# 14. Descriptive-statistics function
# ------------------------------------------------------------

def create_word_count_statistics(series, text_type):
    """
    Generate descriptive statistics for a word-count column.
    """

    valid_values = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    if len(valid_values) == 0:

        return {
            "text_type": text_type,
            "total_articles": len(series),
            "valid_articles": 0,
            "missing_articles": len(series),
            "mean": np.nan,
            "median": np.nan,
            "standard_deviation": np.nan,
            "minimum": np.nan,
            "q1_25_percent": np.nan,
            "q3_75_percent": np.nan,
            "maximum": np.nan
        }

    return {
        "text_type": text_type,
        "total_articles": len(series),
        "valid_articles": len(valid_values),
        "missing_articles": series.isna().sum(),
        "mean": valid_values.mean(),
        "median": valid_values.median(),
        "standard_deviation": valid_values.std(),
        "minimum": valid_values.min(),
        "q1_25_percent": valid_values.quantile(0.25),
        "q3_75_percent": valid_values.quantile(0.75),
        "maximum": valid_values.max()
    }


# ------------------------------------------------------------
# 15. Generate descriptive-statistics table
# ------------------------------------------------------------

word_count_statistics = pd.DataFrame([
    create_word_count_statistics(
        df["全文_word_count"],
        "Full article body"
    ),
    create_word_count_statistics(
        df["首段_word_count"],
        "First paragraph"
    )
])

statistics_columns_to_round = [
    "mean",
    "median",
    "standard_deviation",
    "minimum",
    "q1_25_percent",
    "q3_75_percent",
    "maximum"
]

word_count_statistics[
    statistics_columns_to_round
] = word_count_statistics[
    statistics_columns_to_round
].round(2)

print("\nWord-count descriptive statistics:")

display(word_count_statistics)


# ------------------------------------------------------------
# 16. Detailed percentiles
# ------------------------------------------------------------

def create_detailed_percentiles(series, text_type):
    """
    Calculate selected percentiles.
    """

    valid_values = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    percentile_levels = [
        0.00,
        0.05,
        0.10,
        0.25,
        0.50,
        0.75,
        0.90,
        0.95,
        1.00
    ]

    if len(valid_values) == 0:

        return pd.DataFrame({
            "text_type": [text_type] * len(percentile_levels),
            "percentile": [
                int(level * 100)
                for level in percentile_levels
            ],
            "word_count": [np.nan] * len(percentile_levels)
        })

    result = (
        valid_values
        .quantile(percentile_levels)
        .reset_index()
    )

    result.columns = [
        "percentile",
        "word_count"
    ]

    result["text_type"] = text_type

    result["percentile"] = (
        result["percentile"] * 100
    ).astype(int)

    result["word_count"] = (
        result["word_count"]
        .round(2)
    )

    return result[
        [
            "text_type",
            "percentile",
            "word_count"
        ]
    ]


detailed_percentiles = pd.concat(
    [
        create_detailed_percentiles(
            df["全文_word_count"],
            "Full article body"
        ),
        create_detailed_percentiles(
            df["首段_word_count"],
            "First paragraph"
        )
    ],
    ignore_index=True
)

print("\nDetailed word-count percentiles:")

display(detailed_percentiles)


# ------------------------------------------------------------
# 17. Optional distribution plots
# ------------------------------------------------------------

# Full article word-count distribution
plt.figure(figsize=(10, 6))

plt.hist(
    df["全文_word_count"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.xlabel("Word Count")
plt.ylabel("Number of Articles")
plt.title("Distribution of USA Today Full-Article Word Counts")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR /
    "USA_Today_full_article_word_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# First-paragraph word-count distribution
plt.figure(figsize=(10, 6))

plt.hist(
    df["首段_word_count"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.xlabel("Word Count")
plt.ylabel("Number of Articles")
plt.title("Distribution of USA Today First-Paragraph Word Counts")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR /
    "USA_Today_first_paragraph_word_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 18. Prepare article-level output
# ------------------------------------------------------------

article_level_output_columns = [
    col for col in [
        "file",
        "title",
        "source",
        "date",
        "date_raw",
        "date_clean",
        "date_parsed",
        "year",
        "month",
        "month_name",
        "year_month",
        "content",
        "全文",
        "全文_word_count",
        "首段",
        "首段_word_count",
        "全文提取成功",
        "首段提取成功",
        "首段等于全文"
    ]
    if col in df.columns
]

article_text_output = df[
    article_level_output_columns
].copy()


# ------------------------------------------------------------
# 19. Save separate Excel files
# ------------------------------------------------------------

article_text_output.to_excel(
    OUTPUT_DIR /
    "USA_Today_full_text_and_first_paragraph.xlsx",
    index=False
)

word_count_statistics.to_excel(
    OUTPUT_DIR /
    "USA_Today_word_count_descriptive_statistics.xlsx",
    index=False
)

detailed_percentiles.to_excel(
    OUTPUT_DIR /
    "USA_Today_word_count_percentiles.xlsx",
    index=False
)


# ------------------------------------------------------------
# 20. Save combined Excel workbook
# ------------------------------------------------------------

combined_output_path = (
    OUTPUT_DIR /
    "USA_Today_text_extraction_and_word_count_results.xlsx"
)

with pd.ExcelWriter(
    combined_output_path,
    engine="openpyxl"
) as writer:

    article_text_output.to_excel(
        writer,
        sheet_name="Article-level data",
        index=False
    )

    word_count_statistics.to_excel(
        writer,
        sheet_name="Descriptive statistics",
        index=False
    )

    detailed_percentiles.to_excel(
        writer,
        sheet_name="Percentiles",
        index=False
    )

    if len(failed_full_text_rows) > 0:

        failed_full_text_rows.to_excel(
            writer,
            sheet_name="Failed full text",
            index=False
        )

    if len(same_non_missing_rows) > 0:

        same_non_missing_rows.to_excel(
            writer,
            sheet_name="First equals full",
            index=False
        )

    if len(long_first_paragraph_rows) > 0:

        long_first_paragraph_rows.to_excel(
            writer,
            sheet_name="Long first paragraphs",
            index=False
        )


# ------------------------------------------------------------
# 21. Save updated complete dataset
# ------------------------------------------------------------

updated_dataset_path = (
    OUTPUT_DIR /
    "USA_Today_dataset_with_text_variables.xlsx"
)

df.to_excel(
    updated_dataset_path,
    index=False
)


# ------------------------------------------------------------
# 22. Final verification
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL TEXT EXTRACTION CHECK")
print("=" * 70)

print("Total articles:", len(df))

print("\nFull article body:")
print(
    "Successfully extracted:",
    df["全文"].notna().sum()
)
print(
    "Missing:",
    df["全文"].isna().sum()
)
print(
    "Mean word count:",
    round(df["全文_word_count"].mean(), 2)
)
print(
    "Median word count:",
    round(df["全文_word_count"].median(), 2)
)

print("\nFirst paragraph:")
print(
    "Successfully extracted:",
    df["首段"].notna().sum()
)
print(
    "Missing:",
    df["首段"].isna().sum()
)
print(
    "Mean word count:",
    round(df["首段_word_count"].mean(), 2)
)
print(
    "Median word count:",
    round(df["首段_word_count"].median(), 2)
)

print(
    "\nFirst paragraph equals full text:",
    len(same_non_missing_rows)
)

print(
    "First paragraphs longer than 200 words:",
    len(long_first_paragraph_rows)
)

print("\nCombined results saved to:")
print(combined_output_path)

print("\nUpdated complete dataset saved to:")
print(updated_dataset_path)

print("\nAll output files saved in:")
print(OUTPUT_DIR)
```
